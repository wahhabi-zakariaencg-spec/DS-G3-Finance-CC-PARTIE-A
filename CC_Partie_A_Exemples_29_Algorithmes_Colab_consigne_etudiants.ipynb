## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression linéaire** (*Linear Regression*) — régression supervisée : prédire une valeur continue par une combinaison linéaire des variables.
**Étapes :** données $(X,y)$ → coefficients qui minimisent la somme des carrés des erreurs → prédiction $\hat y$.
$$\hat y_i=\beta_0+\textstyle\sum_{j=1}^{p}\beta_j x_{ij},\qquad J=\tfrac{1}{2n}\sum_i(y_i-\hat y_i)^2$$
$n$ : observations ; $p$ : variables ; $x_{ij}$ : variable $j$ de l'observation $i$ ; $y_i$ : valeur réelle ; $J$ : coût quadratique.
**Paramètres appris :** $\beta_0,\beta_j$. **Hyperparamètre :** `fit_intercept`.
**Avantages :** rapide (solution exacte) ; coefficients interprétables.
**Limites :** relation supposée linéaire ; sensible aux valeurs atypiques et à la multicolinéarité.

### Schéma explicatif

**Comment lire le schéma :** chaque point bleu est une observation ; la droite noire est celle qui rend **minimale la somme des carrés des segments rouges** (résidus). La pente estimée indique de combien $\hat y$ varie quand $x$ augmente d'une unité. Les données sont simulées pour l'illustration.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/gestion) :** prévoir le chiffre d'affaires mensuel d'une enseigne à partir des dépenses marketing, du nombre de points de vente et de la saisonnalité.
- **Cas d'usage 2 :** estimer le prix d'un logement selon la surface, le quartier et l'ancienneté.
- **Données et prérequis :** tableau numérique (variables catégorielles encodées en indicatrices), cible continue ; relation à peu près linéaire, résidus sans structure, faible multicolinéarité, pas de fuite d'information.
- **Peu approprié :** prévoir le défaut de paiement (oui/non) — la cible est binaire, une droite donne des « probabilités » hors de [0 ; 1] → préférer la régression logistique. Également peu adapté aux relations fortement non linéaires (seuils, saturation).
- **Métriques :** MAE, RMSE, R² (sur un jeu de test indépendant) ; analyse des résidus.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** `Xr_ent` (210 × 5) et `yr_ent` issues de `make_regression` (280 lignes, 5 variables, bruit `noise=13`), découpage 75/25 avec `random_state=42`. Données **synthétiques**.
- **Hyperparamètres :** valeurs par défaut (`fit_intercept=True`).
- **Sortie obtenue :** **MAE ≈ 9,35** sur les 70 observations de test.
- **Ce que l'on peut conclure :** l'erreur absolue moyenne est faible comparée à l'échelle de la cible (écart-type de `yr_test` ≈ 97,6 et MAE ≈ 74,0 pour un modèle qui prédit toujours la moyenne — voir la cellule *Repères*). C'est cohérent : `make_regression` génère une relation **exactement linéaire** + bruit gaussien, donc le modèle linéaire est ici le « bon » modèle et l'erreur résiduelle est surtout due au bruit.
- **Ce que l'on ne peut pas conclure :** ce résultat ne dit rien de la performance sur des données économiques réelles ; une seule métrique sur un seul découpage ne mesure ni la variance de l'estimation ni la significativité des coefficients.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression Ridge** (*Ridge Regression*, pénalité L2) — régression supervisée : stabiliser les coefficients quand les variables sont corrélées.
**Étapes :** variables standardisées → minimisation erreur quadratique + pénalité sur la taille des coefficients → prédiction linéaire.
$$J(\beta)=\tfrac{1}{n}\lVert y-X\beta\rVert_2^2+\lambda\textstyle\sum_{j}\beta_j^2$$
$X$ : matrice $n\times p$ ; $y$ : cible ; $\beta$ : coefficients ; $\lVert\cdot\rVert_2$ : norme euclidienne ; $\lambda$ : force de pénalité.
**Paramètres appris :** $\beta$. **Hyperparamètre :** $\lambda$ (`alpha`), choisi par validation.
**Avantages :** robuste à la colinéarité ; réduit le surapprentissage.
**Limites :** n'annule aucun coefficient (pas de sélection) ; impose de standardiser et de régler $\lambda$.

### Schéma explicatif

**Comment lire le schéma :** en abscisse la force de pénalité $\lambda$ (échelle logarithmique), en ordonnée la valeur de chaque coefficient. À gauche (λ faible) on retrouve les moindres carrés ; vers la droite, tous les coefficients **rétrécissent progressivement vers 0 sans s'annuler**. Les variables x1 et x2, construites très corrélées, voient leurs coefficients se rapprocher l'un de l'autre : Ridge répartit l'effet entre variables corrélées. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** prévoir le rendement d'un portefeuille ou la croissance du PIB à partir d'indicateurs macroéconomiques très corrélés (taux, inflation, crédit, masse monétaire).
- **Cas d'usage 2 (marketing) :** modéliser les ventes à partir de nombreux canaux publicitaires dont les dépenses évoluent ensemble (*marketing mix modelling*).
- **Données et prérequis :** variables numériques **standardisées** (sinon la pénalité est injuste entre unités), cible continue ; choix de `alpha` par validation croisée, scaler ajusté sur l'entraînement uniquement (Pipeline).
- **Peu approprié :** quand l'objectif est d'identifier un petit sous-ensemble de variables pertinentes (ex. sélectionner 5 ratios parmi 200) — Ridge les garde tous ; préférer Lasso ou Elastic Net. Inutile aussi si la relation est fortement non linéaire.
- **Métriques :** MAE, RMSE, R² ; courbe de validation en fonction de `alpha`.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques que l'exemple 1 ; pipeline `StandardScaler` → `Ridge`, le scaler étant ajusté sur l'entraînement seulement (pas de fuite).
- **Hyperparamètre :** `alpha=1.0` (fixé, non réglé).
- **Sortie obtenue :** **MAE ≈ 9,37**, quasiment identique à la régression linéaire (≈ 9,35).
- **Conclusion possible :** sur ces données, la pénalité ne change presque rien : il n'y a que 5 variables indépendantes, 210 observations et aucune colinéarité, donc il n'y a pas de variance excessive à réduire. La légère hausse de MAE reflète le petit biais introduit par la pénalité.
- **Limites :** `alpha` n'a pas été choisi par validation croisée ; on ne peut pas conclure que Ridge « n'apporte rien » en général — son intérêt apparaît avec des variables corrélées ou nombreuses, absentes ici.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression Lasso** (*Lasso*, pénalité L1) — régression supervisée : prédire et **sélectionner** les variables utiles.
**Étapes :** variables standardisées → erreur quadratique + pénalité en valeur absolue → certains coefficients deviennent exactement nuls.
$$J(\beta)=\tfrac{1}{2n}\lVert y-X\beta\rVert_2^2+\lambda\textstyle\sum_{j}|\beta_j|$$
$n$ : observations ; $X$ : variables ; $y$ : cible ; $\beta_j$ : coefficients ; $|\cdot|$ : valeur absolue ; $\lambda$ : pénalité L1.
**Paramètres appris :** $\beta$. **Hyperparamètres :** $\lambda$ (`alpha`), `max_iter`.
**Avantages :** sélection intégrée ; modèle simple à expliquer.
**Limites :** instable avec variables très corrélées ; biais sur les grands coefficients.

### Schéma explicatif

**Comment lire le schéma :** comme pour Ridge, chaque courbe suit un coefficient quand $\lambda$ augmente. La différence clé : les courbes **touchent exactement 0** et y restent. Les variables de bruit (pointillés) sont éliminées en premier ; à λ très élevé, même les variables utiles disparaissent. Lire le schéma de gauche à droite revient à lire un **classement d'importance** des variables. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** sélectionner, parmi des dizaines de ratios financiers (liquidité, rentabilité, endettement…), ceux qui expliquent le mieux la note de crédit ou le spread d'une entreprise.
- **Cas d'usage 2 (commerce international) :** identifier les déterminants principaux du volume d'exportations d'un pays parmi de nombreux indicateurs (taux de change, PIB partenaires, tarifs…).
- **Données et prérequis :** variables numériques standardisées, cible continue ; suffisamment d'observations ; `alpha` choisi par validation croisée (`LassoCV`).
- **Peu approprié :** variables très corrélées dont on veut garder l'effet collectif (ex. plusieurs taux d'intérêt de maturités voisines) : le Lasso en retient arbitrairement un et la sélection change d'un échantillon à l'autre → préférer Ridge ou Elastic Net.
- **Métriques :** MAE, RMSE, R² ; nombre de coefficients non nuls et stabilité de la sélection (bootstrap).

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression ; pipeline `StandardScaler` → `Lasso`.
- **Hyperparamètres :** `alpha=0.1` (pénalité faible), `max_iter=10000` pour garantir la convergence.
- **Sortie obtenue :** **R² ≈ 0,986** sur le test : environ 98,6 % de la variance de la cible de test est expliquée.
- **Conclusion possible :** le modèle linéaire pénalisé reproduit très bien une relation qui est, par construction, linéaire. Avec `alpha=0.1` relativement à des coefficients de l'ordre de 10 à 64, la pénalité est trop faible pour annuler une variable : en inspectant `modele[-1].coef_`, les **5 coefficients restent non nuls** (≈ 9,8 ; 16,5 ; 20,6 ; 63,0 ; 63,6), ce qui est logique car les 5 variables générées sont toutes informatives.
- **Limites :** le R² n'est pas comparable directement aux MAE des exemples 1–2 (métrique différente) ; un R² élevé sur données synthétiques ne garantit rien sur données réelles, et la sélection de variables — l'intérêt principal du Lasso — n'est pas illustrée par cet exemple.

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression polynomiale** (*Polynomial Regression*) — régression supervisée : modéliser une relation courbe.
**Étapes :** création des puissances et interactions des variables → régression linéaire (ici Ridge) sur ces nouvelles variables → prédiction non linéaire.
$$\hat y=\beta_0+\beta_1x+\beta_2x^2+\dots+\beta_dx^d$$
$x$ : variable d'entrée ; $d$ : degré maximal ; $\beta_j$ : coefficients ; $\hat y$ : prédiction.
**Paramètres appris :** $\beta_j$. **Hyperparamètres :** $d$ (`degree`), `alpha` de la Ridge.
**Avantages :** capte courbures et interactions simples ; reste un modèle linéaire connu.
**Limites :** explosion du nombre de variables ; surapprentissage et extrapolation dangereuse.

### Schéma explicatif

**Comment lire le schéma :** les points noirs suivent une parabole bruitée. La droite (degré 1, bleu) est trop rigide : **sous-apprentissage**. La courbe de degré 2 (vert) suit la forme réelle. La courbe de degré 15 (rouge) serpente pour passer près de chaque point et diverge aux bords : **surapprentissage** et mauvaise extrapolation. Le degré est donc un hyperparamètre à choisir sur un jeu de validation. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/gestion) :** modéliser une courbe de coût avec économies d'échelle (coût unitaire qui baisse puis remonte avec le volume produit), ou une relation âge–revenu en courbe en cloche.
- **Cas d'usage 2 (marketing) :** rendement décroissant des dépenses publicitaires (effet de saturation) ou effet d'interaction prix × promotion.
- **Données et prérequis :** peu de variables numériques continues, standardisation recommandée (puissances d'échelles très différentes), régularisation (Ridge) et validation du degré.
- **Peu approprié :** données à grande dimension (100 variables en degré 3 → plus de 170 000 termes) ou prévision hors de la plage observée (ex. extrapoler une tendance de prix sur 10 ans) : le polynôme explose. Préférer arbres/boosting ou splines.
- **Métriques :** MAE, RMSE, R² ; comparaison des erreurs d'entraînement et de validation selon le degré.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression (5 variables) ; pipeline `StandardScaler` → `PolynomialFeatures(degree=2, include_bias=False)` (5 variables → 20 variables : 5 linéaires, 5 carrés, 10 interactions) → `Ridge(alpha=10)`.
- **Hyperparamètres :** `degree=2`, `alpha=10`.
- **Sortie obtenue :** **MAE ≈ 10,90**, soit **plus élevée** que la régression linéaire (≈ 9,35).
- **Conclusion possible :** comme la vraie relation est linéaire, les termes quadratiques et d'interaction n'apportent pas d'information ; ils ajoutent de la variance, et la pénalité `alpha=10` (plus forte qu'à l'exemple 2) rétrécit aussi les coefficients linéaires utiles → biais. Plus complexe n'est donc pas meilleur lorsque le modèle simple est déjà correctement spécifié.
- **Limites :** degré et `alpha` n'ont pas été optimisés par validation ; un seul découpage ; on ne peut pas généraliser ce classement à des données réellement non linéaires.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Arbre de régression** (*Regression Tree*, CART) — régression supervisée : prédire une valeur en découpant l'espace en zones homogènes.
**Étapes :** à chaque nœud, choix de la variable et du seuil qui réduisent le plus la variance → découpage récursif → prédiction = moyenne de la feuille.
$$\hat y(x)=\tfrac{1}{|L(x)|}\textstyle\sum_{i\in L(x)}y_i$$
$L(x)$ : observations de la feuille contenant $x$ ; $|L(x)|$ : effectif ; $y_i$ : leurs valeurs.
**Paramètres appris :** variables, seuils, valeurs des feuilles. **Hyperparamètres :** `max_depth`, `min_samples_leaf`.
**Avantages :** capte non-linéarités et interactions ; règles lisibles, sans standardisation.
**Limites :** instable ; prédiction en escalier, aucune extrapolation.

### Schéma explicatif

**Comment lire le schéma :** à droite, chaque nœud pose une question « x ≤ seuil ? » ; on descend à gauche si oui, à droite sinon, jusqu'à une feuille qui contient le nombre d'observations (`samples`) et leur moyenne (`value`). À gauche, la ligne rouge montre que l'arbre prédit une **constante par zone** : 4 feuilles = 4 paliers. Augmenter `max_depth` multiplie les paliers (jusqu'au surapprentissage). Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique) :** estimer le coût d'une expédition selon la distance, le poids, le mode de transport et la saison, avec des règles lisibles pour les opérationnels.
- **Cas d'usage 2 (gestion) :** estimer la durée de traitement d'un dossier de crédit selon le type de client et le montant.
- **Données et prérequis :** tableau mixte numérique/catégoriel, pas de standardisation nécessaire ; contrôle de la profondeur ou élagage pour éviter le surapprentissage.
- **Peu approprié :** prévoir une tendance qui sort de la plage historique (ex. prix de l'immobilier en hausse continue) : l'arbre ne peut pas prédire au-delà de la moyenne de sa feuille la plus extrême. Peu adapté aussi aux relations lisses et linéaires (approximées par marches d'escalier).
- **Métriques :** MAE, RMSE, R² ; comparaison erreur d'entraînement / validation selon `max_depth`.

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression (5 variables, relation linéaire).
- **Hyperparamètres :** `max_depth=5` (au plus 32 feuilles), `random_state=42`.
- **Sortie obtenue :** **MAE ≈ 37,72**, environ **4 fois** celle de la régression linéaire (≈ 9,35), mais meilleure que la prédiction constante par la moyenne (≈ 74,0).
- **Conclusion possible :** l'arbre approxime un hyperplan lisse par des paliers ; avec seulement 210 observations réparties en ≤ 32 feuilles, chaque feuille moyenne peu de points et la discrétisation génère une grosse erreur. Il n'est pas adapté à une relation additive linéaire, ce qui illustre bien sa limite.
- **Limites :** profondeur non réglée par validation ; un arbre unique est instable, la MAE pourrait varier sensiblement avec un autre découpage aléatoire.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Forêt aléatoire — régression** (*Random Forest Regressor*) — régression supervisée par bagging : stabiliser et améliorer l'arbre unique.
**Étapes :** $B$ échantillons bootstrap → un arbre par échantillon (sous-ensemble aléatoire de variables à chaque coupure) → moyenne des prédictions.
$$\hat y(x)=\tfrac{1}{B}\textstyle\sum_{b=1}^{B}T_b(x)$$
$B$ : nombre d'arbres ; $T_b(x)$ : prédiction de l'arbre $b$ ; $x$ : nouvelle observation.
**Paramètres appris :** les arbres. **Hyperparamètres :** `n_estimators`, `max_depth`, `max_features`.
**Avantages :** réduit la variance, robuste ; peu de réglage, importance des variables.
**Limites :** volumineux et peu interprétable ; n'extrapole pas.

### Schéma explicatif

**Comment lire le schéma :** chaque courbe bleue pâle est un arbre entraîné sur un échantillon bootstrap différent : elles sont irrégulières et diffèrent entre elles (**forte variance**). La courbe rouge est leur moyenne : plus lisse et plus proche de la tendance. C'est le principe du *bagging* : agréger des modèles instables et peu corrélés pour réduire la variance. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (commerce international) :** prévoir la demande d'exportation d'un produit selon prix, taux de change, saison et pays de destination.
- **Cas d'usage 2 (finance) :** estimer la valeur d'un bien immobilier en garantie d'un crédit (variables hétérogènes, relations non linéaires).
- **Données et prérequis :** tableau numérique/catégoriel, pas de standardisation ; plusieurs centaines d'observations au minimum ; mémoire et temps de calcul proportionnels au nombre d'arbres.
- **Peu approprié :** prévision d'une série en tendance (inflation, cours en hausse) où il faut extrapoler ; ou quand une explication réglementaire précise de chaque prédiction est exigée (ex. justification d'un refus de crédit) — un modèle linéaire ou un arbre unique est plus défendable.
- **Métriques :** MAE, RMSE, R² ; erreur *out-of-bag* ; importance des variables par permutation.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression.
- **Hyperparamètres :** `n_estimators=80`, arbres de profondeur non limitée (défaut), `random_state=42`.
- **Sortie obtenue :** **MAE ≈ 24,37**, nettement meilleure que l'arbre unique (≈ 37,72) mais toujours bien moins bonne que la régression linéaire (≈ 9,35).
- **Conclusion possible :** l'agrégation réduit bien la variance de l'arbre (gain d'environ 35 % sur la MAE), ce qui illustre l'intérêt du bagging. Mais la moyenne d'arbres reste une fonction en paliers qui approxime mal un hyperplan, surtout aux extrémités de la distribution (pas d'extrapolation).
- **Limites :** pas de validation des hyperparamètres ; données synthétiques linéaires qui avantagent structurellement le modèle linéaire ; ce classement ne se transpose pas à des données réelles non linéaires.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Gradient Boosting — régression** (*Gradient Boosting Regressor*) — régression supervisée par boosting : additionner de petits arbres qui corrigent les erreurs.
**Étapes :** prédiction initiale (moyenne) → arbre ajusté sur les résidus → ajout pondéré par $\eta$ → répétition $M$ fois.
$$F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$$
$F_m$ : modèle à l'itération $m$ ; $h_m$ : arbre ajusté aux résidus ; $\eta$ : taux d'apprentissage ; $M$ : nombre d'itérations.
**Paramètres appris :** les arbres. **Hyperparamètres :** $\eta$ (`learning_rate`), $M$ (`max_iter`), profondeur.
**Avantages :** très performant sur données tabulaires ; gère valeurs manquantes (version Hist).
**Limites :** réglage sensible, risque de surapprentissage ; peu interprétable.

### Schéma explicatif

**Comment lire le schéma :** à gauche, $F_1$ (un seul petit arbre après la moyenne) est grossier ; $F_5$ capte déjà la forme générale ; $F_{50}$ suit finement la courbe. À droite, l'erreur d'entraînement diminue à chaque itération car chaque arbre corrige les résidus du modèle précédent. Attention : l'erreur d'**entraînement** baisse toujours — c'est l'erreur de **validation** qui indique quand s'arrêter. Données simulées, $\eta = 0{,}3$.

### Contextes d'utilisation
- **Cas d'usage 1 (commerce/gestion) :** prévision des ventes journalières par magasin et par produit (promotions, jours fériés, météo) — très utilisé en retail.
- **Cas d'usage 2 (finance) :** estimation de la perte en cas de défaut (LGD) ou de la valeur client (CLV) à partir de nombreuses variables hétérogènes.
- **Données et prérequis :** données tabulaires de taille moyenne à grande, pas de standardisation ; jeu de validation (ou *early stopping*) pour fixer le nombre d'itérations ; découpage chronologique si séries temporelles.
- **Peu approprié :** très petit jeu de données (quelques dizaines de lignes) où il surapprend facilement, ou données non tabulaires (images, texte brut) où les réseaux de neurones dominent ; extrapolation de tendance impossible comme pour tous les arbres.
- **Métriques :** MAE, RMSE, R² ; courbe d'apprentissage (erreur de validation en fonction de $m$).

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression.
- **Hyperparamètres :** `max_iter=80` itérations, autres valeurs par défaut (`learning_rate=0.1`, `max_leaf_nodes=31`), `random_state=42`.
- **Sortie obtenue :** **MAE ≈ 23,27**, légèrement meilleure que la forêt aléatoire (≈ 24,37), mais toujours loin de la régression linéaire (≈ 9,35).
- **Conclusion possible :** le boosting corrige progressivement les erreurs et fait mieux que les autres méthodes à base d'arbres ici, mais avec seulement 210 observations et une relation exactement linéaire, les arbres restent désavantagés. La hiérarchie obtenue (linéaire > boosting ≈ forêt > arbre seul) est propre à ce jeu synthétique.
- **Limites :** `max_iter` et `learning_rate` non réglés (pas de validation ni d'*early stopping*) ; pas de mesure de variance entre plusieurs graines ; aucune conclusion sur des données réelles.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression logistique** (*Logistic Regression*) — classification supervisée : estimer la probabilité d'appartenir à la classe positive.
**Étapes :** score linéaire → sigmoïde → probabilité → classe 1 si $p\ge0{,}5$ ; coefficients estimés par maximum de vraisemblance.
$$P(y=1\mid x)=\frac{1}{1+e^{-(\beta_0+\beta^\top x)}}$$
$x$ : variables ; $\beta_0,\beta$ : coefficients ; $y\in\{0,1\}$ : classe.
**Paramètres appris :** $\beta_0,\beta$. **Hyperparamètres :** $C$ (inverse de la régularisation), seuil de décision.
**Avantages :** probabilités exploitables ; coefficients interprétables (standard en scoring).
**Limites :** frontière linéaire ; sensible au déséquilibre des classes si le seuil n'est pas ajusté.

### Schéma explicatif

**Comment lire le schéma :** à gauche, la sigmoïde transforme n'importe quel score réel en probabilité entre 0 et 1 ; score nul ↔ probabilité 0,5. À droite, la couleur de fond donne la probabilité estimée d'appartenir à la classe 1 (rouge) ; la **ligne noire** est la frontière où $p=0{,}5$ — elle est **droite**, car le score est linéaire. Plus un point est loin de la frontière, plus la prédiction est confiante. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** scoring de crédit — probabilité de défaut d'un emprunteur à partir de son revenu, endettement et historique (ex. dataset *Default of Credit Card Clients*).
- **Cas d'usage 2 (marketing) :** probabilité qu'un client réponde à une campagne ou résilie son abonnement (*churn*).
- **Données et prérequis :** cible binaire (ou multiclasse), variables numériques standardisées, catégorielles encodées ; pas de séparation parfaite des classes ; ajuster le seuil ou pondérer les classes si déséquilibre.
- **Peu approprié :** frontière fortement non linéaire (ex. détection de fraude reposant sur des combinaisons complexes de comportements) sans ingénierie de variables — un modèle à base d'arbres capturera mieux ces interactions.
- **Métriques :** F1, précision, rappel, ROC-AUC (ou PR-AUC si classe rare), log-perte et calibration.

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

- **Hyperparamètres :** pipeline `StandardScaler` → `LogisticRegression(max_iter=1000)` ; régularisation L2 par défaut avec `C=1.0`, seuil 0,5.
- **Sortie obtenue :** **F1 ≈ 0,811** sur la classe positive.
- **Conclusion possible :** le modèle linéaire classe correctement une large majorité des observations, mais il est parmi les plus faibles des classifieurs testés (KNN ≈ 0,889, boosting ≈ 0,950). Cela suggère que la frontière entre classes générée par `make_classification` n'est pas purement linéaire (classes composées de plusieurs amas gaussiens).

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**K plus proches voisins** (*K-Nearest Neighbors*, KNN) — classification supervisée : classer un point selon ses voisins.
**Étapes :** mémorisation des données → distances du nouveau point à tous les points → $K$ plus proches → vote majoritaire.
$$\hat y(x)=\arg\max_c\textstyle\sum_{i\in\mathcal N_K(x)}\mathbb 1(y_i=c)$$
$\mathcal N_K(x)$ : les $K$ voisins les plus proches ; $c$ : classe ; $\mathbb 1$ : indicatrice.
**Paramètres appris :** aucun (le modèle est le jeu d'entraînement). **Hyperparamètres :** $K$, distance, pondération.
**Avantages :** aucune hypothèse sur la frontière ; simple à expliquer.
**Limites :** lent et gourmand en mémoire ; sensible à l'échelle et à la dimension.

### Schéma explicatif

**Comment lire le schéma :** l'étoile rouge est le point à classer. Les pointillés le relient à ses 5 plus proches voisins, contenus dans le cercle. La classe prédite est celle qui domine dans ce cercle (indiquée dans le titre). Avec un $K$ plus petit, la décision dépend de quelques points (frontière irrégulière, variance élevée) ; avec un $K$ plus grand, elle se lisse (biais plus élevé). Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** recommander une offre à un client en s'inspirant des choix des clients au profil le plus proche (âge, panier moyen, fréquence d'achat).
- **Cas d'usage 2 (finance) :** classer une demande de crédit en comparant le demandeur aux dossiers passés les plus similaires ; imputation de valeurs manquantes (KNNImputer).
- **Données et prérequis :** variables numériques **standardisées** (obligatoire), dimension modérée, volume raisonnable (calcul des distances) ; choix de $K$ par validation croisée.
- **Peu approprié :** millions de transactions bancaires à scorer en temps réel (coût de calcul à chaque prédiction), ou données de très grande dimension (ex. 5 000 variables textuelles) où toutes les distances deviennent similaires.
- **Métriques :** F1, précision, rappel, ROC-AUC (avec probabilités = proportion de voisins) ; accuracy si classes équilibrées.

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `StandardScaler` (indispensable pour des distances) → `KNeighborsClassifier(n_neighbors=5)`, distance euclidienne, vote non pondéré.
- **Sortie obtenue :** **F1 ≈ 0,889**, nettement au-dessus de la régression logistique (≈ 0,811).
- **Conclusion possible :** un classifieur purement local, sans hypothèse de linéarité, fait mieux ici : cela confirme que les classes forment des amas que la frontière linéaire sépare mal. Avec seulement 6 variables, la malédiction de la dimension reste limitée.
- **Limites :** $K=5$ non choisi par validation croisée ; sur 100 observations de test, l'écart avec d'autres modèles reste incertain ; aucune information sur le coût de prédiction, qui serait le vrai frein sur un gros jeu réel.

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Bayésien naïf gaussien** (*Gaussian Naive Bayes*) — classification supervisée probabiliste : choisir la classe la plus probable.
**Étapes :** estimation par classe de la fréquence et d'une loi normale par variable → théorème de Bayes avec variables supposées indépendantes → classe de probabilité maximale.
$$P(c\mid x)\propto P(c)\textstyle\prod_{j}P(x_j\mid c)$$
$c$ : classe ; $P(c)$ : probabilité a priori ; $P(x_j\mid c)$ : loi normale de la variable $j$ dans la classe $c$.
**Paramètres appris :** $P(c)$, moyennes et variances $\mu_{jc},\sigma^2_{jc}$. **Hyperparamètre :** `var_smoothing`.
**Avantages :** très rapide, peu de données nécessaires ; efficace en grande dimension (texte).
**Limites :** hypothèse d'indépendance irréaliste ; suppose des lois normales par classe.

### Schéma explicatif

**Comment lire le schéma :** pour une variable, le modèle ajuste une courbe normale par classe (histogrammes = données observées), pondérée par la fréquence de la classe $P(c)$. Pour une nouvelle valeur de $x$, on compare la hauteur des deux courbes : la plus haute donne la classe prédite. Le seuil (pointillés) est l'endroit où elles se croisent. Avec plusieurs variables, le modèle **multiplie** ces densités comme si les variables étaient indépendantes. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/gestion) :** classer automatiquement des emails clients (réclamation, demande d'information, spam) ou analyser le sentiment d'avis produits (version multinomiale sur les mots).
- **Cas d'usage 2 (finance) :** filtre de première ligne pour signaler des transactions suspectes, quand il faut un modèle très rapide et peu de données d'entraînement.
- **Données et prérequis :** variables continues ~normales par classe (GaussianNB) ou comptages/fréquences de mots (MultinomialNB) ; pas de standardisation nécessaire.
- **Peu approprié :** scoring de crédit avec variables fortement corrélées (revenu, patrimoine, montant emprunté) : l'hypothèse d'indépendance « compte deux fois » la même information et les probabilités deviennent trop extrêmes — inutilisables pour fixer un prix du risque.
- **Métriques :** F1, précision, rappel, ROC-AUC (utile car le classement peut rester bon même si les probabilités sont mal calibrées), courbe de calibration.

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** valeurs par défaut (`var_smoothing=1e-9`, a priori estimés sur les fréquences). Pas de standardisation, inutile pour ce modèle.
- **Sortie obtenue :** **F1 ≈ 0,755**, le plus faible des classifieurs de ce notebook.
- **Conclusion possible :** `make_classification` génère chaque classe comme **plusieurs amas gaussiens** (`n_clusters_per_class=2` par défaut) avec des variables informatives corrélées : la distribution par classe n'est ni une seule gaussienne ni à variables indépendantes. L'hypothèse du modèle est violée, d'où la performance moindre.
- **Limites :** on ne peut pas en déduire que Naive Bayes est « mauvais » en général — il excelle sur le texte. Sans matrice de confusion ni ROC-AUC, on ne sait pas si le classement des probabilités reste exploitable avec un autre seuil.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Arbre de décision** (*Decision Tree Classifier*) — classification supervisée : prédire une classe par des règles « si… alors ».
**Étapes :** choix à chaque nœud de la coupure qui rend les sous-groupes les plus purs → découpage récursif → classe majoritaire de la feuille.
$$G(t)=1-\textstyle\sum_{c}p_{c,t}^2$$
$G(t)$ : impureté de Gini du nœud $t$ ; $p_{c,t}$ : proportion de la classe $c$ dans le nœud.
**Paramètres appris :** variables, seuils, classes des feuilles. **Hyperparamètres :** `max_depth`, `min_samples_leaf`, critère.
**Avantages :** très interprétable ; variables mixtes, sans standardisation.
**Limites :** instable, surapprentissage s'il est profond ; frontières en escalier.

### Schéma explicatif

**Comment lire le schéma :** à gauche, chaque nœud affiche sa règle, son indice de Gini, le nombre d'observations et la répartition par classe ; la couleur indique la classe majoritaire et son intensité la pureté. À droite, les mêmes règles découpent le plan en **rectangles** : chaque rectangle est une feuille. Les noms « endettement / ancienneté » ne servent qu'à rendre l'exemple concret ; les données sont simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** grille d'octroi de crédit simple et auditable (« si taux d'endettement > 40 % et ancienneté < 1 an → refus »), exigée pour la transparence réglementaire.
- **Cas d'usage 2 (marketing) :** segmentation explicative des clients qui répondent à une offre, pour construire des règles de ciblage compréhensibles par les équipes commerciales.
- **Données et prérequis :** variables mixtes, aucune standardisation ; contrôle de la profondeur et de la taille minimale des feuilles ; classes pondérées si déséquilibre.
- **Peu approprié :** lorsque la performance maximale prime sur l'interprétabilité (ex. détection de fraude à grande échelle) : un arbre seul est instable et dominé par les forêts ou le boosting ; peu adapté aussi aux frontières obliques (combinaisons linéaires de variables).
- **Métriques :** F1, précision, rappel, ROC-AUC ; écart entraînement/validation pour diagnostiquer le surapprentissage.

```python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `max_depth=4` (au plus 16 feuilles), critère de Gini, `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,891**, comparable au KNN (≈ 0,889) et supérieur à la logistique (≈ 0,811).
- **Conclusion possible :** même peu profond, l'arbre capte les interactions et la structure en amas des classes mieux qu'une frontière linéaire. Contrairement à la régression (exemple 5), la tâche de classification avec frontières par morceaux convient mieux à un arbre.
- **Limites :** un arbre unique est instable : sur un autre découpage, la valeur pourrait varier de plusieurs points ; `max_depth=4` n'a pas été validé ; le F1 n'indique pas la nature des erreurs.

## 12. Forêt aléatoire — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Forêt aléatoire — classification** (*Random Forest Classifier*) — classification supervisée par bagging : combiner des arbres décorrélés.
**Étapes :** échantillons bootstrap → un arbre par échantillon (variables tirées au hasard à chaque coupure) → vote majoritaire.
$$\hat y(x)=\arg\max_c\tfrac{1}{B}\textstyle\sum_{b=1}^{B}\hat p_b(c\mid x)$$
$B$ : nombre d'arbres ; $\hat p_b(c\mid x)$ : proportion de la classe $c$ dans la feuille de l'arbre $b$.
**Paramètres appris :** les arbres. **Hyperparamètres :** `n_estimators`, `max_features`, `max_depth`.
**Avantages :** performant et robuste sans réglage ; importance des variables.
**Limites :** peu interprétable ; coûteux en mémoire et en prédiction.

### Schéma explicatif

**Comment lire le schéma :** à gauche, un arbre profond découpe le plan en zones nettes, avec des « îlots » créés pour isoler des points bruités : **surapprentissage**. À droite, la forêt moyenne 100 arbres : la couleur indique la part d'arbres votant pour la classe rouge ; la frontière est plus lisse et les zones incertaines (couleurs pâles) apparaissent là où les classes se mélangent réellement. Données simulées (`make_moons`).

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** prédiction du défaut de paiement des titulaires de cartes de crédit (variables de paiement, montants, historique).
- **Cas d'usage 2 (logistique/commerce) :** prédire si une commande sera livrée en retard (transporteur, distance, période, stock) pour alerter le client.
- **Données et prérequis :** données tabulaires mixtes, pas de standardisation ; plusieurs centaines d'observations minimum ; `class_weight='balanced'` en cas de classes déséquilibrées.
- **Peu approprié :** décision qui doit être justifiée règle par règle à un régulateur ou un client (un arbre simple ou une logistique est plus défendable), ou prédiction à très faible latence sur matériel limité (des centaines d'arbres à parcourir).
- **Métriques :** F1, rappel, précision, ROC-AUC / PR-AUC selon le déséquilibre ; score *out-of-bag* ; importance par permutation.

```python
from sklearn.ensemble import RandomForestClassifier
modele=RandomForestClassifier(n_estimators=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `n_estimators=80`, profondeur non limitée, `max_features='sqrt'` (défaut), `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,918**, au-dessus de l'arbre unique (≈ 0,891).
- **Conclusion possible :** l'agrégation d'arbres décorrélés améliore la généralisation par rapport à un arbre seul, conformément à la théorie du bagging.
- **Limites :** l'écart de ≈ 0,03 de F1 sur 100 observations correspond à quelques prédictions seulement, il faudrait une validation croisée pour le confirmer ; pas de réglage d'hyperparamètres ; pas de mesure du coût de calcul ni de l'importance des variables.

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Machine à vecteurs de support** (*Support Vector Machine*, SVM) — classification supervisée : séparer les classes avec la marge maximale.
**Étapes :** variables standardisées → frontière la plus éloignée des points les plus proches (vecteurs de support) → noyau RBF pour une frontière courbe.
$$\min_{w,b}\ \tfrac12\lVert w\rVert^2+C\textstyle\sum_i\xi_i$$
$w,b$ : orientation et position de la frontière ; $\xi_i$ : violation de marge du point $i$ ; $C$ : compromis marge/erreurs.
**Paramètres appris :** $w,b$ (via les vecteurs de support). **Hyperparamètres :** $C$, noyau, $\gamma$.
**Avantages :** efficace en grande dimension ; frontières complexes grâce aux noyaux.
**Limites :** lent sur gros volumes ; pas de probabilités natives, réglage sensible.

### Schéma explicatif

**Comment lire le schéma :** la ligne pleine est la frontière de décision ; les pointillés délimitent la **marge** (fonction de décision = ±1). Les points entourés en vert sont les **vecteurs de support** : ce sont les seuls qui déterminent la frontière — supprimer un autre point ne la modifierait pas. À droite, aucune droite ne sépare les deux cercles ; le noyau RBF produit une frontière circulaire. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** prédiction de faillite d'entreprises à partir de quelques dizaines de ratios financiers sur un échantillon de taille moyenne (quelques milliers de sociétés).
- **Cas d'usage 2 (marketing) :** classification de textes (avis clients positifs/négatifs) en grande dimension (représentation TF-IDF), domaine où le SVM linéaire est historiquement très performant.
- **Données et prérequis :** variables numériques **standardisées** (obligatoire), jeu de taille petite à moyenne ; réglage de $C$ et $\gamma$ par validation croisée (GridSearch).
- **Peu approprié :** scoring de millions de transactions (temps d'entraînement quadratique à cubique en $n$), ou besoin de probabilités fiables pour tarifer un risque (il faut une calibration supplémentaire).
- **Métriques :** F1, précision, rappel, ROC-AUC (via `decision_function`) ; accuracy si classes équilibrées.

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `StandardScaler` → `SVC(kernel='rbf', C=1.0)`, `gamma='scale'` par défaut. (`random_state` n'a d'effet que si `probability=True`.)
- **Sortie obtenue :** **F1 ≈ 0,898**, proche de l'arbre (≈ 0,891) et du KNN (≈ 0,889), au-dessus de la logistique (≈ 0,811).
- **Conclusion possible :** le noyau RBF permet une frontière non linéaire, ce qui améliore nettement le résultat par rapport au modèle linéaire sur ces classes en amas.
- **Limites :** $C$ et $\gamma$ ne sont pas réglés (valeurs par défaut) — le SVM est très sensible à ces choix ; pas de probabilités donc pas de ROC-AUC calculée ; données synthétiques et petit test.

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Gradient Boosting — classification** (*Gradient Boosting Classifier*) — classification supervisée par boosting : petits arbres qui corrigent successivement les erreurs.
**Étapes :** score initial → arbre ajusté sur l'écart entre classe réelle et probabilité prédite → ajout pondéré par $\eta$ → sigmoïde pour la probabilité.
$$F_m(x)=F_{m-1}(x)+\eta\,h_m(x),\qquad p(x)=\sigma(F_M(x))$$
$F_m$ : score après $m$ arbres ; $h_m$ : nouvel arbre ; $\eta$ : taux d'apprentissage ; $\sigma$ : sigmoïde.
**Paramètres appris :** les arbres. **Hyperparamètres :** $\eta$, `max_iter`, `max_leaf_nodes`.
**Avantages :** parmi les meilleurs sur données tabulaires ; rapide (version Hist).
**Limites :** sensible aux hyperparamètres ; peu interprétable.

### Schéma explicatif

**Comment lire le schéma :** de gauche à droite, le même modèle avec 1, 10 puis 100 petits arbres (profondeur 2). Avec un seul arbre, la frontière est un simple découpage grossier ; à 10 arbres, la forme en croissants apparaît ; à 100, la frontière épouse les données (avec un début de surapprentissage localement). Le nombre d'arbres est donc un levier de complexité à contrôler par validation. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** détection de fraude à la carte bancaire ou probabilité de défaut, avec des milliers de variables comportementales.
- **Cas d'usage 2 (marketing) :** prédiction de l'attrition client (*churn*) dans les télécoms ou la banque, pour cibler les actions de rétention.
- **Données et prérequis :** données tabulaires de taille moyenne à grande ; validation (ou `early_stopping`) pour fixer $M$ ; `class_weight` si déséquilibre ; découpage chronologique si les données sont datées.
- **Peu approprié :** petits échantillons (quelques centaines de lignes) avec exigence d'explicabilité forte — par exemple une grille de score réglementaire Bâle — où une régression logistique est plus robuste et justifiable.
- **Métriques :** F1, précision, rappel, ROC-AUC, PR-AUC pour classes rares ; courbe d'apprentissage en fonction de $M$.

```python
from sklearn.ensemble import HistGradientBoostingClassifier
modele=HistGradientBoostingClassifier(max_iter=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `max_iter=80`, `learning_rate=0.1` et `max_leaf_nodes=31` par défaut, `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,950**, la **meilleure valeur** des classifieurs du notebook (devant XGBoost ≈ 0,931 et la forêt ≈ 0,918).
- **Conclusion possible :** la correction séquentielle des erreurs exploite bien les interactions et la structure non linéaire de ces données ; c'est cohérent avec la réputation du boosting sur données tabulaires.
- **Limites :** les écarts entre méthodes d'ensemble (0,92 à 0,95) portent sur quelques observations de test et ne sont pas forcément significatifs ; aucun réglage ni validation croisée ; pas de mesure du temps de calcul ni de la calibration.

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**AdaBoost** (*Adaptive Boosting*) — classification supervisée par boosting : combiner des classifieurs faibles (souches).
**Étapes :** poids égaux → souche entraînée → poids de vote selon sa précision → hausse du poids des points mal classés → vote pondéré final.
$$\alpha_m=\tfrac12\ln\frac{1-\varepsilon_m}{\varepsilon_m},\qquad H(x)=\mathrm{sign}\Big(\textstyle\sum_m\alpha_m h_m(x)\Big)$$
$h_m$ : classifieur faible ; $\varepsilon_m$ : son erreur pondérée ; $\alpha_m$ : son poids de vote ; $H$ : classifieur final.
**Paramètres appris :** $h_m$, $\alpha_m$. **Hyperparamètres :** `n_estimators`, `learning_rate`.
**Avantages :** simple, peu de réglages ; rend forts des modèles très faibles.
**Limites :** très sensible au bruit et aux étiquettes erronées ; moins performant que le gradient boosting.

### Schéma explicatif

**Comment lire le schéma :** chaque panneau montre un *stump* (une seule coupure verticale ou horizontale, fond coloré) entraîné avec les poids courants. La **taille des points** représente leur poids : à l'étape 1 tous sont égaux ; les points mal classés grossissent ensuite, forçant le *stump* suivant à les traiter. Le titre donne l'erreur pondérée $\varepsilon$ et le poids de vote $\alpha$ de chaque *stump*. Aucun *stump* seul ne sépare le disque, mais leur vote pondéré s'en approche. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** classification de demandes de crédit en « accordé / refusé » sur un jeu propre et bien étiqueté, comme référence de boosting simple.
- **Cas d'usage 2 :** détection de visages en temps réel (algorithme de Viola-Jones, application historique majeure d'AdaBoost).
- **Données et prérequis :** données tabulaires **propres** (étiquettes fiables, peu de valeurs aberrantes) ; classifieur faible légèrement meilleur que le hasard ; pas de standardisation avec des arbres.
- **Peu approprié :** données avec beaucoup d'étiquettes erronées — par exemple des retours clients classés manuellement de façon incohérente — car l'algorithme augmente sans cesse le poids de ces cas, ce qui dégrade le modèle.
- **Métriques :** F1, précision, rappel, ROC-AUC ; erreur de validation en fonction du nombre d'estimateurs.

```python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `n_estimators=70` *stumps* (arbres de profondeur 1 par défaut), `learning_rate=1.0`, `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,827** — mieux que la logistique (≈ 0,811) et Naive Bayes (≈ 0,755), mais nettement moins bien que le gradient boosting (≈ 0,950) et la forêt (≈ 0,918).
- **Conclusion possible :** des *stumps* ne modélisent chacun qu'une coupure sur une variable ; même combinés, ils capturent mal les interactions entre variables présentes dans ces données, alors que les arbres plus profonds du gradient boosting les captent.
- **Limites :** profondeur du classifieur faible et nombre d'estimateurs non réglés ; écart observé sur un seul découpage de 100 observations ; aucun test de sensibilité au bruit.

## 16. XGBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**XGBoost** (*eXtreme Gradient Boosting*) — classification supervisée : gradient boosting optimisé et régularisé.
**Étapes :** arbres ajoutés séquentiellement sur les erreurs, avec approximation au second ordre de la perte et pénalité sur la complexité des arbres.
$$\mathcal L=\textstyle\sum_i\ell(y_i,\hat y_i)+\sum_m\Omega(f_m),\qquad \Omega(f)=\gamma T+\tfrac12\lambda\sum_j w_j^2$$
$\ell$ : perte ; $f_m$ : arbre $m$ ; $T$ : nombre de feuilles ; $w_j$ : valeur de la feuille $j$ ; $\gamma,\lambda$ : pénalités.
**Paramètres appris :** arbres et $w_j$. **Hyperparamètres :** `n_estimators`, `max_depth`, `learning_rate`, $\gamma$, $\lambda$.
**Avantages :** très performant et rapide ; régularisation et valeurs manquantes gérées.
**Limites :** nombreux hyperparamètres ; peu interprétable.

### Schéma explicatif

**Comment lire le schéma :** trois modèles XGBoost entraînés sur les mêmes données. À gauche, des arbres profonds sans pénalité créent une frontière très irrégulière (îlots autour de points isolés). Au centre, une configuration modérée. À droite, des arbres peu profonds fortement pénalisés donnent une frontière lisse, au risque de sous-apprendre. La régularisation $\Omega$ est ce qui distingue XGBoost du boosting classique. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** modèles de scoring crédit ou de détection de fraude en production dans les banques et fintechs (grands volumes, variables nombreuses).
- **Cas d'usage 2 (logistique/commerce) :** prévision de la probabilité de rupture de stock ou de retard de livraison à partir de données opérationnelles.
- **Données et prérequis :** données tabulaires, valeurs manquantes tolérées ; recherche d'hyperparamètres (validation croisée ou jeu de validation avec `early_stopping_rounds`) ; installation de la bibliothèque `xgboost`.
- **Peu approprié :** problèmes sur images, son ou texte brut (réseaux de neurones plus adaptés) ; ou très petits échantillons avec exigence d'explication simple, où le gain par rapport à une logistique ne justifie pas la complexité.
- **Métriques :** F1, précision, rappel, ROC-AUC, PR-AUC ; log-loss (`eval_metric`) suivie au fil des itérations.

```python
from xgboost import XGBClassifier
modele=XGBClassifier(n_estimators=80,max_depth=3,random_state=GRAINE,eval_metric='logloss').fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

- **Hyperparamètres :** `n_estimators=80`, `max_depth=3`, `eval_metric='logloss'` (perte suivie), `learning_rate=0.3` et `reg_lambda=1` par défaut de la bibliothèque, `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,931**, deuxième meilleur résultat, juste derrière `HistGradientBoostingClassifier` (≈ 0,950) et devant la forêt (≈ 0,918).

- **Limites :** l'écart 0,931 / 0,950 correspond à environ deux erreurs de test et n'est pas interprétable comme une supériorité ; les hyperparamètres ne sont pas réglés ; l'intérêt de XGBoost (grands volumes) n'est pas testé ici.

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**K-moyennes** (*K-means*) — clustering : partitionner les données en $K$ groupes autour de centroïdes.
**Étapes :** $K$ centroïdes initiaux → affectation de chaque point au plus proche → recalcul des centroïdes (moyennes) → répétition jusqu'à stabilité.
$$\min\ \textstyle\sum_{k=1}^{K}\sum_{x_i\in C_k}\lVert x_i-\mu_k\rVert^2$$
$K$ : nombre de groupes ; $C_k$ : points du groupe $k$ ; $\mu_k$ : centroïde du groupe $k$.
**Paramètres appris :** $\mu_k$. **Hyperparamètres :** $K$ (`n_clusters`), `n_init`.
**Avantages :** simple et rapide ; centroïdes faciles à interpréter.
**Limites :** $K$ à fixer ; suppose des groupes sphériques, sensible aux valeurs extrêmes.

### Schéma explicatif

**Comment lire le schéma :** chaque panneau est une itération. Les points sont colorés selon le centroïde (croix rouge) le plus proche : c'est l'étape d'**affectation**. Les flèches montrent l'étape de **mise à jour** : chaque centroïde se déplace vers la moyenne de ses points. Partis d'une initialisation volontairement mauvaise, les centroïdes convergent en quelques itérations vers les centres des trois nuages, et les flèches deviennent de plus en plus courtes. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** segmentation de la clientèle à partir de variables RFM (récence, fréquence, montant des achats) pour adapter les offres — par ex. sur le jeu *Online Retail* (UCI).
- **Cas d'usage 2 (logistique) :** regrouper des points de livraison géographiquement proches pour définir des zones de tournée ou l'emplacement d'entrepôts.
- **Données et prérequis :** variables numériques **standardisées**, valeurs extrêmes traitées ; choix de $K$ par méthode du coude, silhouette ou connaissance métier.
- **Peu approprié :** groupes de formes allongées, imbriquées ou de densités très différentes (ex. trajectoires de transactions frauduleuses) — K-means coupe arbitrairement ces formes ; préférer DBSCAN ou mélanges gaussiens. Peu adapté aussi aux variables purement catégorielles (moyenne sans sens → K-modes).
- **Métriques :** silhouette, inertie (coude), indice de Davies-Bouldin ; stabilité entre graines ; ARI si des étiquettes réelles existent (évaluation externe uniquement).

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Classification hiérarchique ascendante** (*Agglomerative Clustering*, CAH) — clustering : construire une hiérarchie de groupes.
**Étapes :** chaque point est un groupe → fusion des deux groupes les plus proches → répétition → dendrogramme coupé au nombre de groupes voulu.
$$\Delta(A,B)=\frac{|A||B|}{|A|+|B|}\lVert\mu_A-\mu_B\rVert^2\ \ \text{(Ward)}$$
$A,B$ : groupes ; $|A|,|B|$ : effectifs ; $\mu_A,\mu_B$ : centres ; $\Delta$ : hausse d'inertie due à la fusion.
**Paramètres appris :** aucun (hiérarchie des fusions). **Hyperparamètres :** `n_clusters`, liaison, distance.
**Avantages :** $K$ non fixé d'avance ; dendrogramme interprétable.
**Limites :** coûteux ($O(n^2)$) ; fusions jamais remises en cause.

### Schéma explicatif

**Comment lire le schéma :** à droite, chaque feuille en bas est une observation (numéros identiques au nuage de gauche). Deux branches qui se rejoignent indiquent une fusion ; la **hauteur** de la jonction mesure le coût de la fusion (critère de Ward). Les très grands sauts de hauteur signalent que l'on fusionne des groupes vraiment différents. Couper l'arbre à la ligne rouge donne 3 groupes, coloriés de la même façon à gauche. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** regrouper des actions ou des fonds selon la corrélation de leurs rendements (taxonomie de portefeuille, *Hierarchical Risk Parity*).
- **Cas d'usage 2 (commerce international) :** construire une typologie de pays selon leurs indicateurs économiques (PIB par habitant, ouverture commerciale, inflation — données Banque mondiale) et lire les niveaux de proximité.
- **Données et prérequis :** variables numériques standardisées, **échantillon de taille modérée** (≤ quelques milliers de lignes) ; choix de la liaison et de la distance.
- **Peu approprié :** segmentation de plusieurs millions de clients d'un site e-commerce : la matrice de distances $n\times n$ ne tient pas en mémoire ; K-means ou MiniBatchKMeans sont plus adaptés.
- **Métriques :** silhouette, coefficient cophénétique (fidélité du dendrogramme aux distances), lecture des sauts de hauteur ; ARI si étiquettes réelles disponibles.

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

### Analyse de l'expérience Python
- Entrées : `X_clu` (240 × 2) généré par `make_blobs` avec 3 centres et `cluster_std=0.85` — trois groupes sphériques, bien séparés et de même taille. Données **synthétiques** ; les étiquettes de référence existent mais ne sont pas utilisées par le code.
- **Hyperparamètres :** `n_clusters=3`, liaison `ward` et distance euclidienne par défaut.
- **Sortie obtenue :** **silhouette ≈ 0,869**, valeur **identique** à K-means (exemple 17).
- **Conclusion possible :** une silhouette identique à toutes les décimales indique que les deux méthodes produisent la **même partition** (à la numérotation près). C'est logique : Ward minimise aussi l'inertie intra-classe, et des groupes aussi bien séparés ne laissent pas d'ambiguïté.
- **Limites :** le dendrogramme, principal intérêt de la méthode, n'est pas affiché dans l'exemple ; l'égalité avec K-means est propre à ce cas facile et ne se généralise pas ; pas de comparaison entre critères de liaison.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**DBSCAN** (*Density-Based Spatial Clustering of Applications with Noise*) — clustering par densité : trouver des groupes de forme libre et isoler le bruit.
**Étapes :** points cœurs = au moins `min_samples` voisins dans un rayon $\varepsilon$ → cœurs voisins reliés en groupes → points isolés = bruit (−1).
$$N_\varepsilon(x)=\{x':\ d(x,x')\le\varepsilon\},\qquad x\ \text{cœur}\iff|N_\varepsilon(x)|\ge\text{min\_samples}$$
$N_\varepsilon(x)$ : voisinage de rayon $\varepsilon$ ; $d$ : distance.
**Paramètres appris :** aucun (le nombre de groupes est un résultat). **Hyperparamètres :** $\varepsilon$ (`eps`), `min_samples`.
**Avantages :** formes quelconques, $K$ non requis ; détecte les points aberrants.
**Limites :** très sensible à $\varepsilon$ ; échoue si les densités diffèrent.

### Schéma explicatif

**Comment lire le schéma :** les deux croissants sont retrouvés sans préciser leur nombre, ce que K-means ne saurait pas faire (il couperait les croissants en deux par une droite). Les ronds sont des **points cœurs** (≥ 5 voisins dans le cercle de rayon $\varepsilon$, dessiné en pointillé pour un point), les triangles des **points frontières** rattachés à un groupe, et les croix noires le **bruit**, isolé hors des zones denses. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique/commerce) :** repérer des zones de forte concentration de clients ou de livraisons à partir de coordonnées GPS (formes irrégulières suivant les rues), en ignorant les adresses isolées.
- **Cas d'usage 2 (finance) :** détecter des transactions atypiques qui n'appartiennent à aucun groupe dense de comportements habituels (étiquette −1 = alerte à examiner).
- **Données et prérequis :** variables numériques standardisées, dimension faible à moyenne ; choix de $\varepsilon$ via le graphe des distances au k-ième voisin.
- **Peu approprié :** segmentation de clients où un groupe premium très dispersé côtoie un groupe grand public très dense : un seul $\varepsilon$ ne convient pas aux deux densités (l'un est fragmenté ou l'autre absorbé) → HDBSCAN ou mélanges gaussiens. Peu adapté aussi en grande dimension.
- **Métriques :** nombre de groupes et proportion de bruit, silhouette calculée **hors bruit**, stabilité selon $\varepsilon$ ; ARI si étiquettes disponibles.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

### Analyse de l'expérience Python
- Entrées : `X_clu` (240 × 2) généré par `make_blobs` avec 3 centres et `cluster_std=0.85` — trois groupes sphériques, bien séparés et de même taille. Données **synthétiques** ; les étiquettes de référence existent mais ne sont pas utilisées par le code.
- **Hyperparamètres :** `eps=0.8`, `min_samples=5`.
- **Sortie obtenue :** étiquettes `{0, 1, 2, -1}` → **3 groupes** retrouvés et **8 points** classés comme bruit (≈ 3,3 % des 240 points).
- **Conclusion possible :** sans connaître le nombre de groupes, DBSCAN retrouve bien les 3 nuages. Les 8 points de bruit sont des points situés en périphérie des nuages gaussiens, où la densité passe sous le seuil (< 5 voisins dans un rayon de 0,8) : ce ne sont pas des anomalies « réelles », puisque toutes les données ont été générées par les trois groupes.
- **Limites :** aucun indicateur de qualité n'est calculé (pas de silhouette hors bruit) ; le résultat dépend fortement de `eps` — une valeur plus petite fragmenterait les groupes, une plus grande les fusionnerait ; le nombre de points de bruit n'a de sens que relativement à ces réglages.

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Mélange gaussien** (*Gaussian Mixture Model*, GMM) — clustering probabiliste : modéliser les données comme un mélange de $K$ lois normales.
**Étapes :** algorithme EM — étape E : probabilité d'appartenance de chaque point à chaque composante ; étape M : réestimation des paramètres ; répétition.
$$p(x)=\textstyle\sum_{k=1}^{K}\pi_k\,\mathcal N(x\mid\mu_k,\Sigma_k)$$
$\pi_k$ : poids ; $\mu_k$ : moyenne ; $\Sigma_k$ : covariance de la composante $k$ ; $\mathcal N$ : loi normale.
**Paramètres appris :** $\pi_k,\mu_k,\Sigma_k$. **Hyperparamètres :** $K$ (`n_components`), `covariance_type`.
**Avantages :** appartenance probabiliste ; groupes elliptiques de tailles différentes.
**Limites :** suppose des groupes gaussiens, $K$ à fixer ; sensible à l'initialisation.

### Schéma explicatif

**Comment lire le schéma :** chaque composante gaussienne est représentée par deux ellipses (1 et 2 écarts-types) dont la forme et l'orientation viennent de sa matrice de covariance — ici un groupe incliné et un groupe vertical, ce que K-means ne saurait représenter. La couleur de chaque point donne sa **probabilité** d'appartenir à la composante 1 : bleu/rouge francs = appartenance sûre, teintes intermédiaires = zone de recouvrement. Les poids $\pi$ indiquent la part estimée de chaque groupe. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** identifier des **régimes de marché** (calme / volatil) à partir des rendements et de la volatilité d'un indice, avec une probabilité d'être dans chaque régime à chaque date.
- **Cas d'usage 2 (marketing) :** segmentation douce de clients lorsqu'un client peut partager les caractéristiques de plusieurs segments (probabilités d'appartenance utilisées pour personnaliser les offres).
- **Données et prérequis :** variables numériques continues, standardisées, groupes approximativement elliptiques ; suffisamment d'observations par composante pour estimer les covariances ; choix de $K$ par BIC.
- **Peu approprié :** données à forte dimension et peu d'observations (ex. 500 variables pour 200 clients) : les matrices de covariance ne sont pas estimables de façon fiable ; ou groupes de formes très non gaussiennes (croissants, anneaux).
- **Métriques :** BIC/AIC, log-vraisemblance, silhouette (sur l'affectation la plus probable), entropie des responsabilités ; ARI si étiquettes réelles disponibles.

```python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
```

### Analyse de l'expérience Python
- Entrées : `X_clu` (240 × 2) généré par `make_blobs` avec 3 centres et `cluster_std=0.85` — trois groupes sphériques, bien séparés et de même taille. Données **synthétiques** ; les étiquettes de référence existent mais ne sont pas utilisées par le code.
- **Hyperparamètres :** `n_components=3`, `covariance_type='full'` par défaut, `random_state=42`.
- **Sortie obtenue :** poids estimés **[0,333 ; 0,333 ; 0,333]** et **silhouette ≈ 0,869** (sur l'affectation la plus probable).
- **Conclusion possible :** le modèle estime trois composantes de poids égaux (≈ 80 points chacune sur 240), ce qui correspond exactement à la génération par `make_blobs` (groupes de même taille). La silhouette identique à K-means et à la CAH montre que les trois méthodes donnent la même partition sur ce cas idéal.
- **Limites :** les probabilités d'appartenance, principal apport du GMM, ne sont pas exploitées dans la sortie ; $K$ n'a pas été choisi par BIC ; avec des groupes aussi séparés, l'avantage des covariances complètes n'apparaît pas.

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Analyse en composantes principales** (*Principal Component Analysis*, ACP) — réduction de dimension linéaire : résumer des variables corrélées en quelques composantes.
**Étapes :** variables standardisées → directions de variance maximale (vecteurs propres de la covariance) → projection sur les $k$ premières.
$$\Sigma v_j=\lambda_j v_j,\qquad \text{variance expliquée}_j=\lambda_j/\textstyle\sum_l\lambda_l$$
$\Sigma$ : matrice de covariance ; $v_j$ : axe principal $j$ ; $\lambda_j$ : variance portée par cet axe.
**Paramètres appris :** les axes $v_j$. **Hyperparamètre :** $k$ (`n_components`).
**Avantages :** rapide, déterministe, axes interprétables ; réduit bruit et colinéarité.
**Limites :** ne capte que le linéaire ; sensible à l'échelle et aux valeurs extrêmes.

### Schéma explicatif

**Comment lire le schéma :** à gauche, sur un nuage 2D simulé, la flèche CP1 indique la direction où les points sont le plus étalés et CP2 la direction perpendiculaire ; la longueur des flèches est proportionnelle à l'écart-type porté par chaque axe. À droite, sur les vraies données `load_wine`, les barres donnent la part de variance de chacune des 13 composantes et la courbe rouge la part cumulée : on lit combien de composantes il faut garder pour conserver, par exemple, 80 % de l'information.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** résumer les mouvements de la courbe des taux (10 maturités) en trois facteurs — niveau, pente, courbure — ou extraire des facteurs de risque communs à un ensemble d'actions.
- **Cas d'usage 2 (gestion/marketing) :** synthétiser un questionnaire de satisfaction de 30 questions en quelques dimensions interprétables, ou construire un indice composite de développement à partir d'indicateurs Banque mondiale.
- **Données et prérequis :** variables numériques corrélées, **standardisées** (sinon les variables à grande échelle dominent), valeurs extrêmes traitées ; ACP ajustée sur l'entraînement seulement si elle précède un modèle supervisé.
- **Peu approprié :** données où la structure est non linéaire (ex. segments de clients disposés en anneaux ou en spirale) : l'ACP les écrase ; préférer t-SNE/UMAP pour la visualisation. Ou quand l'information discriminante pour une cible se trouve dans une direction de faible variance.
- **Métriques :** part de variance expliquée (individuelle et cumulée), erreur de reconstruction, qualité de représentation (cos²) des individus.

```python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
```

### Analyse de l'expérience Python
- **Entrées :** jeu `load_wine` (178 vins, 13 mesures chimiques), seul jeu réel du notebook — mais il s'agit d'un jeu pédagogique intégré à scikit-learn, sans lien avec une problématique économique. Standardisation sur l'ensemble des données (acceptable ici car aucun modèle supervisé n'est évalué ensuite).
- **Hyperparamètre :** `n_components=2`.
- **Sortie obtenue :** variance expliquée **[0,362 ; 0,192]** → la CP1 conserve ≈ 36,2 % de la variance, la CP2 ≈ 19,2 %, soit **≈ 55,4 %** au total ; et le nuage des 178 vins projetés.
- **Conclusion possible :** la projection 2D résume un peu plus de la moitié de l'information des 13 variables ; elle est suffisante pour une première visualisation mais perd ≈ 45 % de la variance — les distances observées sur le graphique doivent donc être interprétées avec prudence.
- **Limites :** le graphique n'est pas coloré par cépage et n'a pas de titre ni d'axes nommés : on ne peut pas juger visuellement la séparation des classes ; les coefficients des composantes (`components_`) ne sont pas affichés, donc les axes ne sont pas interprétés.

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**t-SNE** (*t-distributed Stochastic Neighbor Embedding*) — réduction de dimension non linéaire, pour la visualisation en 2D.
**Étapes :** similarités de voisinage $p_{ij}$ dans l'espace d'origine → points placés en 2D avec similarités $q_{ij}$ → ajustement pour que $q$ ressemble à $p$.
$$\mathrm{KL}(P\|Q)=\textstyle\sum_{i\neq j}p_{ij}\ln\frac{p_{ij}}{q_{ij}}$$
$p_{ij}$ : similarité d'origine ; $q_{ij}$ : similarité en 2D (loi de Student) ; KL : divergence minimisée.
**Paramètres appris :** les coordonnées 2D. **Hyperparamètres :** `perplexity`, `learning_rate`, `random_state`.
**Avantages :** révèle très bien les groupes non linéaires ; visualisation parlante.
**Limites :** distances entre groupes non interprétables ; lent, non déterministe, pas de projection de nouveaux points.

### Schéma explicatif

**Comment lire le schéma :** chaque point est un vin, coloré par son cépage réel — information **non fournie à l'algorithme**, ajoutée uniquement pour juger la carte. Le premier panneau montre l'ACP pour comparaison. Avec une perplexité faible (3), la carte est plus morcelée et des points d'un même cépage se dispersent en petits îlots (structure trop locale) ; avec 20, les trois cépages forment des groupes nets ; avec 60, la carte se rapproche d'une vue plus globale et les frontières entre groupes se resserrent. Seuls les **voisinages** sont fiables : la distance entre deux groupes ou leur taille apparente ne doivent pas être interprétées.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** visualiser des milliers de clients décrits par des dizaines de variables comportementales pour repérer visuellement des segments avant une classification formelle.
- **Cas d'usage 2 (finance) :** explorer des transactions ou des entreprises décrites par de nombreux ratios afin de voir si les cas frauduleux ou défaillants forment des îlots distincts.
- **Données et prérequis :** variables numériques standardisées, souvent après une ACP préalable (≈ 50 composantes) pour réduire le bruit et le coût ; quelques centaines à quelques dizaines de milliers de points.
- **Peu approprié :** comme étape de prétraitement d'un modèle de prédiction en production (pas de projection de nouvelles observations, résultat non stable) ou pour mesurer des distances économiques entre segments (les distances globales sont déformées).
- **Métriques :** *trustworthiness* (conservation des voisins), divergence KL finale ; évaluation visuelle de séparation avec étiquettes externes ; stabilité selon graine et perplexité.

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

### Analyse de l'expérience Python
- **Entrées :** `load_wine` standardisé (178 × 13).
- **Hyperparamètres :** `n_components=2`, `perplexity=20`, `random_state=42` (pour la reproductibilité, t-SNE étant stochastique).
- **Sortie obtenue :** uniquement un **nuage de points 2D**, sans métrique numérique.
- **Conclusion possible :** le graphique montre si des groupes de vins se distinguent visuellement dans les 13 variables chimiques (dans notre schéma coloré, trois groupes correspondant aux cépages apparaissent nettement).
- **Ce qu'on ne peut pas conclure :** aucune métrique n'est imprimée — on ne peut donc pas quantifier la qualité de la projection (pas de *trustworthiness*) ; le graphique du code n'étant pas coloré, on ne peut pas vérifier que les groupes visibles correspondent aux cépages ; distances et tailles entre groupes ne sont pas interprétables.

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**UMAP** (*Uniform Manifold Approximation and Projection*) — réduction de dimension non linéaire : visualiser en préservant structure locale et en partie globale.
**Étapes :** graphe des $k$ plus proches voisins → représentation 2D initiale → optimisation pour que le graphe 2D ressemble au graphe d'origine.
$$\mathcal C=\textstyle\sum_{(i,j)}\Big[w_{ij}\ln\frac{w_{ij}}{v_{ij}}+(1-w_{ij})\ln\frac{1-w_{ij}}{1-v_{ij}}\Big]$$
$w_{ij}$ : poids de voisinage d'origine ; $v_{ij}$ : poids en 2D ; $\mathcal C$ : entropie croisée minimisée.
**Paramètres appris :** les coordonnées 2D. **Hyperparamètres :** `n_neighbors`, `min_dist`, `n_components`.
**Avantages :** plus rapide que t-SNE ; meilleure structure globale, projette de nouveaux points.
**Limites :** sensible aux hyperparamètres ; axes sans signification.

### Schéma explicatif

**Comment lire le schéma :** chaque panneau est une projection UMAP des 178 vins avec un nombre de voisins différent ; la couleur (cépage) sert uniquement à l'interprétation. Avec `n_neighbors=3`, l'algorithme ne regarde que des voisinages minuscules et la carte se fragmente ; avec 15 (valeur du notebook) ou 80, les trois cépages apparaissent comme des groupes cohérents et leur position relative devient plus stable. C'est le réglage principal entre fidélité locale et vision d'ensemble.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/commerce) :** cartographier les produits d'un catalogue à partir de leurs caractéristiques ou des co-achats pour repérer des familles de produits et des produits « à la frontière ».
- **Cas d'usage 2 (finance) :** réduire des centaines de variables comportementales de clients bancaires à quelques dimensions avant un clustering (UMAP puis HDBSCAN), ou visualiser la structure d'un portefeuille de crédits.
- **Données et prérequis :** variables numériques standardisées (ou métrique adaptée), quelques centaines à plusieurs millions de points ; bibliothèque `umap-learn` ; graine fixée.
- **Peu approprié :** quand il faut des axes **interprétables** économiquement (ex. présenter au comité de direction ce que signifie l'axe 1) : les dimensions UMAP n'ont pas de signification, contrairement aux composantes de l'ACP ; ou sur très petit échantillon (< 50 points) où le graphe de voisins est peu fiable.
- **Métriques :** *trustworthiness* / continuité, conservation des k plus proches voisins, stabilité selon graine et `n_neighbors` ; silhouette ou ARI si un clustering est réalisé ensuite.

```python
import umap
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=umap.UMAP(n_components=2,n_neighbors=15,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('UMAP');plt.show()
```

### Analyse de l'expérience Python
- **Entrées :** `load_wine` standardisé (178 × 13).
- **Hyperparamètres :** `n_components=2`, `n_neighbors=15`, `min_dist=0.1` par défaut, `random_state=42`. Un avertissement indique que fixer `random_state` désactive le calcul parallèle (`n_jobs=1`) : c'est le prix de la reproductibilité.
- **Sortie obtenue :** un **nuage de points 2D** intitulé « UMAP », sans métrique numérique.
- **Conclusion possible :** comme pour t-SNE, le graphique permet une exploration visuelle de la structure des 13 variables ; notre schéma coloré montre que les cépages forment des groupes distincts.
- **Ce qu'on ne peut pas conclure :** aucune mesure de conservation des voisinages n'est calculée ; sans coloration par cépage dans la cellule du code, la correspondance avec les classes réelles n'est pas vérifiable ; on ne peut pas classer UMAP et t-SNE sans métrique commune.

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Forêt d'isolement** (*Isolation Forest*) — détection d'anomalies non supervisée : une anomalie s'isole en peu de coupures.
**Étapes :** arbres à coupures aléatoires → profondeur moyenne nécessaire pour isoler chaque point → score → les plus atypiques (part `contamination`) = −1.
$$s(x,n)=2^{-E[h(x)]/c(n)}$$
$h(x)$ : nombre de coupures pour isoler $x$ ; $E[h(x)]$ : moyenne sur les arbres ; $c(n)$ : normalisation ; $s$ proche de 1 = anomalie.
**Paramètres appris :** les arbres aléatoires. **Hyperparamètres :** `n_estimators`, `max_samples`, `contamination`.
**Avantages :** rapide, sans étiquettes ; aucune hypothèse de distribution.
**Limites :** seuil dépendant de `contamination` ; rate les anomalies proches d'un groupe dense.

### Schéma explicatif

**Comment lire le schéma :** chaque ligne bleue est une coupure aléatoire (verticale ou horizontale) d'un arbre d'isolement. À gauche, le point rouge éloigné du nuage se retrouve seul après **très peu de coupures** ; à droite, un point au cœur du nuage en demande beaucoup plus. L'algorithme répète cette expérience sur de nombreux arbres et transforme la longueur moyenne du chemin en score d'anomalie (titre). Données simulées, un seul arbre illustré.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/audit) :** repérer des écritures comptables ou des transactions bancaires inhabituelles (montant, heure, fréquence, contrepartie) pour prioriser les contrôles d'audit ou de lutte anti-fraude.
- **Cas d'usage 2 (logistique/marchés) :** détecter des journées anormales sur un indice boursier (rendement, volume, volatilité) ou des délais de livraison aberrants dans une chaîne d'approvisionnement.
- **Données et prérequis :** variables numériques (catégorielles encodées), pas besoin de standardisation stricte ; estimation de la proportion d'anomalies attendue ; un expert pour valider les alertes.
- **Peu approprié :** fraude sophistiquée qui imite le comportement normal (anomalies locales noyées dans un groupe dense) ou séries où l'anomalie est **contextuelle** (une vente élevée normale en décembre mais anormale en août) sans variables de contexte : l'algorithme ne la verrait pas.
- **Métriques :** sans étiquettes : distribution des scores, nombre et qualité des alertes jugées par un expert, stabilité entre graines ; avec étiquettes fiables : précision, rappel, PR-AUC.

```python
from sklearn.ensemble import IsolationForest
X_anomal=np.vstack([X_clu,np.array([[9,9],[-8,-8]])])
modele=IsolationForest(contamination=.02,random_state=GRAINE).fit(X_anomal)
print('Observations atypiques:',np.where(modele.predict(X_anomal)==-1)[0])
```

### Analyse de l'expérience Python
- **Entrées :** les 240 points de `X_clu` + **2 points ajoutés manuellement** : (9 ; 9) d'indice 240 et (−8 ; −8) d'indice 241 — 242 observations synthétiques.
- **Hyperparamètres :** `contamination=0.02` (≈ 2 % × 242 ≈ 5 points signalés), `n_estimators=100` par défaut, `random_state=42`.
- **Sortie obtenue :** observations atypiques **[20, 79, 130, 147, 240]**.
- **Conclusion possible :** le point (9 ; 9) (indice 240), très éloigné de tous les groupes, est bien détecté. En revanche le point (−8 ; −8) (indice 241) **n'est pas signalé** : il se trouve près du groupe centré vers (−6,9 ; −6,9) et ne s'isole donc pas facilement. Les 4 autres alertes (indices 20, 79, 130, 147) sont des points des groupes d'origine situés en bordure : ils sont signalés uniquement parce que `contamination` impose environ 5 alertes.
- **Limites :** le résultat illustre que le seuil est fixé par `contamination` et non par les données ; il n'y a pas d'étiquettes fiables pour calculer précision/rappel (on « sait » seulement que 2 points ont été ajoutés, et l'un d'eux n'est pas anormal au sens de l'algorithme) ; les scores eux-mêmes ne sont pas affichés.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Bandits manchots** (*Multi-Armed Bandits*) — renforcement sans état : choisir répétitivement entre plusieurs actions aux gains inconnus.
**Étapes :** choix d'un bras → récompense observée → mise à jour de sa valeur estimée → arbitrage exploration/exploitation (ici : tour de rôle, pure exploration).
$$\hat Q(a)=\frac{\text{somme des gains de }a}{N(a)}$$
$a$ : bras ; $N(a)$ : nombre de fois joué ; $\hat Q(a)$ : gain moyen estimé.
**Paramètres appris :** $\hat Q(a)$. **Hyperparamètres :** horizon (nombre de tours), règle de choix.
**Avantages :** cadre simple de décision sous incertitude ; apprend en ligne, plus efficace qu'un test A/B.
**Limites :** ignore l'état et les effets futurs ; suppose des gains stables.

### Schéma explicatif

**Comment lire le schéma :** à gauche, les courbes pleines sont les estimations $\hat Q(a)$ de chaque bras quand on les joue à tour de rôle, et les pointillés leurs vraies probabilités (0,2 / 0,5 / 0,7) : les estimations se stabilisent autour des vraies valeurs. À droite, le **regret cumulé** (gain perdu par rapport au meilleur bras) augmente en ligne droite : en jouant les mauvais bras deux tiers du temps, on apprend bien mais on gagne mal. C'est précisément ce que les stratégies comme $\varepsilon$-greedy corrigent. Simulation.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** choisir en continu, parmi plusieurs versions d'une bannière publicitaire ou d'un objet d'email, celle qui génère le plus de clics, en réallouant le trafic au fil des résultats.
- **Cas d'usage 2 (finance/commerce) :** tester plusieurs niveaux de prix ou de remise en ligne (*dynamic pricing*) ; allocation de budget entre canaux d'acquisition.
- **Données et prérequis :** pas de dataset fixe : données générées par l'interaction (récompense immédiate observable après chaque choix) ; nombre de bras limité ; environnement à peu près stationnaire.
- **Peu approprié :** décisions dont l'effet est différé ou dépend d'un état (ex. gestion d'un stock où la commande d'aujourd'hui modifie la situation de demain) → il faut un cadre de renforcement complet (Q-learning). Peu adapté aussi si la récompense n'est observée qu'après des mois (octroi de crédit et défaut).
- **Métriques :** récompense cumulée, regret cumulé, proportion de choix du meilleur bras, précision des estimations ; variation entre graines.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

### Analyse de l'expérience Python
- **Entrées :** environnement **simulé** de 3 bras de Bernoulli de probabilités réelles 0,2 / 0,5 / 0,7 (gain 1 ou 0), générateur fixé par `GRAINE=42`. Ce n'est pas un dataset réel.
- **Hyperparamètres :** horizon de 500 tours ; politique de **tour de rôle** (`bras = t % 3`), soit ≈ 167 tirages par bras.
- **Sortie obtenue :** valeurs estimées **[0,180 ; 0,539 ; 0,723]**, proches des vraies valeurs [0,2 ; 0,5 ; 0,7] (écarts de 0,02 à 0,04).
- **Conclusion possible :** avec ≈ 167 essais par bras, la moyenne empirique estime correctement chaque probabilité et le classement des bras est le bon (bras 2 meilleur).
- **Ce qu'on ne peut pas conclure :** cette politique n'**apprend pas à décider** : elle n'exploite jamais l'information, donc le gain cumulé n'est pas optimisé (environ deux tiers des tirages sur des bras sous-optimaux). Le gain et le regret ne sont pas affichés, et une seule graine ne mesure pas la variabilité.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**ε-greedy** (*Epsilon-greedy*) — renforcement : règle d'équilibre exploration/exploitation.
**Étapes :** avec probabilité $\varepsilon$, bras au hasard (exploration) ; sinon, meilleur bras estimé (exploitation) → mise à jour de sa valeur.
$$a_t=\begin{cases}\text{aléatoire} & \text{prob. } \varepsilon\\ \arg\max_a\hat Q(a) & \text{prob. } 1-\varepsilon\end{cases}$$
$\varepsilon$ : taux d'exploration ; $\hat Q(a)$ : valeur estimée du bras $a$.
**Paramètres appris :** $\hat Q(a)$. **Hyperparamètres :** $\varepsilon$, nombre de tours.
**Avantages :** très simple ; explore toujours un minimum.
**Limites :** explore à l'aveugle, même les mauvais bras ; dépend fortement de $\varepsilon$.

### Schéma explicatif

**Comment lire le schéma :** chaque courbe est la moyenne sur 20 simulations indépendantes pour une valeur de $\varepsilon$. À gauche, le gain moyen par tour (l'optimum est 0,7 si l'on jouait toujours le meilleur bras) ; à droite, la proportion de tours où le meilleur bras est joué. $\varepsilon=0,5$ explore trop et plafonne bas ; $\varepsilon=0$ peut rester bloqué sur un bras sous-optimal dans certaines simulations ; une valeur intermédiaire offre en général le meilleur compromis. Simulation.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/e-commerce) :** recommandation de produits ou de contenus : afficher la plupart du temps l'article le plus performant et, de temps en temps, un article nouveau pour mesurer son potentiel.
- **Cas d'usage 2 (logistique/finance) :** choix répété d'un transporteur ou d'un fournisseur parmi plusieurs (coût et délai incertains), ou politique d'exploration dans un agent Q-learning de gestion de stock.
- **Données et prérequis :** environnement interactif avec retour rapide de la récompense ; actions en nombre limité ; choix et éventuelle décroissance de $\varepsilon$.
- **Peu approprié :** situations où une exploration au hasard est coûteuse ou risquée — par ex. tester au hasard un taux d'intérêt très défavorable sur de vrais clients bancaires, ou un dosage en santé : on préfère des méthodes d'exploration dirigée (UCB, Thompson sampling) ou des simulations.
- **Métriques :** récompense cumulée et regret, part des choix optimaux, variance entre graines, vitesse de convergence.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Analyse de l'expérience Python
- **Entrées :** même bandit **simulé** à 3 bras (0,2 / 0,5 / 0,7), graine 42.
- **Hyperparamètres :** $\varepsilon=0,1$ fixe, 1 000 tours, valeur d'exploitation $(S+1)/(N+1)$.
- **Sortie obtenue :** nombre de choix **[25 ; 49 ; 926]** et valeurs estimées **[0,200 ; 0,673 ; 0,693]**.
- **Conclusion possible :** l'agent a identifié le meilleur bras et l'a joué **92,6 %** des tours (contre 33 % pour le tour de rôle de l'exemple 25) : le gain cumulé est donc bien meilleur. Les ≈ 74 choix sur les autres bras proviennent essentiellement de l'exploration aléatoire (≈ 10 % de 1 000 = 100 tirages aléatoires, dont environ deux tiers tombent sur un autre bras).
- **Ce qu'on ne peut pas conclure :** les bras peu joués sont **mal estimés** — le bras 1 est estimé à 0,673 au lieu de 0,5 avec seulement 49 tirages : exploiter réduit la précision sur les options délaissées. Une seule graine est utilisée, donc la robustesse n'est pas évaluée ; le gain total n'est pas affiché.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Q-learning** — renforcement sans modèle, hors-politique : apprendre la valeur de chaque couple (état, action).
**Étapes :** état $s$ → action ε-greedy → récompense $r$ et état $s'$ → mise à jour de $Q(s,a)$ → politique = meilleure action par état.
$$Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\big]$$
$s,s'$ : états ; $a$ : action ; $r$ : récompense ; $\alpha$ : taux d'apprentissage ; $\gamma$ : actualisation du futur.
**Paramètres appris :** la table $Q$. **Hyperparamètres :** $\alpha$, $\gamma$, $\varepsilon$, nombre d'épisodes.
**Avantages :** converge vers la politique optimale sans connaître l'environnement ; simple.
**Limites :** impraticable avec beaucoup d'états (→ DQN) ; lent si récompenses rares.

### Schéma explicatif

**Comment lire le schéma :** à gauche, le cycle du renforcement : l'agent envoie une action, l'environnement renvoie le nouvel état et une récompense, et l'agent corrige sa table $Q$ avec la règle encadrée. À droite, la grille FrozenLake (états numérotés de 0 à 15) avec, dans chaque case praticable, une flèche indiquant l'action de plus grande valeur apprise. En suivant les flèches depuis la case 0 on doit atteindre la case verte (15) en évitant les trous gris. Simulation Gymnasium, mêmes réglages que l'exemple.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique/gestion) :** politique de réapprovisionnement d'un stock : état = niveau de stock et demande récente, action = quantité commandée, récompense = marge − coûts de stockage et de rupture.
- **Cas d'usage 2 (finance) :** stratégie d'exécution d'un gros ordre de bourse découpé dans le temps (état discrétisé : quantité restante, temps restant ; action : quantité à vendre) — dans un simulateur de marché.
- **Données et prérequis :** un **environnement** (réel ou simulé) avec états et actions **discrets et peu nombreux**, récompenses définies ; beaucoup d'épisodes ; réglage de $\alpha$, $\gamma$, $\varepsilon$.
- **Peu approprié :** gestion de portefeuille avec des états continus de grande dimension (prix, volatilités, positions de 50 actifs) : la table Q serait gigantesque et presque jamais remplie → approximation par réseau (DQN). Inadapté aussi lorsque l'on ne peut pas expérimenter (pas de simulateur, erreurs coûteuses).
- **Métriques :** récompense cumulée par épisode, taux de réussite (part d'épisodes atteignant le but), nombre de pas, vitesse de convergence, variation entre graines.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python
- **Entrées :** Environnement **FrozenLake-v1** de Gymnasium, version déterministe (`is_slippery=False`) : grille 4 × 4 (états 0 à 15, départ en 0, but en 15, trous en 5, 7, 11 et 12), 4 actions (0 = gauche, 1 = bas, 2 = droite, 3 = haut), récompense 1 uniquement en atteignant le but. Ce sont des trajectoires **simulées par interaction**, pas un dataset réel.
- **Hyperparamètres :** 4 000 épisodes, ≤ 100 pas par épisode, $\alpha=0,3$, $\gamma=0,95$, $\varepsilon$ décroissant linéairement de 1 à 0,05 sur 2 500 épisodes, graines fixées.
- **Sortie obtenue :** politique `[1 2 1 0 1 0 1 0 2 1 1 0 0 2 2 0]` (action choisie dans chaque état 0 à 15).
- **Conclusion possible :** en suivant la politique depuis le départ : 0 →(bas) 4 →(bas) 8 →(droite) 9 →(bas) 13 →(droite) 14 →(droite) **15 = but**. Ce chemin de 6 pas (le minimum possible) évite tous les trous : l'agent a appris une politique optimale. Les valeurs dans les états 5, 7, 11, 12 (trous) et 15 (but) valent 0 car ce sont des états terminaux où aucune action n'est apprise — l'`argmax` renvoie 0 par défaut et n'a pas de sens.
- **Ce qu'on ne peut pas conclure :** aucun taux de réussite ni récompense cumulée n'est mesuré ; l'environnement non glissant est déterministe et très petit — le succès ne prouve rien sur la version glissante (`is_slippery=True`) ni sur un problème réel ; une seule graine.

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**SARSA** (*State–Action–Reward–State–Action*) — renforcement sans modèle, sur-politique : apprendre la valeur de la politique réellement suivie.
**Étapes :** comme le Q-learning, mais la cible utilise l'action suivante **réellement choisie** $a'$ et non la meilleure.
$$Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\,Q(s',a')-Q(s,a)\big]$$
$s,a$ : état et action ; $r$ : récompense ; $s',a'$ : état et action suivants ; $\alpha$, $\gamma$ : comme en Q-learning.
**Paramètres appris :** la table $Q$. **Hyperparamètres :** $\alpha$, $\gamma$, $\varepsilon$.
**Avantages :** plus prudent quand l'exploration est risquée ; mises à jour plus stables.
**Limites :** optimal seulement si $\varepsilon\to0$ ; limites tabulaires du Q-learning.

### Schéma explicatif

**Comment lire le schéma :** à gauche, depuis l'état $s'$ trois actions sont possibles. Le Q-learning met à jour $Q(s,a)$ avec la **meilleure** valeur (0,9, en rouge), même si l'agent va en réalité explorer ; SARSA utilise la valeur de l'action **réellement jouée** (0,4, en bleu) — il apprend donc la valeur de son propre comportement, exploration incluse (valeurs illustratives). À droite, le taux de réussite des épisodes au cours de l'entraînement sur FrozenLake, moyenné sur 5 graines : il augmente à mesure que $\varepsilon$ diminue. Simulation Gymnasium.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique) :** apprentissage en ligne d'une politique de routage d'entrepôt ou de robot de préparation où une erreur d'exploration est coûteuse (collision, retard) : SARSA tend à apprendre des itinéraires plus sûrs que le Q-learning.
- **Cas d'usage 2 (finance/gestion) :** politique de relance de clients en retard de paiement apprise sur le système réel, où l'on veut évaluer la stratégie effectivement appliquée (avec sa part d'essais) plutôt qu'une stratégie idéale.
- **Données et prérequis :** environnement interactif à états et actions discrets ; l'apprentissage doit se faire **avec la politique utilisée** (pas sur des données collectées par une autre politique) ; décroissance de $\varepsilon$.
- **Peu approprié :** apprendre à partir d'un historique existant collecté par une autre stratégie (ex. journal de décisions passées de gestionnaires) : SARSA est sur-politique et exige des données générées par sa propre politique → préférer le Q-learning ou des méthodes hors-ligne.
- **Métriques :** récompense cumulée et taux de réussite **pendant** et **après** l'apprentissage, risque (nombre de chutes/échecs), variation entre graines.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python
- **Entrées :** Environnement **FrozenLake-v1** de Gymnasium, version déterministe (`is_slippery=False`) : grille 4 × 4 (états 0 à 15, départ en 0, but en 15, trous en 5, 7, 11 et 12), 4 actions (0 = gauche, 1 = bas, 2 = droite, 3 = haut), récompense 1 uniquement en atteignant le but. Ce sont des trajectoires **simulées par interaction**, pas un dataset réel.
- **Hyperparamètres :** identiques à l'exemple 27 (4 000 épisodes, $\alpha=0,3$, $\gamma=0,95$, $\varepsilon$ décroissant de 1 à 0,05), ce qui rend la comparaison équitable.
- **Sortie obtenue :** politique `[1 0 3 0 1 0 1 0 2 1 1 0 0 2 2 0]`.
- **Conclusion possible :** depuis l'état 0, la politique suit **le même chemin que le Q-learning** : 0 → 4 → 8 → 9 → 13 → 14 → 15 (bas, bas, droite, bas, droite, droite), chemin optimal de 6 pas. Les seules différences portent sur les états 1 et 2 (actions 0 et 3 au lieu de 2 et 1), qui ne sont **pas sur le chemin optimal** et ont été peu visités en fin d'apprentissage ; leur valeur importe peu pour la politique suivie depuis le départ.
- **Ce qu'on ne peut pas conclure :** dans un environnement déterministe et sans risque après convergence, la différence de prudence entre SARSA et Q-learning ne peut pas apparaître ; aucune récompense cumulée ni taux de réussite n'est mesuré ; une seule graine.

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Deep Q-Network** (*DQN*) — renforcement profond : Q-learning où la table $Q$ est remplacée par un réseau de neurones.
**Étapes :** transitions stockées en mémoire de rejeu → mini-lots aléatoires → cible calculée avec un réseau cible figé → mise à jour du réseau par gradient.
$$y=r+\gamma(1-d)\max_{a'}Q(s',a';\theta^-),\qquad L=(y-Q(s,a;\theta))^2$$
$y$ : cible ; $d$ : 1 si état final ; $\theta$ : poids du réseau ; $\theta^-$ : poids du réseau cible ; $L$ : perte.
**Paramètres appris :** $\theta$. **Hyperparamètres :** architecture, $\gamma$, taux d'apprentissage, taille de mémoire, fréquence de copie.
**Avantages :** gère des états continus ou nombreux ; mémoire et réseau cible stabilisent l'apprentissage.
**Limites :** instable et coûteux ; actions discrètes seulement, peu interprétable.

### Schéma explicatif

**Comment lire le schéma :** en haut, le réseau en ligne choisit les actions ($\varepsilon$-greedy) ; chaque transition observée est stockée dans la mémoire de rejeu, d'où l'on tire des mini-lots au hasard. En bas, la cible est calculée avec le **réseau cible**, une copie figée du réseau en ligne mise à jour seulement tous les $C$ pas : sans lui, la cible bougerait à chaque mise à jour et l'apprentissage divergerait. La perte entre cible et prédiction sert à corriger les poids $\theta$. Schéma conceptuel original.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** agent de *trading* ou de couverture qui décide d'acheter / conserver / vendre à partir d'un état continu (prix récents, indicateurs techniques, position) — entraîné et évalué dans un simulateur de marché historique.
- **Cas d'usage 2 (logistique/commerce) :** gestion dynamique des stocks multi-produits ou tarification dynamique (prix discrétisés) quand l'état (stocks, demande, saison) est trop riche pour une table.
- **Données et prérequis :** environnement simulable rapidement (des centaines de milliers d'interactions), actions discrètes, bibliothèque de deep learning (PyTorch/TensorFlow) et souvent GPU ; normalisation des états.
- **Peu approprié :** petit problème discret comme FrozenLake 4 × 4 (le Q-learning tabulaire suffit, DQN serait inutilement complexe et instable) ; ou actions continues (montant exact à investir) → méthodes acteur-critique (DDPG, SAC).
- **Métriques :** récompense cumulée moyenne par épisode, taux de réussite, stabilité de la courbe d'apprentissage et variation entre graines ; en finance, rendement ajusté du risque **hors échantillon**.

```python
# Calcul de cible DQN, sans entraîner un réseau neuronal complet
recompenses=np.array([0.,1.,-1.]);termine=np.array([False,True,False]);meilleur_Q_suivant=np.array([2.,3.,1.]);gamma=.95
cibles=recompenses+gamma*(~termine)*meilleur_Q_suivant
print('Cibles de Bellman:',cibles)
# Un DQN complet ajoute réseau en ligne, réseau cible et mémoire de rejeu.
```

### Analyse de l'expérience Python
- **Entrées :** trois transitions **fictives** définies à la main : récompenses `[0, 1, −1]`, indicateurs de fin `[False, True, False]`, meilleures valeurs Q de l'état suivant `[2, 3, 1]`, $\gamma=0,95$. Aucun réseau n'est entraîné : l'exemple est **conceptuel**, comme l'indique le notebook.
- **Calcul effectué :** $y = r + \gamma\,(1-\text{fin})\max Q(s',\cdot)$.
- **Sortie obtenue :** cibles de Bellman **[1,9 ; 1,0 ; −0,05]**. Vérification : $0+0,95\times2=1,9$ ; transition terminale → $y=r=1$ (la valeur future 3 est ignorée) ; $-1+0,95\times1=-0,05$.
- **Conclusion possible :** le code illustre correctement l'étape centrale du DQN : la construction de la cible, et en particulier le rôle du masque `~termine` qui supprime la valeur future en fin d'épisode.
- **Ce qu'on ne peut pas conclure :** rien sur la performance d'un DQN : il n'y a ni réseau, ni mémoire de rejeu, ni réseau cible, ni environnement, ni apprentissage. Une expérience DQN complète (ex. sur `CartPole-v1`) serait facultative et devrait être signalée comme telle.

```python
# Régression linéaire par descente de gradient : données univariées
x=np.linspace(-1,1,100); y=2+3*x
poids=0.;biais=0.;pas=.1
for iteration in range(500):
    prediction=poids*x+biais
    erreur=prediction-y
    poids-=pas*2*np.mean(erreur*x)
    biais-=pas*2*np.mean(erreur)
print('Paramètres appris manuellement:',poids,biais)
from sklearn.linear_model import LinearRegression
print('Référence sklearn:',LinearRegression().fit(x.reshape(-1,1),y).coef_,LinearRegression().fit(x.reshape(-1,1),y).intercept_)
```

```python
# K-means manuel : assignation et recalcul jusqu'à stabilité
valeurs=X_clu.copy(); centroids=valeurs[np.random.default_rng(GRAINE).choice(len(valeurs),size=3,replace=False)].copy()
for iteration in range(100):
    distances=((valeurs[:,None,:]-centroids[None,:,:])**2).sum(axis=2)
    groupes=distances.argmin(axis=1)
    nouveaux=np.array([valeurs[groupes==k].mean(axis=0) if (groupes==k).any() else centroids[k] for k in range(3)])
    if np.allclose(nouveaux,centroids):break
    centroids=nouveaux
print('Centroïdes manuels:',centroids)
from sklearn.cluster import KMeans
print('Centroïdes sklearn (ordre arbitraire):',KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(valeurs).cluster_centers_)
```
