## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression linéaire** (*Linear Regression*, moindres carrés ordinaires) — famille : **régression supervisée**. Objectif : prédire une variable continue $y$ comme combinaison pondérée des caractéristiques.

**Intuition et étapes :** données $(X, y)$ → recherche des coefficients qui minimisent la somme des carrés des écarts entre valeurs observées et prédites → prédiction $\hat y$ pour une nouvelle observation.

$$\hat y_i=\beta_0+\sum_{j=1}^{p}\beta_j x_{ij}\qquad J(\beta)=\frac{1}{2n}\sum_{i=1}^{n}(y_i-\hat y_i)^2$$

$n$ : nombre d'observations ; $p$ : nombre de caractéristiques ; $x_{ij}$ : valeur de la variable $j$ pour l'observation $i$ ; $y_i$ : valeur observée ; $\hat y_i$ : valeur prédite ; $J$ : coût quadratique moyen. **Paramètres appris :** $\beta_0$ (constante) et $\beta_j$ (coefficients). **Hyperparamètre :** `fit_intercept` (estimer ou non la constante) — il n'y a pas de régularisation.

**Avantages :** (1) solution rapide et exacte (forme fermée) ; (2) coefficients directement interprétables comme effets marginaux *conditionnels*.
**Limites :** (1) suppose une relation linéaire — rate les effets non linéaires ; (2) sensible aux valeurs atypiques (le carré amplifie les grosses erreurs) et instable si les variables sont fortement corrélées.

### Schéma explicatif

```python
# Schéma 1 — Régression linéaire : droite des moindres carrés et résidus (illustration originale)
rng=np.random.default_rng(GRAINE)
x=np.linspace(0,10,25); y=3+1.5*x+rng.normal(0,2,25)
b1,b0=np.polyfit(x,y,1); yh=b0+b1*x
fig,ax=plt.subplots(figsize=(7,4))
ax.vlines(x,yh,y,colors='tab:red',lw=1,label='résidus $y_i-\\hat y_i$')
ax.scatter(x,y,s=25,label='observations $(x_i,y_i)$')
ax.plot(x,yh,'k',lw=2,label=f'droite estimée $\\hat y={b0:.1f}+{b1:.2f}x$')
ax.set_xlabel('x (ex. budget publicitaire)');ax.set_ylabel('y (ex. chiffre d’affaires)')
ax.set_title('Moindres carrés : minimiser la somme des carrés des segments rouges');ax.legend();plt.show()
```

**Comment lire le schéma :** chaque point bleu est une observation ; la droite noire est celle qui rend **minimale la somme des carrés des segments rouges** (résidus). La pente estimée indique de combien $\hat y$ varie quand $x$ augmente d'une unité. Les données sont simulées pour l'illustration.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/gestion) :** prévoir le chiffre d'affaires mensuel d'une enseigne à partir des dépenses marketing, du nombre de points de vente et de la saisonnalité.
- **Cas d'usage 2 :** estimer le prix d'un logement selon la surface, le quartier et l'ancienneté.
- **Données et prérequis :** tableau numérique (variables catégorielles encodées en indicatrices), cible continue ; relation à peu près linéaire, résidus sans structure, faible multicolinéarité, pas de fuite d'information.
- **Peu approprié :** prévoir le défaut de paiement (oui/non) — la cible est binaire, une droite donne des « probabilités » hors de [0 ; 1] → préférer la régression logistique. Également peu adapté aux relations fortement non linéaires (seuils, saturation).
- **Métriques :** MAE, RMSE, R² (sur un jeu de test indépendant) ; analyse des résidus.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** `Xr_ent` (210 × 5) et `yr_ent` issues de `make_regression` (280 lignes, 5 variables, bruit `noise=13`), découpage 75/25 avec `random_state=42`. Données **synthétiques**.
- **Hyperparamètres :** valeurs par défaut (`fit_intercept=True`).
- **Sortie obtenue :** **MAE ≈ 9,35** sur les 70 observations de test.
- **Ce que l'on peut conclure :** l'erreur absolue moyenne est faible comparée à l'échelle de la cible (écart-type de `yr_test` ≈ 97,6 et MAE ≈ 74,0 pour un modèle qui prédit toujours la moyenne — voir la cellule *Repères*). C'est cohérent : `make_regression` génère une relation **exactement linéaire** + bruit gaussien, donc le modèle linéaire est ici le « bon » modèle et l'erreur résiduelle est surtout due au bruit.
- **Ce que l'on ne peut pas conclure :** ce résultat ne dit rien de la performance sur des données économiques réelles ; une seule métrique sur un seul découpage ne mesure ni la variance de l'estimation ni la significativité des coefficients.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression Ridge** (*Ridge Regression*, régularisation L2 / Tikhonov) — famille : **régression supervisée**. Objectif : prédire une cible continue en **stabilisant** les coefficients lorsque les variables sont corrélées.

**Intuition et étapes :** variables standardisées → minimisation de l'erreur quadratique **plus** une pénalité sur la taille des coefficients → coefficients « rétrécis » vers 0 → prédiction linéaire.

$$J(\beta)=\frac{1}{n}\lVert y-X\beta\rVert_2^2+\lambda\sum_{j=1}^{p}\beta_j^2$$

$X$ : matrice $n\times p$ des caractéristiques ; $y$ : vecteur cible ; $\beta$ : vecteur des coefficients (**paramètres appris**, constante non pénalisée) ; $\lVert\cdot\rVert_2$ : norme euclidienne ; $\lambda$ (`alpha` dans scikit-learn) : **hyperparamètre** de force de régularisation, choisi par validation.

**Avantages :** (1) stable en présence de multicolinéarité ; (2) réduit la variance donc le surapprentissage quand $p$ est grand.
**Limites :** (1) ne met jamais un coefficient exactement à 0 → pas de sélection de variables ; (2) introduit un biais et nécessite de standardiser les variables et de régler $\lambda$.

### Schéma explicatif

```python
# Schéma 2 — Ridge : chemin des coefficients selon la force de pénalité alpha (illustration)
from sklearn.linear_model import Ridge
rng=np.random.default_rng(GRAINE)
Xs=rng.normal(size=(100,4)); Xs[:,1]=Xs[:,0]+rng.normal(0,.1,100)   # x1 et x2 très corrélées
ys=3*Xs[:,0]+2*Xs[:,1]-Xs[:,2]+rng.normal(0,1,100)
alphas=np.logspace(-2,4,60)
coefs=np.array([Ridge(alpha=a).fit(Xs,ys).coef_ for a in alphas])
fig,ax=plt.subplots(figsize=(7,4))
for j in range(4): ax.plot(alphas,coefs[:,j],label=f'β{j+1}')
ax.axhline(0,c='grey',lw=.8);ax.set_xscale('log')
ax.set_xlabel('alpha (λ) — échelle log');ax.set_ylabel('valeur du coefficient')
ax.set_title('Ridge : plus alpha augmente, plus les coefficients se rapprochent de 0 (sans l’atteindre)');ax.legend();plt.show()
```

**Comment lire le schéma :** en abscisse la force de pénalité $\lambda$ (échelle logarithmique), en ordonnée la valeur de chaque coefficient. À gauche (λ faible) on retrouve les moindres carrés ; vers la droite, tous les coefficients **rétrécissent progressivement vers 0 sans s'annuler**. Les variables x1 et x2, construites très corrélées, voient leurs coefficients se rapprocher l'un de l'autre : Ridge répartit l'effet entre variables corrélées. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** prévoir le rendement d'un portefeuille ou la croissance du PIB à partir d'indicateurs macroéconomiques très corrélés (taux, inflation, crédit, masse monétaire).
- **Cas d'usage 2 (marketing) :** modéliser les ventes à partir de nombreux canaux publicitaires dont les dépenses évoluent ensemble (*marketing mix modelling*).
- **Données et prérequis :** variables numériques **standardisées** (sinon la pénalité est injuste entre unités), cible continue ; choix de `alpha` par validation croisée, scaler ajusté sur l'entraînement uniquement (Pipeline).
- **Peu approprié :** quand l'objectif est d'identifier un petit sous-ensemble de variables pertinentes (ex. sélectionner 5 ratios parmi 200) — Ridge les garde tous ; préférer Lasso ou Elastic Net. Inutile aussi si la relation est fortement non linéaire.
- **Métriques :** MAE, RMSE, R² ; courbe de validation en fonction de `alpha`.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques que l'exemple 1 ; pipeline `StandardScaler` → `Ridge`, le scaler étant ajusté sur l'entraînement seulement (pas de fuite).
- **Hyperparamètre :** `alpha=1.0` (fixé, non réglé).
- **Sortie obtenue :** **MAE ≈ 9,37**, quasiment identique à la régression linéaire (≈ 9,35).
- **Conclusion possible :** sur ces données, la pénalité ne change presque rien : il n'y a que 5 variables indépendantes, 210 observations et aucune colinéarité, donc il n'y a pas de variance excessive à réduire. La légère hausse de MAE reflète le petit biais introduit par la pénalité.
- **Limites :** `alpha` n'a pas été choisi par validation croisée ; on ne peut pas conclure que Ridge « n'apporte rien » en général — son intérêt apparaît avec des variables corrélées ou nombreuses, absentes ici.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression Lasso** (*Least Absolute Shrinkage and Selection Operator*, régularisation L1) — famille : **régression supervisée**. Objectif : prédire une cible continue tout en **sélectionnant** automatiquement les variables utiles.

**Intuition et étapes :** variables standardisées → minimisation de l'erreur quadratique + pénalité sur la **valeur absolue** des coefficients → certains coefficients deviennent exactement nuls → modèle linéaire parcimonieux.

$$J(\beta)=\frac{1}{2n}\lVert y-X\beta\rVert_2^2+\lambda\sum_{j=1}^{p}\lvert\beta_j\rvert$$

$n$ : taille de l'échantillon ; $X$ : matrice explicative ; $y$ : cible ; $\beta_j$ : coefficients (**paramètres appris**) ; $|\cdot|$ : valeur absolue ; $\lambda$ (`alpha`) : **hyperparamètre** de pénalité L1. `max_iter` est un hyperparamètre de l'optimiseur (descente par coordonnées).

**Avantages :** (1) sélection de variables intégrée → modèle plus simple à expliquer ; (2) utile lorsque $p$ est grand devant $n$.
**Limites :** (1) instable avec des variables fortement corrélées (en garde une au hasard) ; (2) biais sur les grands coefficients et sensibilité au choix de $\lambda$.

### Schéma explicatif

```python
# Schéma 3 — Lasso : chemin des coefficients, annulation progressive (illustration)
from sklearn.linear_model import Lasso
rng=np.random.default_rng(GRAINE)
Xs=rng.normal(size=(120,6)); ys=4*Xs[:,0]-3*Xs[:,1]+1.5*Xs[:,2]+rng.normal(0,1,120)  # 3 variables inutiles
alphas=np.logspace(-3,1,60)
coefs=np.array([Lasso(alpha=a,max_iter=10000).fit(Xs,ys).coef_ for a in alphas])
fig,ax=plt.subplots(figsize=(7,4))
for j in range(6): ax.plot(alphas,coefs[:,j],ls='-' if j<3 else '--',label=f'β{j+1}'+(' (utile)' if j<3 else ' (bruit)'))
ax.axhline(0,c='grey',lw=.8);ax.set_xscale('log')
ax.set_xlabel('alpha (λ) — échelle log');ax.set_ylabel('coefficient')
ax.set_title('Lasso : les coefficients atteignent exactement 0 un par un');ax.legend(fontsize=8);plt.show()
```

**Comment lire le schéma :** comme pour Ridge, chaque courbe suit un coefficient quand $\lambda$ augmente. La différence clé : les courbes **touchent exactement 0** et y restent. Les variables de bruit (pointillés) sont éliminées en premier ; à λ très élevé, même les variables utiles disparaissent. Lire le schéma de gauche à droite revient à lire un **classement d'importance** des variables. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** sélectionner, parmi des dizaines de ratios financiers (liquidité, rentabilité, endettement…), ceux qui expliquent le mieux la note de crédit ou le spread d'une entreprise.
- **Cas d'usage 2 (commerce international) :** identifier les déterminants principaux du volume d'exportations d'un pays parmi de nombreux indicateurs (taux de change, PIB partenaires, tarifs…).
- **Données et prérequis :** variables numériques standardisées, cible continue ; suffisamment d'observations ; `alpha` choisi par validation croisée (`LassoCV`).
- **Peu approprié :** variables très corrélées dont on veut garder l'effet collectif (ex. plusieurs taux d'intérêt de maturités voisines) : le Lasso en retient arbitrairement un et la sélection change d'un échantillon à l'autre → préférer Ridge ou Elastic Net.
- **Métriques :** MAE, RMSE, R² ; nombre de coefficients non nuls et stabilité de la sélection (bootstrap).

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression ; pipeline `StandardScaler` → `Lasso`.
- **Hyperparamètres :** `alpha=0.1` (pénalité faible), `max_iter=10000` pour garantir la convergence.
- **Sortie obtenue :** **R² ≈ 0,986** sur le test : environ 98,6 % de la variance de la cible de test est expliquée.
- **Conclusion possible :** le modèle linéaire pénalisé reproduit très bien une relation qui est, par construction, linéaire. Avec `alpha=0.1` relativement à des coefficients de l'ordre de 10 à 64, la pénalité est trop faible pour annuler une variable : en inspectant `modele[-1].coef_`, les **5 coefficients restent non nuls** (≈ 9,8 ; 16,5 ; 20,6 ; 63,0 ; 63,6), ce qui est logique car les 5 variables générées sont toutes informatives.
- **Limites :** le R² n'est pas comparable directement aux MAE des exemples 1–2 (métrique différente) ; un R² élevé sur données synthétiques ne garantit rien sur données réelles, et la sélection de variables — l'intérêt principal du Lasso — n'est pas illustrée par cet exemple.

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression polynomiale** (*Polynomial Regression*) — famille : **régression supervisée**. Objectif : modéliser une relation **courbe** tout en gardant un modèle linéaire dans ses coefficients.

**Intuition et étapes :** variables d'origine → création de nouvelles variables (puissances $x^2, x^3\dots$ et interactions $x_1x_2$) → régression linéaire (souvent pénalisée) sur cet espace enrichi → prédiction non linéaire en $x$.

$$\hat y=\beta_0+\beta_1x+\beta_2x^2+\dots+\beta_dx^d$$

$x$ : variable d'entrée ; $d$ (`degree`) : degré maximal — **hyperparamètre** ; $\beta_j$ : coefficients — **paramètres appris** ; $\hat y$ : prédiction. Dans le code, `alpha` de la Ridge associée est un second hyperparamètre.

**Avantages :** (1) capte des courbures et interactions simples avec un modèle connu et rapide ; (2) reste interprétable pour un faible degré.
**Limites :** (1) le nombre de variables explose avec le degré et la dimension (combinatoire) ; (2) surapprentissage et extrapolation très dangereuse hors de l'intervalle observé.

### Schéma explicatif

```python
# Schéma 4 — Régression polynomiale : effet du degré (sous- et surapprentissage), illustration
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
rng=np.random.default_rng(GRAINE)
x=np.sort(rng.uniform(-3,3,30)); y=0.5*x**2-x+rng.normal(0,1,30)
grille=np.linspace(-3.2,3.2,300)
fig,ax=plt.subplots(figsize=(7,4));ax.scatter(x,y,c='k',s=18,label='observations')
for d,c in [(1,'tab:blue'),(2,'tab:green'),(15,'tab:red')]:
    m=make_pipeline(PolynomialFeatures(d),LinearRegression()).fit(x[:,None],y)
    ax.plot(grille,m.predict(grille[:,None]),c=c,label=f'degré {d}')
ax.set_ylim(-4,12);ax.set_xlabel('x');ax.set_ylabel('y')
ax.set_title('Degré 1 : sous-apprentissage — degré 2 : adapté — degré 15 : surapprentissage');ax.legend();plt.show()
```

**Comment lire le schéma :** les points noirs suivent une parabole bruitée. La droite (degré 1, bleu) est trop rigide : **sous-apprentissage**. La courbe de degré 2 (vert) suit la forme réelle. La courbe de degré 15 (rouge) serpente pour passer près de chaque point et diverge aux bords : **surapprentissage** et mauvaise extrapolation. Le degré est donc un hyperparamètre à choisir sur un jeu de validation. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/gestion) :** modéliser une courbe de coût avec économies d'échelle (coût unitaire qui baisse puis remonte avec le volume produit), ou une relation âge–revenu en courbe en cloche.
- **Cas d'usage 2 (marketing) :** rendement décroissant des dépenses publicitaires (effet de saturation) ou effet d'interaction prix × promotion.
- **Données et prérequis :** peu de variables numériques continues, standardisation recommandée (puissances d'échelles très différentes), régularisation (Ridge) et validation du degré.
- **Peu approprié :** données à grande dimension (100 variables en degré 3 → plus de 170 000 termes) ou prévision hors de la plage observée (ex. extrapoler une tendance de prix sur 10 ans) : le polynôme explose. Préférer arbres/boosting ou splines.
- **Métriques :** MAE, RMSE, R² ; comparaison des erreurs d'entraînement et de validation selon le degré.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression (5 variables) ; pipeline `StandardScaler` → `PolynomialFeatures(degree=2, include_bias=False)` (5 variables → 20 variables : 5 linéaires, 5 carrés, 10 interactions) → `Ridge(alpha=10)`.
- **Hyperparamètres :** `degree=2`, `alpha=10`.
- **Sortie obtenue :** **MAE ≈ 10,90**, soit **plus élevée** que la régression linéaire (≈ 9,35).
- **Conclusion possible :** comme la vraie relation est linéaire, les termes quadratiques et d'interaction n'apportent pas d'information ; ils ajoutent de la variance, et la pénalité `alpha=10` (plus forte qu'à l'exemple 2) rétrécit aussi les coefficients linéaires utiles → biais. Plus complexe n'est donc pas meilleur lorsque le modèle simple est déjà correctement spécifié.
- **Limites :** degré et `alpha` n'ont pas été optimisés par validation ; un seul découpage ; on ne peut pas généraliser ce classement à des données réellement non linéaires.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Arbre de régression** (*Regression Tree*, CART) — famille : **régression supervisée**. Objectif : prédire une valeur continue en découpant l'espace des variables en zones homogènes.

**Intuition et étapes :** données → à chaque nœud, test de toutes les variables et de tous les seuils → choix de la coupure qui réduit le plus la variance (erreur quadratique) des cibles → répétition récursive jusqu'à un critère d'arrêt → prédiction = **moyenne des cibles de la feuille** atteinte.

$$\hat y(x)=\frac{1}{|L(x)|}\sum_{i\in L(x)}y_i$$

$L(x)$ : ensemble des observations d'entraînement de la feuille où tombe $x$ ; $|L(x)|$ : effectif de cette feuille ; $y_i$ : cibles de ces observations. **Paramètres appris :** variables et seuils de coupure, valeurs des feuilles. **Hyperparamètres :** `max_depth`, `min_samples_leaf`, `min_samples_split`.

**Avantages :** (1) capte non-linéarités et interactions sans les spécifier ; (2) lisible (règles « si… alors ») et sans besoin de standardisation.
**Limites :** (1) très instable — une petite variation des données change l'arbre ; (2) prédiction en escalier et incapacité d'extrapoler au-delà des valeurs vues.

### Schéma explicatif

```python
# Schéma 5 — Arbre de régression : prédiction en escalier et structure de l'arbre (illustration)
from sklearn.tree import DecisionTreeRegressor, plot_tree
rng=np.random.default_rng(GRAINE)
x=np.sort(rng.uniform(0,10,80)); y=np.sin(x)*3+x+rng.normal(0,.6,80)
arbre=DecisionTreeRegressor(max_depth=2,random_state=GRAINE).fit(x[:,None],y)
fig,ax=plt.subplots(1,2,figsize=(12,4))
g=np.linspace(0,10,500); ax[0].scatter(x,y,s=12,c='grey',label='observations')
ax[0].plot(g,arbre.predict(g[:,None]),'tab:red',lw=2,label='prédiction (moyenne par feuille)')
ax[0].set_xlabel('x (ex. distance en centaines de km)');ax[0].set_ylabel('y (ex. coût logistique)');ax[0].legend();ax[0].set_title('Fonction en escalier')
plot_tree(arbre,feature_names=['x'],filled=True,ax=ax[1],fontsize=8);ax[1].set_title('Arbre de profondeur 2 : 4 feuilles')
plt.tight_layout();plt.show()
```

**Comment lire le schéma :** à droite, chaque nœud pose une question « x ≤ seuil ? » ; on descend à gauche si oui, à droite sinon, jusqu'à une feuille qui contient le nombre d'observations (`samples`) et leur moyenne (`value`). À gauche, la ligne rouge montre que l'arbre prédit une **constante par zone** : 4 feuilles = 4 paliers. Augmenter `max_depth` multiplie les paliers (jusqu'au surapprentissage). Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique) :** estimer le coût d'une expédition selon la distance, le poids, le mode de transport et la saison, avec des règles lisibles pour les opérationnels.
- **Cas d'usage 2 (gestion) :** estimer la durée de traitement d'un dossier de crédit selon le type de client et le montant.
- **Données et prérequis :** tableau mixte numérique/catégoriel, pas de standardisation nécessaire ; contrôle de la profondeur ou élagage pour éviter le surapprentissage.
- **Peu approprié :** prévoir une tendance qui sort de la plage historique (ex. prix de l'immobilier en hausse continue) : l'arbre ne peut pas prédire au-delà de la moyenne de sa feuille la plus extrême. Peu adapté aussi aux relations lisses et linéaires (approximées par marches d'escalier).
- **Métriques :** MAE, RMSE, R² ; comparaison erreur d'entraînement / validation selon `max_depth`.

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression (5 variables, relation linéaire).
- **Hyperparamètres :** `max_depth=5` (au plus 32 feuilles), `random_state=42`.
- **Sortie obtenue :** **MAE ≈ 37,72**, environ **4 fois** celle de la régression linéaire (≈ 9,35), mais meilleure que la prédiction constante par la moyenne (≈ 74,0).
- **Conclusion possible :** l'arbre approxime un hyperplan lisse par des paliers ; avec seulement 210 observations réparties en ≤ 32 feuilles, chaque feuille moyenne peu de points et la discrétisation génère une grosse erreur. Il n'est pas adapté à une relation additive linéaire, ce qui illustre bien sa limite.
- **Limites :** profondeur non réglée par validation ; un arbre unique est instable, la MAE pourrait varier sensiblement avec un autre découpage aléatoire.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Forêt aléatoire — régression** (*Random Forest Regressor*) — famille : **régression supervisée, méthode d'ensemble (bagging)**. Objectif : obtenir une prédiction plus stable et précise qu'un arbre unique.

**Intuition et étapes :** données → tirage de $B$ échantillons bootstrap (avec remise) → pour chacun, entraînement d'un arbre profond où chaque coupure n'examine qu'un sous-ensemble aléatoire de variables → nouvelle observation passée dans tous les arbres → **moyenne** des prédictions.

$$\hat y(x)=\frac{1}{B}\sum_{b=1}^{B}T_b(x)$$

$B$ (`n_estimators`) : nombre d'arbres — **hyperparamètre** ; $T_b(x)$ : prédiction de l'arbre $b$ ; $x$ : nouvelle observation. **Paramètres appris :** structure et feuilles de chaque arbre. Autres hyperparamètres : `max_depth`, `max_features`, `min_samples_leaf`.

**Avantages :** (1) réduit fortement la variance d'un arbre unique, robuste au bruit et aux valeurs atypiques ; (2) peu de réglage nécessaire, fournit une importance des variables et une erreur *out-of-bag*.
**Limites :** (1) modèle volumineux et moins interprétable qu'un arbre ; (2) n'extrapole pas hors de la plage des cibles observées.

### Schéma explicatif

```python
# Schéma 6 — Forêt aléatoire : agrégation de plusieurs arbres bootstrap (illustration)
from sklearn.tree import DecisionTreeRegressor
rng=np.random.default_rng(GRAINE)
x=np.sort(rng.uniform(0,10,80)); y=np.sin(x)*3+x+rng.normal(0,.8,80); g=np.linspace(0,10,400)
preds=[]
fig,ax=plt.subplots(figsize=(8,4));ax.scatter(x,y,s=10,c='k',label='observations')
for b in range(15):
    idx=rng.integers(0,80,80)                              # échantillon bootstrap
    p=DecisionTreeRegressor(max_depth=4,random_state=b).fit(x[idx,None],y[idx]).predict(g[:,None]); preds.append(p)
    ax.plot(g,p,c='tab:blue',alpha=.25,lw=1,label='arbre individuel $T_b$' if b==0 else None)
ax.plot(g,np.mean(preds,axis=0),'tab:red',lw=2.5,label='moyenne des 15 arbres (forêt)')
ax.set_xlabel('x');ax.set_ylabel('y');ax.set_title('Chaque arbre est bruité ; leur moyenne est plus lisse et plus stable');ax.legend();plt.show()
```

**Comment lire le schéma :** chaque courbe bleue pâle est un arbre entraîné sur un échantillon bootstrap différent : elles sont irrégulières et diffèrent entre elles (**forte variance**). La courbe rouge est leur moyenne : plus lisse et plus proche de la tendance. C'est le principe du *bagging* : agréger des modèles instables et peu corrélés pour réduire la variance. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (commerce international) :** prévoir la demande d'exportation d'un produit selon prix, taux de change, saison et pays de destination.
- **Cas d'usage 2 (finance) :** estimer la valeur d'un bien immobilier en garantie d'un crédit (variables hétérogènes, relations non linéaires).
- **Données et prérequis :** tableau numérique/catégoriel, pas de standardisation ; plusieurs centaines d'observations au minimum ; mémoire et temps de calcul proportionnels au nombre d'arbres.
- **Peu approprié :** prévision d'une série en tendance (inflation, cours en hausse) où il faut extrapoler ; ou quand une explication réglementaire précise de chaque prédiction est exigée (ex. justification d'un refus de crédit) — un modèle linéaire ou un arbre unique est plus défendable.
- **Métriques :** MAE, RMSE, R² ; erreur *out-of-bag* ; importance des variables par permutation.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression.
- **Hyperparamètres :** `n_estimators=80`, arbres de profondeur non limitée (défaut), `random_state=42`.
- **Sortie obtenue :** **MAE ≈ 24,37**, nettement meilleure que l'arbre unique (≈ 37,72) mais toujours bien moins bonne que la régression linéaire (≈ 9,35).
- **Conclusion possible :** l'agrégation réduit bien la variance de l'arbre (gain d'environ 35 % sur la MAE), ce qui illustre l'intérêt du bagging. Mais la moyenne d'arbres reste une fonction en paliers qui approxime mal un hyperplan, surtout aux extrémités de la distribution (pas d'extrapolation).
- **Limites :** pas de validation des hyperparamètres ; données synthétiques linéaires qui avantagent structurellement le modèle linéaire ; ce classement ne se transpose pas à des données réelles non linéaires.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Gradient Boosting — régression** (*Gradient Boosting Regressor*, ici la version histogramme `HistGradientBoostingRegressor`) — famille : **régression supervisée, ensemble séquentiel (boosting)**. Objectif : construire un modèle précis en additionnant de petits arbres qui corrigent les erreurs des précédents.

**Intuition et étapes :** prédiction initiale (moyenne) → calcul des pseudo-résidus (gradient de la perte, = résidus pour la perte quadratique) → entraînement d'un petit arbre sur ces résidus → ajout de cet arbre multiplié par un taux d'apprentissage → répétition $M$ fois.

$$F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$$

$F_m$ : modèle après $m$ itérations ; $h_m$ : nouvel arbre faible ajusté aux pseudo-résidus de $F_{m-1}$ ; $\eta$ (`learning_rate`) : taux d'apprentissage ; $M$ (`max_iter`) : nombre d'itérations. **Hyperparamètres :** $\eta$, $M$, `max_depth`/`max_leaf_nodes`, régularisation. **Paramètres appris :** structure et valeurs des arbres. La version « Hist » discrétise les variables en 255 intervalles pour accélérer le calcul.

**Avantages :** (1) souvent parmi les meilleures performances sur données tabulaires ; (2) gère non-linéarités, interactions et (version Hist) valeurs manquantes nativement.
**Limites :** (1) nombreux hyperparamètres interdépendants, surapprentissage si $M$ trop grand ; (2) apprentissage séquentiel (moins parallélisable) et interprétation difficile.

### Schéma explicatif

```python
# Schéma 7 — Gradient boosting : correction séquentielle des résidus (illustration)
from sklearn.tree import DecisionTreeRegressor
rng=np.random.default_rng(GRAINE)
x=np.sort(rng.uniform(0,10,100)); y=np.sin(x)*3+x+rng.normal(0,.5,100); g=np.linspace(0,10,400)
eta=0.3; F=np.full_like(y,y.mean()); Fg=np.full_like(g,y.mean())
fig,ax=plt.subplots(1,2,figsize=(12,4)); ax[0].scatter(x,y,s=8,c='k')
erreurs=[np.mean((y-F)**2)]
for m in range(1,51):
    h=DecisionTreeRegressor(max_depth=2).fit(x[:,None],y-F)      # arbre sur les résidus
    F+=eta*h.predict(x[:,None]); Fg+=eta*h.predict(g[:,None]); erreurs.append(np.mean((y-F)**2))
    if m in (1,5,50): ax[0].plot(g,Fg,label=f'F_{m}')
ax[0].set_title('Le modèle se rapproche des données itération après itération');ax[0].legend();ax[0].set_xlabel('x')
ax[1].plot(erreurs,'o-',ms=3);ax[1].set_xlabel('itération m');ax[1].set_ylabel('MSE entraînement');ax[1].set_title('Chaque arbre réduit l’erreur restante')
plt.tight_layout();plt.show()
```

**Comment lire le schéma :** à gauche, $F_1$ (un seul petit arbre après la moyenne) est grossier ; $F_5$ capte déjà la forme générale ; $F_{50}$ suit finement la courbe. À droite, l'erreur d'entraînement diminue à chaque itération car chaque arbre corrige les résidus du modèle précédent. Attention : l'erreur d'**entraînement** baisse toujours — c'est l'erreur de **validation** qui indique quand s'arrêter. Données simulées, $\eta = 0{,}3$.

### Contextes d'utilisation
- **Cas d'usage 1 (commerce/gestion) :** prévision des ventes journalières par magasin et par produit (promotions, jours fériés, météo) — très utilisé en retail.
- **Cas d'usage 2 (finance) :** estimation de la perte en cas de défaut (LGD) ou de la valeur client (CLV) à partir de nombreuses variables hétérogènes.
- **Données et prérequis :** données tabulaires de taille moyenne à grande, pas de standardisation ; jeu de validation (ou *early stopping*) pour fixer le nombre d'itérations ; découpage chronologique si séries temporelles.
- **Peu approprié :** très petit jeu de données (quelques dizaines de lignes) où il surapprend facilement, ou données non tabulaires (images, texte brut) où les réseaux de neurones dominent ; extrapolation de tendance impossible comme pour tous les arbres.
- **Métriques :** MAE, RMSE, R² ; courbe d'apprentissage (erreur de validation en fonction de $m$).

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression.
- **Hyperparamètres :** `max_iter=80` itérations, autres valeurs par défaut (`learning_rate=0.1`, `max_leaf_nodes=31`), `random_state=42`.
- **Sortie obtenue :** **MAE ≈ 23,27**, légèrement meilleure que la forêt aléatoire (≈ 24,37), mais toujours loin de la régression linéaire (≈ 9,35).
- **Conclusion possible :** le boosting corrige progressivement les erreurs et fait mieux que les autres méthodes à base d'arbres ici, mais avec seulement 210 observations et une relation exactement linéaire, les arbres restent désavantagés. La hiérarchie obtenue (linéaire > boosting ≈ forêt > arbre seul) est propre à ce jeu synthétique.
- **Limites :** `max_iter` et `learning_rate` non réglés (pas de validation ni d'*early stopping*) ; pas de mesure de variance entre plusieurs graines ; aucune conclusion sur des données réelles.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression logistique** (*Logistic Regression*) — famille : **classification supervisée** (malgré son nom). Objectif : estimer la **probabilité** qu'une observation appartienne à la classe positive.

**Intuition et étapes :** variables (standardisées) → score linéaire $z=\beta_0+\beta^\top x$ → transformation par la fonction sigmoïde en probabilité → estimation des coefficients par maximum de vraisemblance (minimisation de la log-perte) → classe prédite = 1 si $p\ge$ seuil (0,5 par défaut).

$$P(y=1\mid x)=\sigma(z)=\frac{1}{1+e^{-(\beta_0+\beta^\top x)}}\qquad \mathcal L=-\frac1n\sum_i\big[y_i\ln p_i+(1-y_i)\ln(1-p_i)\big]+\frac{1}{2C}\lVert\beta\rVert^2$$

$x$ : vecteur des caractéristiques ; $\beta_0,\beta$ : **paramètres appris** ; $\sigma$ : sigmoïde ; $p_i$ : probabilité prédite ; $y_i\in\{0,1\}$ : classe réelle ; $C$ : **hyperparamètre** inverse de la régularisation L2 (défaut 1). Le seuil de décision est aussi un choix (hyperparamètre métier).

**Avantages :** (1) produit des probabilités, généralement bien calibrées ; (2) coefficients interprétables (en *odds ratio* via $e^{\beta_j}$), standard réglementaire en scoring.
**Limites :** (1) frontière de décision linéaire ; (2) sensible à la multicolinéarité et aux classes très déséquilibrées si le seuil n'est pas ajusté.

### Schéma explicatif

```python
# Schéma 8 — Régression logistique : sigmoïde et frontière linéaire (illustration)
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification
Xs,ys=make_classification(n_samples=200,n_features=2,n_redundant=0,n_informative=2,n_clusters_per_class=1,class_sep=1.2,random_state=GRAINE)
m=LogisticRegression().fit(Xs,ys)
fig,ax=plt.subplots(1,2,figsize=(12,4))
z=np.linspace(-6,6,200);ax[0].plot(z,1/(1+np.exp(-z)),lw=2);ax[0].axhline(.5,ls='--',c='grey');ax[0].axvline(0,ls='--',c='grey')
ax[0].set_xlabel('score linéaire z = β0 + βᵀx');ax[0].set_ylabel('P(y=1|x)');ax[0].set_title('Sigmoïde : score → probabilité')
xx,yy=np.meshgrid(np.linspace(Xs[:,0].min()-1,Xs[:,0].max()+1,200),np.linspace(Xs[:,1].min()-1,Xs[:,1].max()+1,200))
P=m.predict_proba(np.c_[xx.ravel(),yy.ravel()])[:,1].reshape(xx.shape)
cs=ax[1].contourf(xx,yy,P,levels=20,cmap='RdBu_r',alpha=.6);ax[1].contour(xx,yy,P,levels=[.5],colors='k')
ax[1].scatter(Xs[:,0],Xs[:,1],c=ys,cmap='bwr',edgecolor='white',linewidth=.5,s=18);plt.colorbar(cs,ax=ax[1],label='probabilité classe 1')
ax[1].set_title('Frontière p = 0,5 (ligne noire) : une droite');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** à gauche, la sigmoïde transforme n'importe quel score réel en probabilité entre 0 et 1 ; score nul ↔ probabilité 0,5. À droite, la couleur de fond donne la probabilité estimée d'appartenir à la classe 1 (rouge) ; la **ligne noire** est la frontière où $p=0{,}5$ — elle est **droite**, car le score est linéaire. Plus un point est loin de la frontière, plus la prédiction est confiante. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** scoring de crédit — probabilité de défaut d'un emprunteur à partir de son revenu, endettement et historique (ex. dataset *Default of Credit Card Clients*).
- **Cas d'usage 2 (marketing) :** probabilité qu'un client réponde à une campagne ou résilie son abonnement (*churn*).
- **Données et prérequis :** cible binaire (ou multiclasse), variables numériques standardisées, catégorielles encodées ; pas de séparation parfaite des classes ; ajuster le seuil ou pondérer les classes si déséquilibre.
- **Peu approprié :** frontière fortement non linéaire (ex. détection de fraude reposant sur des combinaisons complexes de comportements) sans ingénierie de variables — un modèle à base d'arbres capturera mieux ces interactions.
- **Métriques :** F1, précision, rappel, ROC-AUC (ou PR-AUC si classe rare), log-perte et calibration.

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** pipeline `StandardScaler` → `LogisticRegression(max_iter=1000)` ; régularisation L2 par défaut avec `C=1.0`, seuil 0,5.
- **Sortie obtenue :** **F1 ≈ 0,811** sur la classe positive.
- **Conclusion possible :** le modèle linéaire classe correctement une large majorité des observations, mais il est parmi les plus faibles des classifieurs testés (KNN ≈ 0,889, boosting ≈ 0,950). Cela suggère que la frontière entre classes générée par `make_classification` n'est pas purement linéaire (classes composées de plusieurs amas gaussiens).
- **Limites :** le F1 seul ne dit pas si l'erreur vient des faux positifs ou des faux négatifs (il faudrait la matrice de confusion) ; aucune mesure de la qualité des probabilités (ROC-AUC, calibration) ; un seul découpage de 100 observations — une différence de quelques points de F1 entre modèles peut relever du hasard.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**K plus proches voisins** (*K-Nearest Neighbors*, KNN) — famille : **classification supervisée** (existe aussi en régression). Objectif : classer une observation selon la classe majoritaire de ses voisins les plus proches.

**Intuition et étapes :** variables standardisées → « apprentissage » = simple mémorisation du jeu d'entraînement → pour un nouveau point : calcul des distances à tous les points mémorisés → sélection des $K$ plus proches → **vote majoritaire** des classes.

$$\hat y(x)=\arg\max_{c}\sum_{i\in\mathcal N_K(x)}\mathbb 1(y_i=c)\qquad d(x,x_i)=\sqrt{\sum_{j=1}^{p}(x_j-x_{ij})^2}$$

$\mathcal N_K(x)$ : indices des $K$ voisins les plus proches de $x$ ; $c$ : une classe ; $\mathbb 1$ : indicatrice (1 si vrai) ; $d$ : distance euclidienne. **Pas de paramètres appris** (méthode non paramétrique, le modèle *est* le jeu d'entraînement). **Hyperparamètres :** $K$ (`n_neighbors`), la métrique de distance, la pondération des votes.

**Avantages :** (1) aucune hypothèse sur la forme de la frontière, s'adapte à des formes complexes ; (2) simple à comprendre et à expliquer (« ressemble à ces clients »).
**Limites :** (1) prédiction lente et mémoire élevée sur gros volumes (distance à tous les points) ; (2) très sensible à l'échelle des variables et à la malédiction de la dimension.

### Schéma explicatif

```python
# Schéma 9 — KNN : vote des K plus proches voisins (illustration)
rng=np.random.default_rng(GRAINE)
A_=rng.normal([2,2],1,(25,2)); B_=rng.normal([4.5,4.5],1,(25,2)); Xs=np.vstack([A_,B_]); ys=np.r_[np.zeros(25),np.ones(25)]
q=np.array([3.4,3.3]); d=np.linalg.norm(Xs-q,axis=1); K=5; vois=np.argsort(d)[:K]
fig,ax=plt.subplots(figsize=(6,5))
ax.scatter(*A_.T,c='tab:blue',label='classe 0');ax.scatter(*B_.T,c='tab:orange',label='classe 1')
for i in vois: ax.plot([q[0],Xs[i,0]],[q[1],Xs[i,1]],'k--',lw=.8)
ax.add_patch(plt.Circle(q,d[vois[-1]],fill=False,ls=':',color='k'))
ax.scatter(*q,marker='*',s=300,c='red',edgecolor='k',label='nouveau point')
v=int(ys[vois].sum());ax.set_title(f'K=5 : {K-v} voisins classe 0, {v} voisins classe 1 → prédiction : classe {int(v>K/2)}')
ax.legend();ax.set_xlabel('variable 1 (standardisée)');ax.set_ylabel('variable 2 (standardisée)');plt.show()
```

**Comment lire le schéma :** l'étoile rouge est le point à classer. Les pointillés le relient à ses 5 plus proches voisins, contenus dans le cercle. La classe prédite est celle qui domine dans ce cercle (indiquée dans le titre). Avec un $K$ plus petit, la décision dépend de quelques points (frontière irrégulière, variance élevée) ; avec un $K$ plus grand, elle se lisse (biais plus élevé). Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** recommander une offre à un client en s'inspirant des choix des clients au profil le plus proche (âge, panier moyen, fréquence d'achat).
- **Cas d'usage 2 (finance) :** classer une demande de crédit en comparant le demandeur aux dossiers passés les plus similaires ; imputation de valeurs manquantes (KNNImputer).
- **Données et prérequis :** variables numériques **standardisées** (obligatoire), dimension modérée, volume raisonnable (calcul des distances) ; choix de $K$ par validation croisée.
- **Peu approprié :** millions de transactions bancaires à scorer en temps réel (coût de calcul à chaque prédiction), ou données de très grande dimension (ex. 5 000 variables textuelles) où toutes les distances deviennent similaires.
- **Métriques :** F1, précision, rappel, ROC-AUC (avec probabilités = proportion de voisins) ; accuracy si classes équilibrées.

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `StandardScaler` (indispensable pour des distances) → `KNeighborsClassifier(n_neighbors=5)`, distance euclidienne, vote non pondéré.
- **Sortie obtenue :** **F1 ≈ 0,889**, nettement au-dessus de la régression logistique (≈ 0,811).
- **Conclusion possible :** un classifieur purement local, sans hypothèse de linéarité, fait mieux ici : cela confirme que les classes forment des amas que la frontière linéaire sépare mal. Avec seulement 6 variables, la malédiction de la dimension reste limitée.
- **Limites :** $K=5$ non choisi par validation croisée ; sur 100 observations de test, l'écart avec d'autres modèles reste incertain ; aucune information sur le coût de prédiction, qui serait le vrai frein sur un gros jeu réel.

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Classifieur bayésien naïf** (*Naive Bayes*, ici *Gaussian Naive Bayes*) — famille : **classification supervisée** (modèle génératif probabiliste). Objectif : attribuer la classe la plus probable sachant les caractéristiques observées.

**Intuition et étapes :** données → estimation, pour chaque classe, de sa fréquence (a priori) et de la distribution de chaque variable (ici une loi normale : moyenne et variance) → application du théorème de Bayes en supposant les variables **indépendantes conditionnellement à la classe** → classe de probabilité a posteriori maximale.

$$P(c\mid x)\propto P(c)\prod_{j=1}^{p}P(x_j\mid c),\qquad P(x_j\mid c)=\frac{1}{\sqrt{2\pi\sigma_{jc}^2}}\exp\!\Big(-\frac{(x_j-\mu_{jc})^2}{2\sigma_{jc}^2}\Big)$$

$c$ : classe ; $P(c)$ : probabilité a priori ; $x_j$ : valeur de la variable $j$ ; $\mu_{jc},\sigma^2_{jc}$ : moyenne et variance de $x_j$ dans la classe $c$ — **paramètres appris** ; $\propto$ : « proportionnel à ». **Hyperparamètres :** `var_smoothing` (lissage numérique des variances), éventuellement `priors`.

**Avantages :** (1) extrêmement rapide, fonctionne avec très peu de données ; (2) gère bien un grand nombre de variables (texte).
**Limites :** (1) hypothèse d'indépendance rarement vraie → probabilités mal calibrées ; (2) la version gaussienne suppose des distributions normales par classe.

### Schéma explicatif

```python
# Schéma 10 — Naive Bayes gaussien : une loi normale par classe et par variable (illustration)
from scipy.stats import norm
rng=np.random.default_rng(GRAINE)
x0=rng.normal(2,1,200); x1=rng.normal(5,1.3,120)        # variable x pour classe 0 et classe 1
g=np.linspace(-2,10,400); p0=len(x0)/320; p1=len(x1)/320
d0=p0*norm.pdf(g,x0.mean(),x0.std()); d1=p1*norm.pdf(g,x1.mean(),x1.std())
fig,ax=plt.subplots(figsize=(8,4))
ax.hist(x0,bins=25,density=True,alpha=.25,color='tab:blue');ax.hist(x1,bins=25,density=True,alpha=.25,color='tab:orange')
ax.plot(g,d0,'tab:blue',lw=2,label='P(c=0)·P(x|c=0)');ax.plot(g,d1,'tab:orange',lw=2,label='P(c=1)·P(x|c=1)')
croise=np.where(np.diff(np.sign(d0-d1)))[0]; s=[g[i] for i in croise if 2<g[i]<6][0]
ax.axvline(s,c='k',ls='--',label=f'seuil de décision ≈ {s:.2f}')
ax.set_xlabel('valeur de la variable x');ax.set_ylabel('densité');ax.set_title('On choisit la classe dont la courbe pondérée est la plus haute');ax.legend();plt.show()
```

**Comment lire le schéma :** pour une variable, le modèle ajuste une courbe normale par classe (histogrammes = données observées), pondérée par la fréquence de la classe $P(c)$. Pour une nouvelle valeur de $x$, on compare la hauteur des deux courbes : la plus haute donne la classe prédite. Le seuil (pointillés) est l'endroit où elles se croisent. Avec plusieurs variables, le modèle **multiplie** ces densités comme si les variables étaient indépendantes. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/gestion) :** classer automatiquement des emails clients (réclamation, demande d'information, spam) ou analyser le sentiment d'avis produits (version multinomiale sur les mots).
- **Cas d'usage 2 (finance) :** filtre de première ligne pour signaler des transactions suspectes, quand il faut un modèle très rapide et peu de données d'entraînement.
- **Données et prérequis :** variables continues ~normales par classe (GaussianNB) ou comptages/fréquences de mots (MultinomialNB) ; pas de standardisation nécessaire.
- **Peu approprié :** scoring de crédit avec variables fortement corrélées (revenu, patrimoine, montant emprunté) : l'hypothèse d'indépendance « compte deux fois » la même information et les probabilités deviennent trop extrêmes — inutilisables pour fixer un prix du risque.
- **Métriques :** F1, précision, rappel, ROC-AUC (utile car le classement peut rester bon même si les probabilités sont mal calibrées), courbe de calibration.

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** valeurs par défaut (`var_smoothing=1e-9`, a priori estimés sur les fréquences). Pas de standardisation, inutile pour ce modèle.
- **Sortie obtenue :** **F1 ≈ 0,755**, le plus faible des classifieurs de ce notebook.
- **Conclusion possible :** `make_classification` génère chaque classe comme **plusieurs amas gaussiens** (`n_clusters_per_class=2` par défaut) avec des variables informatives corrélées : la distribution par classe n'est ni une seule gaussienne ni à variables indépendantes. L'hypothèse du modèle est violée, d'où la performance moindre.
- **Limites :** on ne peut pas en déduire que Naive Bayes est « mauvais » en général — il excelle sur le texte. Sans matrice de confusion ni ROC-AUC, on ne sait pas si le classement des probabilités reste exploitable avec un autre seuil.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Arbre de décision** (*Decision Tree Classifier*, CART) — famille : **classification supervisée**. Objectif : prédire une classe au moyen d'une suite de règles « si… alors ».

**Intuition et étapes :** données étiquetées → à chaque nœud, recherche de la variable et du seuil qui rendent les sous-groupes les plus **purs** (une seule classe dominante) → découpage récursif jusqu'à la profondeur maximale → une observation descend dans l'arbre et reçoit la classe majoritaire de sa feuille.

$$G(t)=1-\sum_{c=1}^{C}p_{c,t}^2\qquad \Delta G=G(t)-\frac{n_g}{n_t}G(t_g)-\frac{n_d}{n_t}G(t_d)$$

$G(t)$ : impureté de Gini du nœud $t$ ; $p_{c,t}$ : proportion de la classe $c$ dans le nœud ; $C$ : nombre de classes ; $n_t,n_g,n_d$ : effectifs du nœud et de ses enfants gauche/droite ; $\Delta G$ : gain d'une coupure, que l'on maximise. **Paramètres appris :** variables, seuils et classes des feuilles. **Hyperparamètres :** `max_depth`, `min_samples_leaf`, `criterion` (gini/entropie), `ccp_alpha` (élagage).

**Avantages :** (1) très interprétable, visualisable et explicable à un non-spécialiste ; (2) gère variables mixtes et interactions, sans standardisation.
**Limites :** (1) instable et sujet au surapprentissage s'il est profond ; (2) frontières en escalier parallèles aux axes, souvent moins performant qu'un ensemble.

### Schéma explicatif

```python
# Schéma 11 — Arbre de décision : règles lisibles et frontière en rectangles (illustration)
from sklearn.tree import DecisionTreeClassifier, plot_tree
rng=np.random.default_rng(GRAINE)
Xs=rng.uniform(0,10,(200,2)); ys=((Xs[:,0]>4)&(Xs[:,1]>3)|(Xs[:,0]>8)).astype(int)
ys=np.where(rng.random(200)<.08,1-ys,ys)                 # 8 % de bruit
arbre=DecisionTreeClassifier(max_depth=2,random_state=GRAINE).fit(Xs,ys)
fig,ax=plt.subplots(1,2,figsize=(13,4.5))
plot_tree(arbre,feature_names=['endettement','ancienneté'],class_names=['sain','défaut'],filled=True,ax=ax[0],fontsize=8)
xx,yy=np.meshgrid(np.linspace(0,10,300),np.linspace(0,10,300))
ax[1].contourf(xx,yy,arbre.predict(np.c_[xx.ravel(),yy.ravel()]).reshape(xx.shape),alpha=.3,cmap='coolwarm')
ax[1].scatter(Xs[:,0],Xs[:,1],c=ys,cmap='coolwarm',s=12,edgecolor='k');ax[1].set_xlabel('endettement');ax[1].set_ylabel('ancienneté')
ax[1].set_title('Zones de décision : rectangles alignés sur les axes');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** à gauche, chaque nœud affiche sa règle, son indice de Gini, le nombre d'observations et la répartition par classe ; la couleur indique la classe majoritaire et son intensité la pureté. À droite, les mêmes règles découpent le plan en **rectangles** : chaque rectangle est une feuille. Les noms « endettement / ancienneté » ne servent qu'à rendre l'exemple concret ; les données sont simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** grille d'octroi de crédit simple et auditable (« si taux d'endettement > 40 % et ancienneté < 1 an → refus »), exigée pour la transparence réglementaire.
- **Cas d'usage 2 (marketing) :** segmentation explicative des clients qui répondent à une offre, pour construire des règles de ciblage compréhensibles par les équipes commerciales.
- **Données et prérequis :** variables mixtes, aucune standardisation ; contrôle de la profondeur et de la taille minimale des feuilles ; classes pondérées si déséquilibre.
- **Peu approprié :** lorsque la performance maximale prime sur l'interprétabilité (ex. détection de fraude à grande échelle) : un arbre seul est instable et dominé par les forêts ou le boosting ; peu adapté aussi aux frontières obliques (combinaisons linéaires de variables).
- **Métriques :** F1, précision, rappel, ROC-AUC ; écart entraînement/validation pour diagnostiquer le surapprentissage.

```python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `max_depth=4` (au plus 16 feuilles), critère de Gini, `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,891**, comparable au KNN (≈ 0,889) et supérieur à la logistique (≈ 0,811).
- **Conclusion possible :** même peu profond, l'arbre capte les interactions et la structure en amas des classes mieux qu'une frontière linéaire. Contrairement à la régression (exemple 5), la tâche de classification avec frontières par morceaux convient mieux à un arbre.
- **Limites :** un arbre unique est instable : sur un autre découpage, la valeur pourrait varier de plusieurs points ; `max_depth=4` n'a pas été validé ; le F1 n'indique pas la nature des erreurs.

## 12. Forêt aléatoire — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Forêt aléatoire — classification** (*Random Forest Classifier*) — famille : **classification supervisée, ensemble par bagging**. Objectif : obtenir un classifieur robuste en combinant de nombreux arbres décorrélés.

**Intuition et étapes :** données → $B$ échantillons bootstrap → un arbre profond par échantillon, chaque coupure ne considérant qu'un sous-ensemble aléatoire de $m$ variables ($m\approx\sqrt p$ par défaut) → chaque arbre vote → classe finale = **vote majoritaire** (ou moyenne des probabilités).

$$\hat y(x)=\arg\max_c\frac1B\sum_{b=1}^{B}\hat p_b(c\mid x)$$

$B$ (`n_estimators`) : nombre d'arbres ; $\hat p_b(c\mid x)$ : proportion de la classe $c$ dans la feuille de l'arbre $b$ atteinte par $x$. **Hyperparamètres :** $B$, `max_features` ($m$), `max_depth`, `min_samples_leaf`, `class_weight`. **Paramètres appris :** les arbres.

**Avantages :** (1) très bonne performance « prête à l'emploi », robuste au bruit et au surapprentissage ; (2) importance des variables et estimation *out-of-bag* sans jeu de validation séparé.
**Limites :** (1) boîte noire relative (des centaines de règles) ; (2) coûteux en mémoire et en temps de prédiction pour de très grandes forêts.

### Schéma explicatif

```python
# Schéma 12 — Forêt aléatoire : vote de plusieurs arbres vs frontière d'un arbre unique (illustration)
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_moons
Xs,ys=make_moons(n_samples=250,noise=.3,random_state=GRAINE)
xx,yy=np.meshgrid(np.linspace(-2,3,250),np.linspace(-1.5,2,250)); G=np.c_[xx.ravel(),yy.ravel()]
modeles=[('Arbre unique (profond)',DecisionTreeClassifier(random_state=GRAINE)),('Forêt de 100 arbres',RandomForestClassifier(n_estimators=100,random_state=GRAINE))]
fig,ax=plt.subplots(1,2,figsize=(12,4))
for a,(nom,m) in zip(ax,modeles):
    m.fit(Xs,ys); P=m.predict_proba(G)[:,1].reshape(xx.shape)
    a.contourf(xx,yy,P,levels=20,cmap='RdBu_r',alpha=.6);a.scatter(Xs[:,0],Xs[:,1],c=ys,cmap='bwr',edgecolor='white',linewidth=.5,s=14);a.set_title(nom)
fig.suptitle('Couleur = proportion d’arbres votant pour la classe rouge');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** à gauche, un arbre profond découpe le plan en zones nettes, avec des « îlots » créés pour isoler des points bruités : **surapprentissage**. À droite, la forêt moyenne 100 arbres : la couleur indique la part d'arbres votant pour la classe rouge ; la frontière est plus lisse et les zones incertaines (couleurs pâles) apparaissent là où les classes se mélangent réellement. Données simulées (`make_moons`).

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** prédiction du défaut de paiement des titulaires de cartes de crédit (variables de paiement, montants, historique).
- **Cas d'usage 2 (logistique/commerce) :** prédire si une commande sera livrée en retard (transporteur, distance, période, stock) pour alerter le client.
- **Données et prérequis :** données tabulaires mixtes, pas de standardisation ; plusieurs centaines d'observations minimum ; `class_weight='balanced'` en cas de classes déséquilibrées.
- **Peu approprié :** décision qui doit être justifiée règle par règle à un régulateur ou un client (un arbre simple ou une logistique est plus défendable), ou prédiction à très faible latence sur matériel limité (des centaines d'arbres à parcourir).
- **Métriques :** F1, rappel, précision, ROC-AUC / PR-AUC selon le déséquilibre ; score *out-of-bag* ; importance par permutation.

```python
from sklearn.ensemble import RandomForestClassifier
modele=RandomForestClassifier(n_estimators=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `n_estimators=80`, profondeur non limitée, `max_features='sqrt'` (défaut), `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,918**, au-dessus de l'arbre unique (≈ 0,891).
- **Conclusion possible :** l'agrégation d'arbres décorrélés améliore la généralisation par rapport à un arbre seul, conformément à la théorie du bagging.
- **Limites :** l'écart de ≈ 0,03 de F1 sur 100 observations correspond à quelques prédictions seulement, il faudrait une validation croisée pour le confirmer ; pas de réglage d'hyperparamètres ; pas de mesure du coût de calcul ni de l'importance des variables.

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Machine à vecteurs de support** (*Support Vector Machine*, SVM ; classifieur `SVC`) — famille : **classification supervisée**. Objectif : trouver la frontière qui sépare les classes avec la **plus grande marge** possible.

**Intuition et étapes :** variables standardisées → recherche de l'hyperplan qui maximise la distance aux points les plus proches de chaque classe (les **vecteurs de support**) en tolérant quelques erreurs → avec un **noyau** (ex. RBF), cette séparation se fait dans un espace transformé, ce qui donne une frontière courbe dans l'espace d'origine → classe = signe de la fonction de décision.

$$\min_{w,b,\xi}\ \frac12\lVert w\rVert^2+C\sum_{i=1}^{n}\xi_i\quad\text{s.c. } y_i(w^\top\phi(x_i)+b)\ge1-\xi_i,\ \xi_i\ge0\qquad K(x,x')=e^{-\gamma\lVert x-x'\rVert^2}$$

$w,b$ : orientation et décalage de l'hyperplan (**paramètres appris**, exprimés via les vecteurs de support) ; $y_i\in\{-1,+1\}$ : classe ; $\xi_i$ : variable d'écart (violation de marge) ; $\phi$ : transformation implicite définie par le noyau $K$ ; $C$ : **hyperparamètre** de compromis marge/erreurs ; $\gamma$ : **hyperparamètre** de portée du noyau RBF ; la marge vaut $2/\lVert w\rVert$.

**Avantages :** (1) très efficace en dimension élevée et avec des frontières complexes (noyaux) ; (2) solution dépendant seulement des vecteurs de support, robuste.
**Limites :** (1) coût d'entraînement qui croît fortement avec $n$ (≫ 50 000 lignes devient lent) ; (2) pas de probabilités natives et réglage sensible de $C$ et $\gamma$.

### Schéma explicatif

```python
# Schéma 13 — SVM : marge maximale et vecteurs de support (illustration, noyau linéaire puis RBF)
from sklearn.svm import SVC
from sklearn.datasets import make_blobs, make_circles
Xa,ya=make_blobs(n_samples=60,centers=2,cluster_std=1.1,random_state=6)
Xb,yb=make_circles(n_samples=150,factor=.4,noise=.08,random_state=GRAINE)
fig,ax=plt.subplots(1,2,figsize=(12,4.5))
for a,(Xs,ys,k,tit) in zip(ax,[(Xa,ya,'linear','Noyau linéaire : marge maximale'),(Xb,yb,'rbf','Noyau RBF : frontière courbe')]):
    m=SVC(kernel=k,C=1.0).fit(Xs,ys)
    xx,yy=np.meshgrid(np.linspace(Xs[:,0].min()-.5,Xs[:,0].max()+.5,250),np.linspace(Xs[:,1].min()-.5,Xs[:,1].max()+.5,250))
    Z=m.decision_function(np.c_[xx.ravel(),yy.ravel()]).reshape(xx.shape)
    a.contour(xx,yy,Z,levels=[-1,0,1],colors='k',linestyles=['--','-','--'])
    a.scatter(Xs[:,0],Xs[:,1],c=ys,cmap='coolwarm',s=18,edgecolor='k')
    a.scatter(*m.support_vectors_.T,s=110,facecolors='none',edgecolors='green',lw=1.5,label='vecteurs de support')
    a.set_title(tit);a.legend(loc='lower right')
plt.tight_layout();plt.show()
```

**Comment lire le schéma :** la ligne pleine est la frontière de décision ; les pointillés délimitent la **marge** (fonction de décision = ±1). Les points entourés en vert sont les **vecteurs de support** : ce sont les seuls qui déterminent la frontière — supprimer un autre point ne la modifierait pas. À droite, aucune droite ne sépare les deux cercles ; le noyau RBF produit une frontière circulaire. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** prédiction de faillite d'entreprises à partir de quelques dizaines de ratios financiers sur un échantillon de taille moyenne (quelques milliers de sociétés).
- **Cas d'usage 2 (marketing) :** classification de textes (avis clients positifs/négatifs) en grande dimension (représentation TF-IDF), domaine où le SVM linéaire est historiquement très performant.
- **Données et prérequis :** variables numériques **standardisées** (obligatoire), jeu de taille petite à moyenne ; réglage de $C$ et $\gamma$ par validation croisée (GridSearch).
- **Peu approprié :** scoring de millions de transactions (temps d'entraînement quadratique à cubique en $n$), ou besoin de probabilités fiables pour tarifer un risque (il faut une calibration supplémentaire).
- **Métriques :** F1, précision, rappel, ROC-AUC (via `decision_function`) ; accuracy si classes équilibrées.

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

- **Hyperparamètres :** `StandardScaler` → `SVC(kernel='rbf', C=1.0)`, `gamma='scale'` par défaut. (`random_state` n'a d'effet que si `probability=True`.)
- **Sortie obtenue :** **F1 ≈ 0,898**, proche de l'arbre (≈ 0,891) et du KNN (≈ 0,889), au-dessus de la logistique (≈ 0,811).

- **Limites :** $C$ et $\gamma$ ne sont pas réglés (valeurs par défaut) — le SVM est très sensible à ces choix ; pas de probabilités donc pas de ROC-AUC calculée ; données synthétiques et petit test.

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Gradient Boosting — classification** (*Gradient Boosting Classifier*, version histogramme `HistGradientBoostingClassifier`) — famille : **classification supervisée, ensemble séquentiel**. Objectif : prédire une classe en additionnant des petits arbres qui corrigent successivement les erreurs.

**Intuition et étapes :** score initial (log-odds de la fréquence de la classe positive) → calcul du gradient de la log-perte (≈ $y_i-p_i$, l'écart entre la classe réelle et la probabilité prédite) → ajustement d'un petit arbre sur ces gradients → ajout pondéré par $\eta$ → répétition → probabilité finale via la sigmoïde.

$$F_m(x)=F_{m-1}(x)+\eta\,h_m(x),\qquad p(x)=\sigma(F_M(x)),\qquad r_{im}=y_i-\sigma(F_{m-1}(x_i))$$

$F_m$ : score (log-odds) après $m$ arbres ; $h_m$ : arbre ajusté aux pseudo-résidus $r_{im}$ ; $\eta$ (`learning_rate`) ; $M$ (`max_iter`) ; $\sigma$ : sigmoïde. **Hyperparamètres :** $\eta$, $M$, `max_leaf_nodes`, `max_depth`, `l2_regularization`, `early_stopping`. **Paramètres appris :** les arbres et leurs valeurs de feuilles.

**Avantages :** (1) performances de premier plan sur données tabulaires ; (2) version Hist rapide sur grands volumes, gestion native des valeurs manquantes.
**Limites :** (1) sensible aux hyperparamètres, surapprentissage possible ; (2) peu interprétable (nécessite SHAP ou importance par permutation).

### Schéma explicatif

```python
# Schéma 14 — Gradient boosting en classification : la frontière se construit par itérations (illustration)
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.datasets import make_moons
Xs,ys=make_moons(n_samples=300,noise=.25,random_state=GRAINE)
xx,yy=np.meshgrid(np.linspace(-2,3,250),np.linspace(-1.5,2,250)); G=np.c_[xx.ravel(),yy.ravel()]
fig,ax=plt.subplots(1,3,figsize=(15,4))
for a,M in zip(ax,[1,10,100]):
    m=HistGradientBoostingClassifier(max_iter=M,learning_rate=.3,max_depth=2,random_state=GRAINE).fit(Xs,ys)
    a.contourf(xx,yy,m.predict_proba(G)[:,1].reshape(xx.shape),levels=20,cmap='RdBu_r',alpha=.6)
    a.scatter(Xs[:,0],Xs[:,1],c=ys,cmap='bwr',edgecolor='white',linewidth=.5,s=12);a.set_title(f'M = {M} arbre(s)')
fig.suptitle('Couleur = probabilité prédite de la classe rouge ; chaque arbre ajouté affine la frontière');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** de gauche à droite, le même modèle avec 1, 10 puis 100 petits arbres (profondeur 2). Avec un seul arbre, la frontière est un simple découpage grossier ; à 10 arbres, la forme en croissants apparaît ; à 100, la frontière épouse les données (avec un début de surapprentissage localement). Le nombre d'arbres est donc un levier de complexité à contrôler par validation. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** détection de fraude à la carte bancaire ou probabilité de défaut, avec des milliers de variables comportementales.
- **Cas d'usage 2 (marketing) :** prédiction de l'attrition client (*churn*) dans les télécoms ou la banque, pour cibler les actions de rétention.
- **Données et prérequis :** données tabulaires de taille moyenne à grande ; validation (ou `early_stopping`) pour fixer $M$ ; `class_weight` si déséquilibre ; découpage chronologique si les données sont datées.
- **Peu approprié :** petits échantillons (quelques centaines de lignes) avec exigence d'explicabilité forte — par exemple une grille de score réglementaire Bâle — où une régression logistique est plus robuste et justifiable.
- **Métriques :** F1, précision, rappel, ROC-AUC, PR-AUC pour classes rares ; courbe d'apprentissage en fonction de $M$.

```python
from sklearn.ensemble import HistGradientBoostingClassifier
modele=HistGradientBoostingClassifier(max_iter=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `max_iter=80`, `learning_rate=0.1` et `max_leaf_nodes=31` par défaut, `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,950**, la **meilleure valeur** des classifieurs du notebook (devant XGBoost ≈ 0,931 et la forêt ≈ 0,918).
- **Conclusion possible :** la correction séquentielle des erreurs exploite bien les interactions et la structure non linéaire de ces données ; c'est cohérent avec la réputation du boosting sur données tabulaires.
- **Limites :** les écarts entre méthodes d'ensemble (0,92 à 0,95) portent sur quelques observations de test et ne sont pas forcément significatifs ; aucun réglage ni validation croisée ; pas de mesure du temps de calcul ni de la calibration.

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**AdaBoost** (*Adaptive Boosting*) — famille : **classification supervisée, ensemble séquentiel**. Objectif : combiner des classifieurs faibles (souvent des arbres à une seule coupure, *stumps*) en un classifieur fort.

**Intuition et étapes :** poids égaux pour toutes les observations → entraînement d'un *stump* → calcul de son erreur pondérée → attribution d'un poids de vote au *stump* (d'autant plus grand qu'il est précis) → **augmentation du poids des observations mal classées** → le *stump* suivant se concentre sur elles → vote pondéré final.

$$\varepsilon_m=\sum_i w_i\,\mathbb 1(h_m(x_i)\neq y_i),\quad \alpha_m=\tfrac12\ln\frac{1-\varepsilon_m}{\varepsilon_m},\quad w_i\leftarrow w_i\,e^{-\alpha_m y_i h_m(x_i)},\quad H(x)=\operatorname{sign}\Big(\sum_{m=1}^{M}\alpha_m h_m(x)\Big)$$

$w_i$ : poids (normalisé) de l'observation $i$ ; $h_m$ : classifieur faible à l'étape $m$ ; $y_i\in\{-1,+1\}$ ; $\varepsilon_m$ : erreur pondérée ; $\alpha_m$ : poids de vote du classifieur (**paramètre appris**) ; $H$ : classifieur final. **Hyperparamètres :** $M$ (`n_estimators`), `learning_rate`, type de classifieur faible.

**Avantages :** (1) simple, peu d'hyperparamètres, transforme des modèles très faibles en modèle performant ; (2) les poids des observations révèlent les cas « difficiles ».
**Limites :** (1) très sensible au bruit et aux étiquettes erronées (il s'acharne sur elles) ; (2) en général moins performant que le gradient boosting moderne.

### Schéma explicatif

```python
# Schéma 15 — AdaBoost : repondération des observations mal classées (illustration)
from sklearn.tree import DecisionTreeClassifier
rng=np.random.default_rng(GRAINE)
Xs=rng.uniform(-1,1,(60,2)); ys=np.where(Xs[:,0]**2+Xs[:,1]**2<.45,1,-1)
w=np.full(60,1/60); fig,ax=plt.subplots(1,3,figsize=(15,4))
xx,yy=np.meshgrid(np.linspace(-1,1,200),np.linspace(-1,1,200))
for m in range(3):
    h=DecisionTreeClassifier(max_depth=1).fit(Xs,ys,sample_weight=w); pred=h.predict(Xs)
    eps=w[pred!=ys].sum(); alpha=.5*np.log((1-eps)/eps)
    ax[m].contourf(xx,yy,h.predict(np.c_[xx.ravel(),yy.ravel()]).reshape(xx.shape),alpha=.2,cmap='coolwarm')
    ax[m].scatter(Xs[:,0],Xs[:,1],c=ys,cmap='coolwarm',s=w*60*40,edgecolor='k')
    ax[m].set_title(f'Étape {m+1} : erreur ε={eps:.2f}, vote α={alpha:.2f}')
    w=w*np.exp(-alpha*ys*pred); w/=w.sum()
fig.suptitle('Taille des points = poids avant l’étape ; les erreurs grossissent à l’étape suivante');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** chaque panneau montre un *stump* (une seule coupure verticale ou horizontale, fond coloré) entraîné avec les poids courants. La **taille des points** représente leur poids : à l'étape 1 tous sont égaux ; les points mal classés grossissent ensuite, forçant le *stump* suivant à les traiter. Le titre donne l'erreur pondérée $\varepsilon$ et le poids de vote $\alpha$ de chaque *stump*. Aucun *stump* seul ne sépare le disque, mais leur vote pondéré s'en approche. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** classification de demandes de crédit en « accordé / refusé » sur un jeu propre et bien étiqueté, comme référence de boosting simple.
- **Cas d'usage 2 :** détection de visages en temps réel (algorithme de Viola-Jones, application historique majeure d'AdaBoost).
- **Données et prérequis :** données tabulaires **propres** (étiquettes fiables, peu de valeurs aberrantes) ; classifieur faible légèrement meilleur que le hasard ; pas de standardisation avec des arbres.
- **Peu approprié :** données avec beaucoup d'étiquettes erronées — par exemple des retours clients classés manuellement de façon incohérente — car l'algorithme augmente sans cesse le poids de ces cas, ce qui dégrade le modèle.
- **Métriques :** F1, précision, rappel, ROC-AUC ; erreur de validation en fonction du nombre d'estimateurs.

```python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `n_estimators=70` *stumps* (arbres de profondeur 1 par défaut), `learning_rate=1.0`, `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,827** — mieux que la logistique (≈ 0,811) et Naive Bayes (≈ 0,755), mais nettement moins bien que le gradient boosting (≈ 0,950) et la forêt (≈ 0,918).
- **Conclusion possible :** des *stumps* ne modélisent chacun qu'une coupure sur une variable ; même combinés, ils capturent mal les interactions entre variables présentes dans ces données, alors que les arbres plus profonds du gradient boosting les captent.
- **Limites :** profondeur du classifieur faible et nombre d'estimateurs non réglés ; écart observé sur un seul découpage de 100 observations ; aucun test de sensibilité au bruit.

## 16. XGBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**XGBoost** (*eXtreme Gradient Boosting*) — famille : **classification supervisée** (et régression), **gradient boosting régularisé**. Objectif : une implémentation optimisée et régularisée du boosting d'arbres, rapide et performante.

**Intuition et étapes :** identique au gradient boosting (arbres ajoutés séquentiellement sur les erreurs) mais : approximation de second ordre de la perte (gradient $g_i$ **et** hessien $h_i$), pénalité explicite sur la complexité des arbres, recherche de coupures parallélisée et gestion des valeurs manquantes.

$$\mathcal{L}^{(m)}=\sum_{i=1}^{n}\ell\big(y_i,\hat y_i^{(m-1)}+f_m(x_i)\big)+\Omega(f_m),\qquad \Omega(f)=\gamma T+\tfrac12\lambda\sum_{j=1}^{T}w_j^2$$

$\ell$ : perte (log-perte en classification) ; $\hat y^{(m-1)}$ : prédiction après $m-1$ arbres ; $f_m$ : nouvel arbre ; $T$ : nombre de feuilles ; $w_j$ : valeur de la feuille $j$ (**paramètres appris**) ; $\gamma$ (`gamma`), $\lambda$ (`reg_lambda`) : **hyperparamètres** de pénalité. Autres hyperparamètres : `n_estimators`, `max_depth`, `learning_rate`, `subsample`, `colsample_bytree`.

**Avantages :** (1) très performant et rapide (parallélisme, GPU), référence des compétitions sur données tabulaires ; (2) régularisation intégrée et gestion native des valeurs manquantes.
**Limites :** (1) nombreux hyperparamètres, réglage long ; (2) interprétation difficile et risque de surapprentissage si mal paramétré.

### Schéma explicatif

```python
# Schéma 16 — XGBoost : effet de la régularisation et de la profondeur sur la complexité (illustration)
from xgboost import XGBClassifier
from sklearn.datasets import make_moons
Xs,ys=make_moons(n_samples=300,noise=.3,random_state=GRAINE)
xx,yy=np.meshgrid(np.linspace(-2,3,250),np.linspace(-1.5,2,250)); G=np.c_[xx.ravel(),yy.ravel()]
configs=[('max_depth=6, λ=0 (peu contraint)',dict(max_depth=6,reg_lambda=0,gamma=0)),
         ('max_depth=3, λ=1 (défaut proche du notebook)',dict(max_depth=3,reg_lambda=1)),
         ('max_depth=2, λ=20, γ=2 (très régularisé)',dict(max_depth=2,reg_lambda=20,gamma=2))]
fig,ax=plt.subplots(1,3,figsize=(15,4))
for a,(t,p) in zip(ax,configs):
    m=XGBClassifier(n_estimators=150,learning_rate=.2,random_state=GRAINE,**p).fit(Xs,ys)
    a.contourf(xx,yy,m.predict_proba(G)[:,1].reshape(xx.shape),levels=20,cmap='RdBu_r',alpha=.6)
    a.scatter(Xs[:,0],Xs[:,1],c=ys,cmap='bwr',edgecolor='white',linewidth=.5,s=12);a.set_title(t,fontsize=10)
fig.suptitle('La pénalité Ω (γ, λ) et la profondeur contrôlent la complexité de la frontière');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** trois modèles XGBoost entraînés sur les mêmes données. À gauche, des arbres profonds sans pénalité créent une frontière très irrégulière (îlots autour de points isolés). Au centre, une configuration modérée. À droite, des arbres peu profonds fortement pénalisés donnent une frontière lisse, au risque de sous-apprendre. La régularisation $\Omega$ est ce qui distingue XGBoost du boosting classique. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** modèles de scoring crédit ou de détection de fraude en production dans les banques et fintechs (grands volumes, variables nombreuses).
- **Cas d'usage 2 (logistique/commerce) :** prévision de la probabilité de rupture de stock ou de retard de livraison à partir de données opérationnelles.
- **Données et prérequis :** données tabulaires, valeurs manquantes tolérées ; recherche d'hyperparamètres (validation croisée ou jeu de validation avec `early_stopping_rounds`) ; installation de la bibliothèque `xgboost`.
- **Peu approprié :** problèmes sur images, son ou texte brut (réseaux de neurones plus adaptés) ; ou très petits échantillons avec exigence d'explication simple, où le gain par rapport à une logistique ne justifie pas la complexité.
- **Métriques :** F1, précision, rappel, ROC-AUC, PR-AUC ; log-loss (`eval_metric`) suivie au fil des itérations.

```python
from xgboost import XGBClassifier
modele=XGBClassifier(n_estimators=80,max_depth=3,random_state=GRAINE,eval_metric='logloss').fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python
- Entrées : `Xc_ent` (300 × 6) / `Xc_test` (100 × 6) issues de `make_classification` (400 lignes, 6 variables dont 4 informatives, 2 classes), découpage stratifié 75/25 — test équilibré 50/50. Données **synthétiques**.
- **Hyperparamètres :** `n_estimators=80`, `max_depth=3`, `eval_metric='logloss'` (perte suivie), `learning_rate=0.3` et `reg_lambda=1` par défaut de la bibliothèque, `random_state=42`.
- **Sortie obtenue :** **F1 ≈ 0,931**, deuxième meilleur résultat, juste derrière `HistGradientBoostingClassifier` (≈ 0,950) et devant la forêt (≈ 0,918).
- **Conclusion possible :** comme les autres méthodes de boosting d'arbres, XGBoost capte bien la structure non linéaire ; sur ce petit jeu, ses optimisations (vitesse, régularisation) n'apportent pas d'avantage décisif par rapport à l'implémentation de scikit-learn.
- **Limites :** l'écart 0,931 / 0,950 correspond à environ deux erreurs de test et n'est pas interprétable comme une supériorité ; les hyperparamètres ne sont pas réglés ; l'intérêt de XGBoost (grands volumes) n'est pas testé ici.

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**K-moyennes** (*K-means*) — famille : **clustering (non supervisé)**. Objectif : partitionner les observations en $K$ groupes homogènes autour de centres (centroïdes).

**Intuition et étapes :** données sans étiquette (standardisées) → initialisation de $K$ centroïdes (k-means++) → **affectation** de chaque point au centroïde le plus proche → **mise à jour** de chaque centroïde comme moyenne de ses points → répétition jusqu'à stabilité → étiquette de groupe pour chaque observation.

$$\min_{\{C_k\},\{\mu_k\}}\ \text{WCSS}=\sum_{k=1}^{K}\sum_{x_i\in C_k}\lVert x_i-\mu_k\rVert^2$$

$K$ (`n_clusters`) : nombre de groupes — **hyperparamètre** ; $C_k$ : ensemble des points du groupe $k$ ; $\mu_k$ : centroïde du groupe $k$ — **paramètre appris** ; $\lVert\cdot\rVert$ : distance euclidienne ; WCSS : inertie intra-classe. Autre hyperparamètre : `n_init` (nombre d'initialisations, on garde la meilleure).

**Avantages :** (1) simple, rapide et adapté aux grands volumes ; (2) résultat facile à interpréter via les centroïdes (« profil type » de chaque groupe).
**Limites :** (1) $K$ doit être fixé à l'avance ; (2) suppose des groupes sphériques de tailles comparables, sensible aux valeurs extrêmes et à l'échelle des variables.

### Schéma explicatif

```python
# Schéma 17 — K-means : alternance affectation / mise à jour des centroïdes (illustration)
rng=np.random.default_rng(7)
P=np.vstack([rng.normal(c,.6,(40,2)) for c in [(0,0),(4,1),(2,4)]])
C=np.array([[0.,3.],[1.,1.5],[4.5,3.5]])            # initialisation volontairement mauvaise
fig,ax=plt.subplots(1,4,figsize=(16,3.8))
for t in range(4):
    lab=((P[:,None]-C[None])**2).sum(2).argmin(1)
    ax[t].scatter(P[:,0],P[:,1],c=lab,cmap='viridis',s=12)
    ax[t].scatter(C[:,0],C[:,1],marker='X',s=200,c='red',edgecolor='k',label='centroïdes')
    nouv=np.array([P[lab==k].mean(0) for k in range(3)])
    for k in range(3): ax[t].annotate('',nouv[k],C[k],arrowprops=dict(arrowstyle='->',color='red'))
    ax[t].set_title(f'Itération {t+1}'); C=nouv
ax[0].legend();fig.suptitle('Couleur = groupe affecté ; flèche = déplacement du centroïde vers la moyenne de son groupe');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** chaque panneau est une itération. Les points sont colorés selon le centroïde (croix rouge) le plus proche : c'est l'étape d'**affectation**. Les flèches montrent l'étape de **mise à jour** : chaque centroïde se déplace vers la moyenne de ses points. Partis d'une initialisation volontairement mauvaise, les centroïdes convergent en quelques itérations vers les centres des trois nuages, et les flèches deviennent de plus en plus courtes. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** segmentation de la clientèle à partir de variables RFM (récence, fréquence, montant des achats) pour adapter les offres — par ex. sur le jeu *Online Retail* (UCI).
- **Cas d'usage 2 (logistique) :** regrouper des points de livraison géographiquement proches pour définir des zones de tournée ou l'emplacement d'entrepôts.
- **Données et prérequis :** variables numériques **standardisées**, valeurs extrêmes traitées ; choix de $K$ par méthode du coude, silhouette ou connaissance métier.
- **Peu approprié :** groupes de formes allongées, imbriquées ou de densités très différentes (ex. trajectoires de transactions frauduleuses) — K-means coupe arbitrairement ces formes ; préférer DBSCAN ou mélanges gaussiens. Peu adapté aussi aux variables purement catégorielles (moyenne sans sens → K-modes).
- **Métriques :** silhouette, inertie (coude), indice de Davies-Bouldin ; stabilité entre graines ; ARI si des étiquettes réelles existent (évaluation externe uniquement).

```python
from sklearn.cluster import KMeans
modele=KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(X_clu)
print('Silhouette:',silhouette_score(X_clu,modele.labels_))
plt.scatter(X_clu[:,0],X_clu[:,1],c=modele.labels_,s=12);plt.title('K-means');plt.show()
```

- **Sortie obtenue :** **silhouette ≈ 0,869** et un nuage de points coloré en trois groupes.
- **Conclusion possible :** une silhouette proche de 1 indique que chaque point est en moyenne beaucoup plus proche de son groupe que du groupe voisin : la partition est nette. C'est attendu car les données sont exactement dans le cas idéal de K-means (groupes sphériques, de même variance et taille).
- **Limites :** la silhouette mesure la cohérence géométrique, pas l'utilité métier ; $K$ n'a pas été choisi (pas de comparaison de plusieurs $K$) ; les étiquettes de référence n'ont pas servi à une évaluation externe (ARI) ; aucune conclusion possible sur des données réelles plus désordonnées.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Classification hiérarchique ascendante** (*Agglomerative Hierarchical Clustering*, CAH) — famille : **clustering (non supervisé)**. Objectif : construire une hiérarchie de regroupements, du point isolé jusqu'à un groupe unique.

**Intuition et étapes :** chaque observation forme d'abord son propre groupe → calcul des distances entre groupes → **fusion des deux groupes les plus proches** selon un critère de liaison → répétition jusqu'à un seul groupe → l'arbre obtenu (**dendrogramme**) est coupé à la hauteur qui donne le nombre de groupes souhaité.

$$\Delta(A,B)=\frac{|A|\,|B|}{|A|+|B|}\,\lVert\mu_A-\mu_B\rVert^2\quad\text{(critère de Ward)}$$

$A,B$ : deux groupes candidats à la fusion ; $|A|,|B|$ : leurs effectifs ; $\mu_A,\mu_B$ : leurs centres ; $\Delta$ : augmentation de l'inertie intra-classe provoquée par la fusion (on fusionne la paire qui l'augmente le moins). **Hyperparamètres :** `n_clusters` (ou `distance_threshold`), `linkage` (ward, complete, average, single), la métrique. **Pas de paramètres appris** au sens strict : le résultat est la hiérarchie de fusions.

**Avantages :** (1) pas besoin de fixer $K$ à l'avance : le dendrogramme montre toutes les partitions possibles ; (2) résultat déterministe et visuellement interprétable.
**Limites :** (1) coût mémoire/temps élevé ($O(n^2)$ distances) → inadapté à plus de quelques dizaines de milliers de points ; (2) une fusion n'est jamais remise en cause, sensible au bruit selon la liaison choisie.

### Schéma explicatif

```python
# Schéma 18 — CAH : dendrogramme et coupure en groupes (illustration)
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
rng=np.random.default_rng(GRAINE)
P=np.vstack([rng.normal(c,.5,(8,2)) for c in [(0,0),(4,0),(2,3.5)]])
Z=linkage(P,method='ward'); lab=fcluster(Z,t=3,criterion='maxclust')
fig,ax=plt.subplots(1,2,figsize=(13,4))
ax[0].scatter(P[:,0],P[:,1],c=lab,cmap='viridis',s=40,edgecolor='k')
for i,(x,y) in enumerate(P): ax[0].annotate(str(i),(x,y),fontsize=7,xytext=(3,3),textcoords='offset points')
ax[0].set_title('24 observations et les 3 groupes obtenus')
dendrogram(Z,ax=ax[1],color_threshold=Z[-3,2]+.01); seuil=(Z[-3,2]+Z[-2,2])/2
ax[1].axhline(seuil,c='red',ls='--',label='coupure → 3 groupes');ax[1].set_xlabel('indice de l’observation');ax[1].set_ylabel('distance de fusion (Ward)')
ax[1].legend();ax[1].set_title('Dendrogramme');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** à droite, chaque feuille en bas est une observation (numéros identiques au nuage de gauche). Deux branches qui se rejoignent indiquent une fusion ; la **hauteur** de la jonction mesure le coût de la fusion (critère de Ward). Les très grands sauts de hauteur signalent que l'on fusionne des groupes vraiment différents. Couper l'arbre à la ligne rouge donne 3 groupes, coloriés de la même façon à gauche. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** regrouper des actions ou des fonds selon la corrélation de leurs rendements (taxonomie de portefeuille, *Hierarchical Risk Parity*).
- **Cas d'usage 2 (commerce international) :** construire une typologie de pays selon leurs indicateurs économiques (PIB par habitant, ouverture commerciale, inflation — données Banque mondiale) et lire les niveaux de proximité.
- **Données et prérequis :** variables numériques standardisées, **échantillon de taille modérée** (≤ quelques milliers de lignes) ; choix de la liaison et de la distance.
- **Peu approprié :** segmentation de plusieurs millions de clients d'un site e-commerce : la matrice de distances $n\times n$ ne tient pas en mémoire ; K-means ou MiniBatchKMeans sont plus adaptés.
- **Métriques :** silhouette, coefficient cophénétique (fidélité du dendrogramme aux distances), lecture des sauts de hauteur ; ARI si étiquettes réelles disponibles.

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

### Analyse de l'expérience Python
- Entrées : `X_clu` (240 × 2) généré par `make_blobs` avec 3 centres et `cluster_std=0.85` — trois groupes sphériques, bien séparés et de même taille. Données **synthétiques** ; les étiquettes de référence existent mais ne sont pas utilisées par le code.
- **Hyperparamètres :** `n_clusters=3`, liaison `ward` et distance euclidienne par défaut.
- **Sortie obtenue :** **silhouette ≈ 0,869**, valeur **identique** à K-means (exemple 17).
- **Conclusion possible :** une silhouette identique à toutes les décimales indique que les deux méthodes produisent la **même partition** (à la numérotation près). C'est logique : Ward minimise aussi l'inertie intra-classe, et des groupes aussi bien séparés ne laissent pas d'ambiguïté.
- **Limites :** le dendrogramme, principal intérêt de la méthode, n'est pas affiché dans l'exemple ; l'égalité avec K-means est propre à ce cas facile et ne se généralise pas ; pas de comparaison entre critères de liaison.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**DBSCAN** (*Density-Based Spatial Clustering of Applications with Noise*) — famille : **clustering (non supervisé) par densité**, utilisable aussi pour repérer des anomalies. Objectif : trouver des groupes de forme quelconque comme zones denses séparées par des zones vides, et isoler le **bruit**.

**Intuition et étapes :** pour chaque point, compter ses voisins dans un rayon $\varepsilon$ → un point ayant au moins `min_samples` voisins est un **point cœur** → les points cœurs voisins sont reliés en un même groupe, qui absorbe les **points frontières** atteignables → les points restants sont étiquetés **bruit** (−1).

$$N_\varepsilon(x)=\{x'\in X:\ d(x,x')\le\varepsilon\},\qquad x\ \text{cœur}\iff |N_\varepsilon(x)|\ge \text{min\_samples}$$

$N_\varepsilon(x)$ : voisinage de rayon $\varepsilon$ autour de $x$ ; $d$ : distance (euclidienne par défaut) ; $|\cdot|$ : nombre de points (incluant $x$). **Hyperparamètres :** $\varepsilon$ (`eps`) et `min_samples`. **Pas de paramètres appris** : le nombre de groupes est un **résultat**, non une entrée.

**Avantages :** (1) découvre le nombre de groupes et des formes non convexes ; (2) identifie explicitement les points aberrants (bruit).
**Limites :** (1) très sensible au choix de $\varepsilon$ ; (2) échoue quand les groupes ont des densités très différentes et se dégrade en grande dimension.

### Schéma explicatif

```python
# Schéma 19 — DBSCAN : points cœurs, frontières et bruit (illustration)
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
Xs,_=make_moons(n_samples=200,noise=.07,random_state=GRAINE)
Xs=np.vstack([Xs,np.random.default_rng(GRAINE).uniform([-1.5,-1],[2.5,1.5],(12,2))])
db=DBSCAN(eps=.2,min_samples=5).fit(Xs); lab=db.labels_
coeur=np.zeros(len(Xs),bool); coeur[db.core_sample_indices_]=True
fig,ax=plt.subplots(figsize=(8,4.5))
for k in set(lab)-{-1}:
    ax.scatter(*Xs[(lab==k)&coeur].T,s=30,label=f'groupe {k} — points cœurs');ax.scatter(*Xs[(lab==k)&~coeur].T,s=30,marker='^',label=f'groupe {k} — frontières')
ax.scatter(*Xs[lab==-1].T,c='k',marker='x',s=50,label='bruit (−1)')
ax.add_patch(plt.Circle(Xs[db.core_sample_indices_[0]],.2,fill=False,ls='--',color='grey'))
ax.set_title('Deux groupes non convexes retrouvés sans fixer K ; cercle pointillé = rayon ε');ax.legend(fontsize=8,loc='lower left');plt.show()
```

**Comment lire le schéma :** les deux croissants sont retrouvés sans préciser leur nombre, ce que K-means ne saurait pas faire (il couperait les croissants en deux par une droite). Les ronds sont des **points cœurs** (≥ 5 voisins dans le cercle de rayon $\varepsilon$, dessiné en pointillé pour un point), les triangles des **points frontières** rattachés à un groupe, et les croix noires le **bruit**, isolé hors des zones denses. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique/commerce) :** repérer des zones de forte concentration de clients ou de livraisons à partir de coordonnées GPS (formes irrégulières suivant les rues), en ignorant les adresses isolées.
- **Cas d'usage 2 (finance) :** détecter des transactions atypiques qui n'appartiennent à aucun groupe dense de comportements habituels (étiquette −1 = alerte à examiner).
- **Données et prérequis :** variables numériques standardisées, dimension faible à moyenne ; choix de $\varepsilon$ via le graphe des distances au k-ième voisin.
- **Peu approprié :** segmentation de clients où un groupe premium très dispersé côtoie un groupe grand public très dense : un seul $\varepsilon$ ne convient pas aux deux densités (l'un est fragmenté ou l'autre absorbé) → HDBSCAN ou mélanges gaussiens. Peu adapté aussi en grande dimension.
- **Métriques :** nombre de groupes et proportion de bruit, silhouette calculée **hors bruit**, stabilité selon $\varepsilon$ ; ARI si étiquettes disponibles.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

### Analyse de l'expérience Python
- Entrées : `X_clu` (240 × 2) généré par `make_blobs` avec 3 centres et `cluster_std=0.85` — trois groupes sphériques, bien séparés et de même taille. Données **synthétiques** ; les étiquettes de référence existent mais ne sont pas utilisées par le code.
- **Hyperparamètres :** `eps=0.8`, `min_samples=5`.
- **Sortie obtenue :** étiquettes `{0, 1, 2, -1}` → **3 groupes** retrouvés et **8 points** classés comme bruit (≈ 3,3 % des 240 points).
- **Conclusion possible :** sans connaître le nombre de groupes, DBSCAN retrouve bien les 3 nuages. Les 8 points de bruit sont des points situés en périphérie des nuages gaussiens, où la densité passe sous le seuil (< 5 voisins dans un rayon de 0,8) : ce ne sont pas des anomalies « réelles », puisque toutes les données ont été générées par les trois groupes.
- **Limites :** aucun indicateur de qualité n'est calculé (pas de silhouette hors bruit) ; le résultat dépend fortement de `eps` — une valeur plus petite fragmenterait les groupes, une plus grande les fusionnerait ; le nombre de points de bruit n'a de sens que relativement à ces réglages.

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Modèle de mélange gaussien** (*Gaussian Mixture Model*, GMM) — famille : **clustering probabiliste (non supervisé)**, aussi estimation de densité. Objectif : modéliser les données comme un mélange de $K$ lois normales et attribuer à chaque point une **probabilité d'appartenance** à chaque groupe.

**Intuition et étapes :** initialisation des $K$ gaussiennes → **étape E** : calcul, pour chaque point, de la probabilité d'avoir été généré par chaque composante (responsabilités) → **étape M** : réestimation des poids, moyennes et covariances à partir de ces responsabilités → répétition (algorithme EM) jusqu'à convergence de la vraisemblance.

$$p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal N(x\mid\mu_k,\Sigma_k),\qquad \gamma_{ik}=\frac{\pi_k\mathcal N(x_i\mid\mu_k,\Sigma_k)}{\sum_{l}\pi_l\mathcal N(x_i\mid\mu_l,\Sigma_l)}$$

$\pi_k$ : poids de la composante $k$ ($\sum_k\pi_k=1$) ; $\mu_k$ : moyenne ; $\Sigma_k$ : matrice de covariance — tous **paramètres appris** ; $\mathcal N$ : densité normale multivariée ; $\gamma_{ik}$ : responsabilité (probabilité que $x_i$ appartienne au groupe $k$). **Hyperparamètres :** $K$ (`n_components`), `covariance_type` (full, diag, spherical, tied), `n_init`.

**Avantages :** (1) affectation « douce » (probabilités) et groupes elliptiques de tailles et orientations différentes ; (2) modèle génératif : permet aussi de calculer la densité d'un point (détection d'anomalies).
**Limites :** (1) suppose des groupes approximativement gaussiens, $K$ à fixer (choisi via BIC/AIC) ; (2) sensible à l'initialisation et coûteux en grande dimension (covariances).

### Schéma explicatif

```python
# Schéma 20 — Mélange gaussien : ellipses de covariance et appartenance probabiliste (illustration)
from sklearn.mixture import GaussianMixture
from matplotlib.patches import Ellipse
rng=np.random.default_rng(GRAINE)
P=np.vstack([rng.multivariate_normal([0,0],[[2,1.5],[1.5,1.5]],150),rng.multivariate_normal([4,0],[[.4,0],[0,2]],100)])
g=GaussianMixture(n_components=2,random_state=GRAINE).fit(P); proba=g.predict_proba(P)[:,0]
fig,ax=plt.subplots(figsize=(8,4.5))
sc=ax.scatter(P[:,0],P[:,1],c=proba,cmap='coolwarm',s=14);plt.colorbar(sc,label='probabilité d’appartenir à la 1re composante')
for k in range(2):
    val,vec=np.linalg.eigh(g.covariances_[k]); ang=np.degrees(np.arctan2(vec[1,1],vec[0,1]))
    for n in (1,2): ax.add_patch(Ellipse(g.means_[k],2*n*np.sqrt(val[1]),2*n*np.sqrt(val[0]),angle=ang,fill=False,lw=1.5,ls='-' if n==1 else '--'))
    ax.annotate(f'π={g.weights_[k]:.2f}',g.means_[k],fontsize=11,weight='bold',bbox=dict(fc='white',alpha=.8))
ax.set_title('Ellipses = 1 et 2 écarts-types de chaque gaussienne ; couleur intermédiaire = appartenance incertaine');plt.show()
```

**Comment lire le schéma :** chaque composante gaussienne est représentée par deux ellipses (1 et 2 écarts-types) dont la forme et l'orientation viennent de sa matrice de covariance — ici un groupe incliné et un groupe vertical, ce que K-means ne saurait représenter. La couleur de chaque point donne sa **probabilité** d'appartenir à la composante 1 : bleu/rouge francs = appartenance sûre, teintes intermédiaires = zone de recouvrement. Les poids $\pi$ indiquent la part estimée de chaque groupe. Données simulées.

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** identifier des **régimes de marché** (calme / volatil) à partir des rendements et de la volatilité d'un indice, avec une probabilité d'être dans chaque régime à chaque date.
- **Cas d'usage 2 (marketing) :** segmentation douce de clients lorsqu'un client peut partager les caractéristiques de plusieurs segments (probabilités d'appartenance utilisées pour personnaliser les offres).
- **Données et prérequis :** variables numériques continues, standardisées, groupes approximativement elliptiques ; suffisamment d'observations par composante pour estimer les covariances ; choix de $K$ par BIC.
- **Peu approprié :** données à forte dimension et peu d'observations (ex. 500 variables pour 200 clients) : les matrices de covariance ne sont pas estimables de façon fiable ; ou groupes de formes très non gaussiennes (croissants, anneaux).
- **Métriques :** BIC/AIC, log-vraisemblance, silhouette (sur l'affectation la plus probable), entropie des responsabilités ; ARI si étiquettes réelles disponibles.

```python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
```

### Analyse de l'expérience Python
- Entrées : `X_clu` (240 × 2) généré par `make_blobs` avec 3 centres et `cluster_std=0.85` — trois groupes sphériques, bien séparés et de même taille. Données **synthétiques** ; les étiquettes de référence existent mais ne sont pas utilisées par le code.
- **Hyperparamètres :** `n_components=3`, `covariance_type='full'` par défaut, `random_state=42`.
- **Sortie obtenue :** poids estimés **[0,333 ; 0,333 ; 0,333]** et **silhouette ≈ 0,869** (sur l'affectation la plus probable).
- **Conclusion possible :** le modèle estime trois composantes de poids égaux (≈ 80 points chacune sur 240), ce qui correspond exactement à la génération par `make_blobs` (groupes de même taille). La silhouette identique à K-means et à la CAH montre que les trois méthodes donnent la même partition sur ce cas idéal.
- **Limites :** les probabilités d'appartenance, principal apport du GMM, ne sont pas exploitées dans la sortie ; $K$ n'a pas été choisi par BIC ; avec des groupes aussi séparés, l'avantage des covariances complètes n'apparaît pas.

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Analyse en composantes principales** (*Principal Component Analysis*, ACP / PCA) — famille : **réduction de dimension (non supervisée, linéaire)**. Objectif : résumer $p$ variables corrélées par un petit nombre de nouvelles variables non corrélées (les composantes) qui conservent le maximum de variance.

**Intuition et étapes :** variables centrées-réduites → calcul de la matrice de covariance (ou SVD) → recherche des directions de variance maximale, orthogonales entre elles (vecteurs propres) → projection des observations sur les $k$ premières directions.

$$\Sigma=\frac{1}{n-1}X^\top X,\qquad \Sigma v_j=\lambda_j v_j,\qquad Z=XV_k,\qquad \text{part de variance}_j=\frac{\lambda_j}{\sum_{l=1}^{p}\lambda_l}$$

$X$ : données centrées-réduites ($n\times p$) ; $\Sigma$ : matrice de covariance ; $v_j$ : $j$-ième axe principal (vecteur propre, **paramètre appris**) ; $\lambda_j$ : variance portée par cet axe (valeur propre) ; $V_k$ : matrice des $k$ premiers axes ; $Z$ : coordonnées projetées. **Hyperparamètre :** $k$ (`n_components`, nombre ou part de variance à conserver).

**Avantages :** (1) rapide, déterministe, composantes interprétables via leurs coefficients (cercle des corrélations) ; (2) réduit bruit et multicolinéarité avant un autre modèle.
**Limites :** (1) ne capte que des structures **linéaires** ; (2) sensible à l'échelle et aux valeurs extrêmes, et la variance maximale n'est pas forcément l'information utile pour une cible.

### Schéma explicatif

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**t-SNE** (*t-distributed Stochastic Neighbor Embedding*) — famille : **réduction de dimension non linéaire (non supervisée)**, utilisée essentiellement pour la **visualisation**. Objectif : représenter en 2D (ou 3D) des données de grande dimension en préservant les **voisinages locaux**.

**Intuition et étapes :** calcul, dans l'espace d'origine, de probabilités de voisinage $p_{ij}$ (gaussiennes, dont la largeur dépend de la perplexité) → initialisation aléatoire/ACP en 2D → calcul de similarités $q_{ij}$ en 2D avec une loi de Student (queues lourdes) → déplacement des points par descente de gradient pour que $q$ ressemble à $p$.

$$\mathrm{KL}(P\,\|\,Q)=\sum_{i\neq j}p_{ij}\ln\frac{p_{ij}}{q_{ij}},\qquad q_{ij}=\frac{(1+\lVert y_i-y_j\rVert^2)^{-1}}{\sum_{k\neq l}(1+\lVert y_k-y_l\rVert^2)^{-1}}$$

$p_{ij}$ : similarité entre les points $i$ et $j$ dans l'espace d'origine ; $q_{ij}$ : similarité entre leurs images $y_i, y_j$ en 2D ; KL : divergence de Kullback-Leibler minimisée. **Paramètres appris :** les coordonnées $y_i$ elles-mêmes. **Hyperparamètres :** `perplexity` (≈ nombre effectif de voisins, 5–50), `learning_rate`, `max_iter`, `random_state`.

**Avantages :** (1) révèle très bien les groupes et structures non linéaires ; (2) visualisations parlantes même en très grande dimension.
**Limites :** (1) les distances **entre groupes** et la taille des groupes n'ont pas de sens ; (2) non déterministe, lent ($O(n^2)$ ou $O(n\log n)$), et pas de `transform` pour de nouveaux points.

### Schéma explicatif

```python
# Schéma 22 — t-SNE : effet de la perplexité sur la carte obtenue (jeu vin, coloré par cépage)
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
vin=load_wine(); Xv=StandardScaler().fit_transform(vin.data)
fig,ax=plt.subplots(1,4,figsize=(17,3.8))
p=PCA(2).fit_transform(Xv); ax[0].scatter(p[:,0],p[:,1],c=vin.target,cmap='viridis',s=10); ax[0].set_title('Référence : ACP (linéaire)')
for a,perp in zip(ax[1:],[3,20,60]):
    e=TSNE(n_components=2,perplexity=perp,random_state=GRAINE).fit_transform(Xv)
    a.scatter(e[:,0],e[:,1],c=vin.target,cmap='viridis',s=10); a.set_title(f't-SNE, perplexité = {perp}')
fig.suptitle('Couleur = cépage réel (non utilisé par l’algorithme, seulement pour lire la carte)');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** chaque point est un vin, coloré par son cépage réel — information **non fournie à l'algorithme**, ajoutée uniquement pour juger la carte. Le premier panneau montre l'ACP pour comparaison. Avec une perplexité faible (3), la carte est plus morcelée et des points d'un même cépage se dispersent en petits îlots (structure trop locale) ; avec 20, les trois cépages forment des groupes nets ; avec 60, la carte se rapproche d'une vue plus globale et les frontières entre groupes se resserrent. Seuls les **voisinages** sont fiables : la distance entre deux groupes ou leur taille apparente ne doivent pas être interprétées.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** visualiser des milliers de clients décrits par des dizaines de variables comportementales pour repérer visuellement des segments avant une classification formelle.
- **Cas d'usage 2 (finance) :** explorer des transactions ou des entreprises décrites par de nombreux ratios afin de voir si les cas frauduleux ou défaillants forment des îlots distincts.
- **Données et prérequis :** variables numériques standardisées, souvent après une ACP préalable (≈ 50 composantes) pour réduire le bruit et le coût ; quelques centaines à quelques dizaines de milliers de points.
- **Peu approprié :** comme étape de prétraitement d'un modèle de prédiction en production (pas de projection de nouvelles observations, résultat non stable) ou pour mesurer des distances économiques entre segments (les distances globales sont déformées).
- **Métriques :** *trustworthiness* (conservation des voisins), divergence KL finale ; évaluation visuelle de séparation avec étiquettes externes ; stabilité selon graine et perplexité.

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

### Analyse de l'expérience Python
- **Entrées :** `load_wine` standardisé (178 × 13).
- **Hyperparamètres :** `n_components=2`, `perplexity=20`, `random_state=42` (pour la reproductibilité, t-SNE étant stochastique).
- **Sortie obtenue :** uniquement un **nuage de points 2D**, sans métrique numérique.
- **Conclusion possible :** le graphique montre si des groupes de vins se distinguent visuellement dans les 13 variables chimiques (dans notre schéma coloré, trois groupes correspondant aux cépages apparaissent nettement).
- **Ce qu'on ne peut pas conclure :** aucune métrique n'est imprimée — on ne peut donc pas quantifier la qualité de la projection (pas de *trustworthiness*) ; le graphique du code n'étant pas coloré, on ne peut pas vérifier que les groupes visibles correspondent aux cépages ; distances et tailles entre groupes ne sont pas interprétables.

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**UMAP** (*Uniform Manifold Approximation and Projection*) — famille : **réduction de dimension non linéaire (non supervisée)**, pour la visualisation et le prétraitement. Objectif : projeter des données en faible dimension en préservant la structure locale **et**, mieux que t-SNE, une partie de la structure globale.

**Intuition et étapes :** construction d'un graphe des $k$ plus proches voisins dans l'espace d'origine, avec des poids de connexion flous → initialisation d'une représentation en 2D (souvent spectrale) → optimisation des coordonnées pour que le graphe en 2D ressemble au graphe d'origine (minimisation d'une entropie croisée floue).

$$\mathcal C=\sum_{(i,j)}\Big[w_{ij}\ln\frac{w_{ij}}{v_{ij}}+(1-w_{ij})\ln\frac{1-w_{ij}}{1-v_{ij}}\Big],\qquad v_{ij}=\big(1+a\lVert y_i-y_j\rVert^{2b}\big)^{-1}$$

$w_{ij}$ : poids de l'arête entre $i$ et $j$ dans le graphe d'origine ; $v_{ij}$ : poids correspondant en basse dimension ; $y_i$ : coordonnées projetées (**paramètres appris**) ; $a,b$ : constantes déduites de `min_dist`. **Hyperparamètres :** `n_neighbors` (local ↔ global), `min_dist` (compacité des groupes), `n_components`, `metric`.

**Avantages :** (1) plus rapide que t-SNE et passe mieux à l'échelle ; (2) conserve mieux la structure globale et permet de projeter de nouveaux points (`transform`).
**Limites :** (1) résultat sensible aux hyperparamètres et non déterministe sans graine ; (2) distances et densités dans la carte restent difficiles à interpréter, base théorique plus complexe.

### Schéma explicatif

```python
# Schéma 23 — UMAP : influence de n_neighbors (structure locale vs globale), jeu vin coloré par cépage
import umap, warnings; warnings.filterwarnings('ignore')
vin=load_wine(); Xv=StandardScaler().fit_transform(vin.data)
fig,ax=plt.subplots(1,3,figsize=(14,4))
for a,k in zip(ax,[3,15,80]):
    e=umap.UMAP(n_components=2,n_neighbors=k,random_state=GRAINE).fit_transform(Xv)
    a.scatter(e[:,0],e[:,1],c=vin.target,cmap='viridis',s=10); a.set_title(f'UMAP, n_neighbors = {k}')
fig.suptitle('Petit k : détails locaux, groupes fragmentés — grand k : vue plus globale (couleur = cépage, non utilisé par l’algorithme)')
plt.tight_layout();plt.show()
```

**Comment lire le schéma :** chaque panneau est une projection UMAP des 178 vins avec un nombre de voisins différent ; la couleur (cépage) sert uniquement à l'interprétation. Avec `n_neighbors=3`, l'algorithme ne regarde que des voisinages minuscules et la carte se fragmente ; avec 15 (valeur du notebook) ou 80, les trois cépages apparaissent comme des groupes cohérents et leur position relative devient plus stable. C'est le réglage principal entre fidélité locale et vision d'ensemble.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/commerce) :** cartographier les produits d'un catalogue à partir de leurs caractéristiques ou des co-achats pour repérer des familles de produits et des produits « à la frontière ».
- **Cas d'usage 2 (finance) :** réduire des centaines de variables comportementales de clients bancaires à quelques dimensions avant un clustering (UMAP puis HDBSCAN), ou visualiser la structure d'un portefeuille de crédits.
- **Données et prérequis :** variables numériques standardisées (ou métrique adaptée), quelques centaines à plusieurs millions de points ; bibliothèque `umap-learn` ; graine fixée.
- **Peu approprié :** quand il faut des axes **interprétables** économiquement (ex. présenter au comité de direction ce que signifie l'axe 1) : les dimensions UMAP n'ont pas de signification, contrairement aux composantes de l'ACP ; ou sur très petit échantillon (< 50 points) où le graphe de voisins est peu fiable.
- **Métriques :** *trustworthiness* / continuité, conservation des k plus proches voisins, stabilité selon graine et `n_neighbors` ; silhouette ou ARI si un clustering est réalisé ensuite.

```python
import umap
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=umap.UMAP(n_components=2,n_neighbors=15,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('UMAP');plt.show()
```

### Analyse de l'expérience Python
- **Entrées :** `load_wine` standardisé (178 × 13).
- **Hyperparamètres :** `n_components=2`, `n_neighbors=15`, `min_dist=0.1` par défaut, `random_state=42`. Un avertissement indique que fixer `random_state` désactive le calcul parallèle (`n_jobs=1`) : c'est le prix de la reproductibilité.
- **Sortie obtenue :** un **nuage de points 2D** intitulé « UMAP », sans métrique numérique.
- **Conclusion possible :** comme pour t-SNE, le graphique permet une exploration visuelle de la structure des 13 variables ; notre schéma coloré montre que les cépages forment des groupes distincts.
- **Ce qu'on ne peut pas conclure :** aucune mesure de conservation des voisinages n'est calculée ; sans coloration par cépage dans la cellule du code, la correspondance avec les classes réelles n'est pas vérifiable ; on ne peut pas classer UMAP et t-SNE sans métrique commune.

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Forêt d'isolement** (*Isolation Forest*) — famille : **détection d'anomalies (non supervisée)**. Objectif : attribuer à chaque observation un score d'anomalie, en partant de l'idée qu'une anomalie est **facile à isoler**.

**Intuition et étapes :** construction de nombreux arbres aléatoires sur des sous-échantillons : à chaque nœud, choix aléatoire d'une variable et d'un seuil entre son min et son max → une observation atypique se retrouve seule dans une feuille après **peu de coupures** → la profondeur moyenne d'isolement est transformée en score → les observations au score le plus élevé (proportion `contamination`) sont étiquetées −1.

$$s(x,n)=2^{-\frac{E[h(x)]}{c(n)}},\qquad c(n)=2H(n-1)-\frac{2(n-1)}{n}$$

$h(x)$ : longueur du chemin (nombre de coupures) pour isoler $x$ dans un arbre ; $E[h(x)]$ : moyenne sur les arbres ; $c(n)$ : longueur moyenne attendue pour un échantillon de taille $n$ (normalisation) ; $H$ : nombre harmonique ; $s$ proche de 1 → anomalie, proche de 0,5 ou moins → normal. **Paramètres appris :** les arbres aléatoires. **Hyperparamètres :** `n_estimators`, `max_samples`, `contamination` (part attendue d'anomalies, qui fixe le seuil).

**Avantages :** (1) rapide, linéaire en $n$, efficace en dimension modérée à élevée ; (2) ne nécessite ni étiquettes ni hypothèse de distribution.
**Limites :** (1) le seuil dépend de `contamination`, souvent inconnue ; (2) détecte mal les anomalies « locales » situées près ou à l'intérieur d'un groupe dense, et les scores sont difficiles à expliquer.

### Schéma explicatif

```python
# Schéma 24 — Isolation Forest : un point atypique s'isole en peu de coupures (illustration)
from sklearn.ensemble import IsolationForest
rng=np.random.default_rng(3)
P=np.vstack([rng.normal(0,1,(150,2)),[[4.5,4]]]); iso=IsolationForest(random_state=GRAINE).fit(P)
def isoler(pt,P,rng,lim):
    box=[P[:,0].min(),P[:,0].max(),P[:,1].min(),P[:,1].max()]; S=P.copy(); coupes=[]
    while len(S)>1 and len(coupes)<lim:
        j=rng.integers(2); s=rng.uniform(S[:,j].min(),S[:,j].max()); coupes.append((j,s)); S=S[S[:,j]<=s] if pt[j]<=s else S[S[:,j]>s]
    return coupes
fig,ax=plt.subplots(1,2,figsize=(13,4.5))
for a,pt,nom in [(ax[0],P[-1],'point atypique'),(ax[1],P[3],'point normal')]:
    c=isoler(pt,P,np.random.default_rng(1),40); a.scatter(P[:,0],P[:,1],s=10,c='grey')
    for j,s in c: (a.axvline if j==0 else a.axhline)(s,c='tab:blue',lw=.8,alpha=.6)
    a.scatter(*pt,c='red',s=120,edgecolor='k',zorder=3); a.set_title(f'{nom} : isolé en {len(c)} coupure'+('s' if len(c)>1 else ''))
sc=iso.score_samples(P); fig.suptitle(f'Score d’anomalie moyen : point atypique = {-sc[-1]:.2f}, points normaux = {-sc[:-1].mean():.2f} (plus élevé = plus anormal)')
plt.tight_layout();plt.show()
```

**Comment lire le schéma :** chaque ligne bleue est une coupure aléatoire (verticale ou horizontale) d'un arbre d'isolement. À gauche, le point rouge éloigné du nuage se retrouve seul après **très peu de coupures** ; à droite, un point au cœur du nuage en demande beaucoup plus. L'algorithme répète cette expérience sur de nombreux arbres et transforme la longueur moyenne du chemin en score d'anomalie (titre). Données simulées, un seul arbre illustré.

### Contextes d'utilisation
- **Cas d'usage 1 (finance/audit) :** repérer des écritures comptables ou des transactions bancaires inhabituelles (montant, heure, fréquence, contrepartie) pour prioriser les contrôles d'audit ou de lutte anti-fraude.
- **Cas d'usage 2 (logistique/marchés) :** détecter des journées anormales sur un indice boursier (rendement, volume, volatilité) ou des délais de livraison aberrants dans une chaîne d'approvisionnement.
- **Données et prérequis :** variables numériques (catégorielles encodées), pas besoin de standardisation stricte ; estimation de la proportion d'anomalies attendue ; un expert pour valider les alertes.
- **Peu approprié :** fraude sophistiquée qui imite le comportement normal (anomalies locales noyées dans un groupe dense) ou séries où l'anomalie est **contextuelle** (une vente élevée normale en décembre mais anormale en août) sans variables de contexte : l'algorithme ne la verrait pas.
- **Métriques :** sans étiquettes : distribution des scores, nombre et qualité des alertes jugées par un expert, stabilité entre graines ; avec étiquettes fiables : précision, rappel, PR-AUC.

```python
from sklearn.ensemble import IsolationForest
X_anomal=np.vstack([X_clu,np.array([[9,9],[-8,-8]])])
modele=IsolationForest(contamination=.02,random_state=GRAINE).fit(X_anomal)
print('Observations atypiques:',np.where(modele.predict(X_anomal)==-1)[0])
```

### Analyse de l'expérience Python
- **Entrées :** les 240 points de `X_clu` + **2 points ajoutés manuellement** : (9 ; 9) d'indice 240 et (−8 ; −8) d'indice 241 — 242 observations synthétiques.
- **Hyperparamètres :** `contamination=0.02` (≈ 2 % × 242 ≈ 5 points signalés), `n_estimators=100` par défaut, `random_state=42`.
- **Sortie obtenue :** observations atypiques **[20, 79, 130, 147, 240]**.
- **Conclusion possible :** le point (9 ; 9) (indice 240), très éloigné de tous les groupes, est bien détecté. En revanche le point (−8 ; −8) (indice 241) **n'est pas signalé** : il se trouve près du groupe centré vers (−6,9 ; −6,9) et ne s'isole donc pas facilement. Les 4 autres alertes (indices 20, 79, 130, 147) sont des points des groupes d'origine situés en bordure : ils sont signalés uniquement parce que `contamination` impose environ 5 alertes.
- **Limites :** le résultat illustre que le seuil est fixé par `contamination` et non par les données ; il n'y a pas d'étiquettes fiables pour calculer précision/rappel (on « sait » seulement que 2 points ont été ajoutés, et l'un d'eux n'est pas anormal au sens de l'algorithme) ; les scores eux-mêmes ne sont pas affichés.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Bandits manchots** (*Multi-Armed Bandits*) — famille : **apprentissage par renforcement** (cas simplifié sans état). Objectif : choisir, de façon répétée, entre plusieurs actions (« bras ») aux gains aléatoires de moyennes inconnues, pour **maximiser le gain cumulé**.

**Intuition et étapes :** à chaque tour $t$, l'agent choisit un bras → il observe une récompense aléatoire → il met à jour l'estimation de la valeur de ce bras (moyenne empirique) → il utilise ces estimations pour ses choix suivants. Tout le problème est l'arbitrage **exploration** (essayer pour mieux estimer) / **exploitation** (jouer le bras qui semble le meilleur). L'exemple du notebook utilise une politique de pure exploration (tour de rôle), comme référence.

$$\hat Q_t(a)=\frac{\sum_{s<t}r_s\,\mathbb 1(a_s=a)}{N_t(a)},\qquad \text{Regret}(T)=T\,\mu^*-\sum_{t=1}^{T}\mu_{a_t}$$

$a$ : bras ; $r_s$ : récompense reçue au tour $s$ ; $N_t(a)$ : nombre de fois où $a$ a été joué ; $\hat Q_t(a)$ : valeur estimée (**paramètre appris**) ; $\mu_a$ : vraie espérance de gain (inconnue) ; $\mu^*$ : celle du meilleur bras ; Regret : manque à gagner par rapport à l'oracle. **Hyperparamètres :** horizon $T$, règle de choix (tour de rôle, $\varepsilon$, UCB, Thompson…).

**Avantages :** (1) cadre simple et formel de la décision sous incertitude avec apprentissage en ligne ; (2) apprend pendant l'exploitation, sans jeu de données préalable — plus efficace qu'un test A/B classique.
**Limites :** (1) ignore l'état/le contexte et les conséquences futures d'une action ; (2) suppose des gains stationnaires — performance dégradée si les probabilités évoluent.

### Schéma explicatif

```python
# Schéma 25 — Bandits : convergence des estimations et regret d'une exploration uniforme (simulation)
rng=np.random.default_rng(GRAINE); mu=np.array([.2,.5,.7]); T=500
S=np.zeros(3);N=np.zeros(3);hist=[];regret=[]
for t in range(T):
    a=t%3; r=rng.random()<mu[a]; S[a]+=r;N[a]+=1; hist.append(S/np.maximum(N,1)); regret.append(mu.max()-mu[a])
hist=np.array(hist)
fig,ax=plt.subplots(1,2,figsize=(13,4))
for k,c in enumerate(['tab:blue','tab:orange','tab:green']):
    ax[0].plot(hist[:,k],c=c,label=f'bras {k} : estimation'); ax[0].axhline(mu[k],c=c,ls='--',lw=.8)
ax[0].set_xlabel('tour t');ax[0].set_ylabel('valeur estimée');ax[0].set_title('Estimations → vraies probabilités (pointillés)');ax[0].legend()
ax[1].plot(np.cumsum(regret));ax[1].set_xlabel('tour t');ax[1].set_ylabel('regret cumulé');ax[1].set_title('Explorer sans exploiter : regret qui croît linéairement')
plt.tight_layout();plt.show()
```

**Comment lire le schéma :** à gauche, les courbes pleines sont les estimations $\hat Q(a)$ de chaque bras quand on les joue à tour de rôle, et les pointillés leurs vraies probabilités (0,2 / 0,5 / 0,7) : les estimations se stabilisent autour des vraies valeurs. À droite, le **regret cumulé** (gain perdu par rapport au meilleur bras) augmente en ligne droite : en jouant les mauvais bras deux tiers du temps, on apprend bien mais on gagne mal. C'est précisément ce que les stratégies comme $\varepsilon$-greedy corrigent. Simulation.

- **Cas d'usage 1 (marketing) :** choisir en continu, parmi plusieurs versions d'une bannière publicitaire ou d'un objet d'email, celle qui génère le plus de clics, en réallouant le trafic au fil des résultats.
- **Cas d'usage 2 (finance/commerce) :** tester plusieurs niveaux de prix ou de remise en ligne (*dynamic pricing*) ; allocation de budget entre canaux d'acquisition.

Peu adapté aussi si la récompense n'est observée qu'après des mois (octroi de crédit et défaut).
- **Métriques :** récompense cumulée, regret cumulé, proportion de choix du meilleur bras, précision des estimations ; variation entre graines.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Stratégie ε-gloutonne** (*Epsilon-greedy*) — famille : **apprentissage par renforcement** (politique d'exploration, ici appliquée à un bandit). Objectif : équilibrer exploration et exploitation par une règle simple.

**Intuition et étapes :** à chaque tour, avec probabilité $\varepsilon$ on choisit un bras **au hasard** (exploration) ; sinon on choisit le bras de **meilleure valeur estimée** (exploitation) → on observe le gain → on met à jour la moyenne du bras joué. La même règle sert à choisir les actions dans Q-learning et SARSA.

$$a_t=\begin{cases}\text{bras aléatoire uniforme} & \text{avec probabilité } \varepsilon\\ \arg\max_a \hat Q_t(a) & \text{avec probabilité } 1-\varepsilon\end{cases}\qquad \hat Q(a)\leftarrow\hat Q(a)+\frac{1}{N(a)}\big(r-\hat Q(a)\big)$$

$\varepsilon\in[0,1]$ : **hyperparamètre** de taux d'exploration (fixe ou décroissant) ; $\hat Q(a)$ : valeur estimée du bras (**paramètre appris**) ; $N(a)$ : nombre de tirages ; $r$ : récompense observée. Dans le code, la valeur utilisée pour l'exploitation est $(S(a)+1)/(N(a)+1)$ — un a priori optimiste qui pousse à essayer chaque bras au début.

**Avantages :** (1) très simple à implémenter et à expliquer ; (2) garantit une exploration minimale de toutes les actions, donc une adaptation partielle aux changements.
**Limites :** (1) explore à l'aveugle, même les bras manifestement mauvais (part $\varepsilon$ de choix gaspillés indéfiniment si $\varepsilon$ est fixe) ; (2) la performance dépend fortement du choix de $\varepsilon$.

### Schéma explicatif

```python
# Schéma 26 — ε-greedy : effet de epsilon sur le gain moyen et la part de choix optimaux (simulation, 20 graines)
mu=np.array([.2,.5,.7]); T=1000
def jouer(eps,graine):
    g=np.random.default_rng(graine);S=np.zeros(3);N=np.zeros(3);gains=[];opt=[]
    for t in range(T):
        a=int(g.integers(3)) if g.random()<eps else int(np.argmax((S+1)/(N+1)))
        r=int(g.random()<mu[a]);S[a]+=r;N[a]+=1;gains.append(r);opt.append(a==2)
    return np.cumsum(gains)/np.arange(1,T+1),np.cumsum(opt)/np.arange(1,T+1)
fig,ax=plt.subplots(1,2,figsize=(13,4))
for eps in [0,.01,.1,.5]:
    R=np.array([jouer(eps,s) for s in range(20)])
    ax[0].plot(R[:,0].mean(0),label=f'ε={eps}');ax[1].plot(R[:,1].mean(0),label=f'ε={eps}')
ax[0].axhline(.7,c='k',ls=':',label='optimum 0,7');ax[0].set_title('Gain moyen cumulé');ax[1].set_title('Part des tours jouant le meilleur bras')
for a in ax: a.set_xlabel('tour t');a.legend()
plt.tight_layout();plt.show()
```

**Comment lire le schéma :** chaque courbe est la moyenne sur 20 simulations indépendantes pour une valeur de $\varepsilon$. À gauche, le gain moyen par tour (l'optimum est 0,7 si l'on jouait toujours le meilleur bras) ; à droite, la proportion de tours où le meilleur bras est joué. $\varepsilon=0,5$ explore trop et plafonne bas ; $\varepsilon=0$ peut rester bloqué sur un bras sous-optimal dans certaines simulations ; une valeur intermédiaire offre en général le meilleur compromis. Simulation.

### Contextes d'utilisation
- **Cas d'usage 1 (marketing/e-commerce) :** recommandation de produits ou de contenus : afficher la plupart du temps l'article le plus performant et, de temps en temps, un article nouveau pour mesurer son potentiel.
- **Cas d'usage 2 (logistique/finance) :** choix répété d'un transporteur ou d'un fournisseur parmi plusieurs (coût et délai incertains), ou politique d'exploration dans un agent Q-learning de gestion de stock.
- **Données et prérequis :** environnement interactif avec retour rapide de la récompense ; actions en nombre limité ; choix et éventuelle décroissance de $\varepsilon$.
- **Peu approprié :** situations où une exploration au hasard est coûteuse ou risquée — par ex. tester au hasard un taux d'intérêt très défavorable sur de vrais clients bancaires, ou un dosage en santé : on préfère des méthodes d'exploration dirigée (UCB, Thompson sampling) ou des simulations.
- **Métriques :** récompense cumulée et regret, part des choix optimaux, variance entre graines, vitesse de convergence.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Analyse de l'expérience Python
- **Entrées :** même bandit **simulé** à 3 bras (0,2 / 0,5 / 0,7), graine 42.
- **Hyperparamètres :** $\varepsilon=0,1$ fixe, 1 000 tours, valeur d'exploitation $(S+1)/(N+1)$.
- **Sortie obtenue :** nombre de choix **[25 ; 49 ; 926]** et valeurs estimées **[0,200 ; 0,673 ; 0,693]**.
- **Conclusion possible :** l'agent a identifié le meilleur bras et l'a joué **92,6 %** des tours (contre 33 % pour le tour de rôle de l'exemple 25) : le gain cumulé est donc bien meilleur. Les ≈ 74 choix sur les autres bras proviennent essentiellement de l'exploration aléatoire (≈ 10 % de 1 000 = 100 tirages aléatoires, dont environ deux tiers tombent sur un autre bras).
- **Ce qu'on ne peut pas conclure :** les bras peu joués sont **mal estimés** — le bras 1 est estimé à 0,673 au lieu de 0,5 avec seulement 49 tirages : exploiter réduit la précision sur les options délaissées. Une seule graine est utilisée, donc la robustesse n'est pas évaluée ; le gain total n'est pas affiché.

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Q-learning** (apprentissage Q, *Q-learning*) — famille : **apprentissage par renforcement, sans modèle, hors-politique** (*off-policy*). Objectif : apprendre la valeur de chaque couple (état, action) pour en déduire la politique optimale.

**Intuition et étapes :** l'agent observe un état $s$ → choisit une action $a$ ($\varepsilon$-greedy) → reçoit une récompense $r$ et arrive en $s'$ → met à jour $Q(s,a)$ vers « récompense immédiate + meilleure valeur future possible » → répète sur de nombreux épisodes → politique finale : dans chaque état, l'action de plus grande valeur $Q$.

$$Q(s,a)\leftarrow Q(s,a)+\alpha\Big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\Big]$$

$s,s'$ : état courant et suivant ; $a$ : action choisie ; $r$ : récompense ; $Q(s,a)$ : valeur estimée de l'action $a$ dans l'état $s$ (**paramètres appris**, tableau états × actions) ; $\alpha$ : **taux d'apprentissage** ; $\gamma\in[0,1]$ : **facteur d'actualisation** (poids du futur) ; $\max_{a'}$ : meilleure action dans $s'$, quelle que soit celle que l'agent jouera réellement (d'où « hors-politique »). **Hyperparamètres :** $\alpha$, $\gamma$, $\varepsilon$ et sa décroissance, nombre d'épisodes.

**Avantages :** (1) converge vers la politique optimale (sous conditions d'exploration suffisante) sans connaître les règles de l'environnement ; (2) simple et interprétable en version tabulaire.
**Limites :** (1) table $Q$ impraticable quand les états sont nombreux ou continus (→ DQN) ; (2) apprentissage lent avec récompenses rares, et risque de surestimation des valeurs à cause du $\max$.

### Schéma explicatif

```python
# Schéma 27 — Q-learning : boucle agent–environnement et politique apprise sur FrozenLake (schéma + carte)
import gymnasium as gym
from matplotlib.patches import FancyBboxPatch
fig,ax=plt.subplots(1,2,figsize=(14,4.5))
a=ax[0];a.axis('off');a.set_xlim(0,10);a.set_ylim(0,6)
for (x,y,t,c) in [(1,2.2,'AGENT\ntable Q(s,a)\npolitique ε-greedy','#cfe2f3'),(6.5,2.2,'ENVIRONNEMENT\nFrozenLake 4×4','#d9ead3')]:
    a.add_patch(FancyBboxPatch((x,y),2.8,1.8,boxstyle='round,pad=.1',fc=c,ec='k'));a.text(x+1.4,y+.9,t,ha='center',va='center',fontsize=10)
a.annotate('',(6.4,3.6),(3.9,3.6),arrowprops=dict(arrowstyle='->',lw=2));a.text(5.1,3.85,'action a',ha='center')
a.annotate('',(3.9,2.4),(6.4,2.4),arrowprops=dict(arrowstyle='->',lw=2));a.text(5.1,1.9,'état s\', récompense r',ha='center')
a.text(5,.6,'mise à jour : Q(s,a) ← Q(s,a) + α[r + γ·max Q(s\',·) − Q(s,a)]',ha='center',fontsize=10,bbox=dict(fc='lightyellow'))
a.set_title('Boucle agent–environnement')
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));g=np.random.default_rng(GRAINE)
for ep in range(4000):
    s,_=env.reset(seed=GRAINE+ep)
    for _ in range(100):
        act=int(g.integers(4)) if g.random()<max(.05,1-ep/2500) else int(np.argmax(Q[s]))
        s2,r,te,tr,_=env.step(act);Q[s,act]+=.3*(r+(0 if te else .95*Q[s2].max())-Q[s,act]);s=s2
        if te or tr:break
carte=np.array([list(l.decode()) for l in env.unwrapped.desc.ravel()]).reshape(4,4); env.close()
couleur={'S':'#cfe2f3','F':'white','H':'#555555','G':'#b6d7a8'}; fl={0:(-.3,0),1:(0,.3),2:(.3,0),3:(0,-.3)}
b=ax[1]
for i in range(4):
    for j in range(4):
        e=i*4+j;b.add_patch(plt.Rectangle((j-.5,i-.5),1,1,fc=couleur[carte[i,j]],ec='k'))
        b.text(j-.42,i-.32,str(e),fontsize=8,color='red' if carte[i,j]=='H' else 'k')
        if carte[i,j] in 'SF': dx,dy=fl[int(Q[e].argmax())];b.arrow(j-dx/2,i-dy/2,dx,dy,head_width=.12,color='tab:blue')
b.set_xlim(-.5,3.5);b.set_ylim(3.5,-.5);b.set_xticks([]);b.set_yticks([]);b.set_aspect('equal')
b.set_title('Politique apprise : flèche = argmax Q (gris = trou, vert = but)');plt.tight_layout();plt.show()
```

**Comment lire le schéma :** à gauche, le cycle du renforcement : l'agent envoie une action, l'environnement renvoie le nouvel état et une récompense, et l'agent corrige sa table $Q$ avec la règle encadrée. À droite, la grille FrozenLake (états numérotés de 0 à 15) avec, dans chaque case praticable, une flèche indiquant l'action de plus grande valeur apprise. En suivant les flèches depuis la case 0 on doit atteindre la case verte (15) en évitant les trous gris. Simulation Gymnasium, mêmes réglages que l'exemple.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique/gestion) :** politique de réapprovisionnement d'un stock : état = niveau de stock et demande récente, action = quantité commandée, récompense = marge − coûts de stockage et de rupture.
- **Cas d'usage 2 (finance) :** stratégie d'exécution d'un gros ordre de bourse découpé dans le temps (état discrétisé : quantité restante, temps restant ; action : quantité à vendre) — dans un simulateur de marché.
- **Données et prérequis :** un **environnement** (réel ou simulé) avec états et actions **discrets et peu nombreux**, récompenses définies ; beaucoup d'épisodes ; réglage de $\alpha$, $\gamma$, $\varepsilon$.
- **Peu approprié :** gestion de portefeuille avec des états continus de grande dimension (prix, volatilités, positions de 50 actifs) : la table Q serait gigantesque et presque jamais remplie → approximation par réseau (DQN). Inadapté aussi lorsque l'on ne peut pas expérimenter (pas de simulateur, erreurs coûteuses).
- **Métriques :** récompense cumulée par épisode, taux de réussite (part d'épisodes atteignant le but), nombre de pas, vitesse de convergence, variation entre graines.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python
- **Entrées :** Environnement **FrozenLake-v1** de Gymnasium, version déterministe (`is_slippery=False`) : grille 4 × 4 (états 0 à 15, départ en 0, but en 15, trous en 5, 7, 11 et 12), 4 actions (0 = gauche, 1 = bas, 2 = droite, 3 = haut), récompense 1 uniquement en atteignant le but. Ce sont des trajectoires **simulées par interaction**, pas un dataset réel.
- **Hyperparamètres :** 4 000 épisodes, ≤ 100 pas par épisode, $\alpha=0,3$, $\gamma=0,95$, $\varepsilon$ décroissant linéairement de 1 à 0,05 sur 2 500 épisodes, graines fixées.
- **Sortie obtenue :** politique `[1 2 1 0 1 0 1 0 2 1 1 0 0 2 2 0]` (action choisie dans chaque état 0 à 15).
- **Conclusion possible :** en suivant la politique depuis le départ : 0 →(bas) 4 →(bas) 8 →(droite) 9 →(bas) 13 →(droite) 14 →(droite) **15 = but**. Ce chemin de 6 pas (le minimum possible) évite tous les trous : l'agent a appris une politique optimale. Les valeurs dans les états 5, 7, 11, 12 (trous) et 15 (but) valent 0 car ce sont des états terminaux où aucune action n'est apprise — l'`argmax` renvoie 0 par défaut et n'a pas de sens.
- **Ce qu'on ne peut pas conclure :** aucun taux de réussite ni récompense cumulée n'est mesuré ; l'environnement non glissant est déterministe et très petit — le succès ne prouve rien sur la version glissante (`is_slippery=True`) ni sur un problème réel ; une seule graine.

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**SARSA** (*State–Action–Reward–State–Action*) — famille : **apprentissage par renforcement, sans modèle, sur-politique** (*on-policy*). Objectif : apprendre la valeur $Q(s,a)$ **de la politique réellement suivie** (exploration comprise).

**Intuition et étapes :** identique au Q-learning, à une différence près : pour la cible, on utilise la valeur de l'**action suivante effectivement choisie** $a'$ (par la même règle $\varepsilon$-greedy), et non le maximum. Le nom reprend le quintuplet utilisé à chaque mise à jour : $(s, a, r, s', a')$.

$$Q(s,a)\leftarrow Q(s,a)+\alpha\Big[r+\gamma\,Q(s',a')-Q(s,a)\Big]$$

$s, a$ : état et action courants ; $r$ : récompense ; $s'$ : état suivant ; $a'$ : action suivante tirée par la politique $\varepsilon$-greedy ; $Q$ : table des valeurs (**paramètres appris**) ; $\alpha$, $\gamma$, $\varepsilon$ : **hyperparamètres** (mêmes rôles que pour le Q-learning).

**Avantages :** (1) plus prudent : comme il tient compte des erreurs d'exploration, il apprend des trajectoires plus sûres quand explorer est risqué ; (2) mises à jour souvent plus stables (moins de surestimation que le $\max$).
**Limites :** (1) converge vers la politique optimale seulement si $\varepsilon$ décroît vers 0 ; (2) mêmes limites tabulaires que le Q-learning (états discrets et peu nombreux) et apprentissage plus lent si l'exploration est forte.

### Schéma explicatif

```python
# Schéma 28 — Q-learning vs SARSA : la cible de mise à jour diffère (schéma) + courbes d'apprentissage (simulation)
import gymnasium as gym
fig,ax=plt.subplots(1,2,figsize=(14,4.5))
a=ax[0];a.axis('off');a.set_xlim(0,10);a.set_ylim(0,6)
a.scatter([1],[3],s=900,c='#cfe2f3',edgecolor='k');a.text(1,3,'s,a',ha='center',va='center')
a.scatter([4.5],[3],s=900,c='#d9ead3',edgecolor='k');a.text(4.5,3,"s'",ha='center',va='center')
a.annotate('',(4.1,3),(1.4,3),arrowprops=dict(arrowstyle='->'));a.text(2.75,3.2,'r',ha='center')
for y,lab,c in [(5,'a′₁ : Q=0,9  ← max (Q-learning)','tab:red'),(3,'a′₂ : Q=0,4  ← action réellement jouée (SARSA)','tab:blue'),(1,'a′₃ : Q=0,1','grey')]:
    a.annotate('',(7,y),(4.9,3),arrowprops=dict(arrowstyle='->',color=c));a.text(7.1,y,lab,va='center',color=c,fontsize=9)
a.set_title("Q-learning vise le meilleur Q(s',·) ; SARSA vise Q(s',a′) de l'action choisie")
def entrainer(methode,graine):
    env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));g=np.random.default_rng(graine);res=[]
    pol=lambda s,e:int(g.integers(4)) if g.random()<e else int(np.argmax(Q[s]))
    for ep in range(3000):
        s,_=env.reset(seed=graine+ep);e=max(.05,1-ep/2000);act=pol(s,e);tot=0
        for _ in range(100):
            s2,r,te,tr,_=env.step(act);a2=pol(s2,e);tot+=r
            suite=Q[s2].max() if methode=='Q' else Q[s2,a2]
            Q[s,act]+=.3*(r+(0 if te else .95*suite)-Q[s,act]);s,act=s2,a2
            if te or tr:break
        res.append(tot)
    env.close();return np.array(res)
for m,c,nom in [('Q','tab:red','Q-learning'),('S','tab:blue','SARSA')]:
    R=np.array([entrainer(m,s) for s in range(5)]).mean(0)
    ax[1].plot(np.convolve(R,np.ones(100)/100,'valid'),c=c,label=nom)
ax[1].set_xlabel('épisode');ax[1].set_ylabel('taux de réussite (moyenne glissante 100, 5 graines)')
ax[1].set_title('Pendant l’apprentissage (avec exploration ε)');ax[1].legend();plt.tight_layout();plt.show()
```

**Comment lire le schéma :** à gauche, depuis l'état $s'$ trois actions sont possibles. Le Q-learning met à jour $Q(s,a)$ avec la **meilleure** valeur (0,9, en rouge), même si l'agent va en réalité explorer ; SARSA utilise la valeur de l'action **réellement jouée** (0,4, en bleu) — il apprend donc la valeur de son propre comportement, exploration incluse (valeurs illustratives). À droite, le taux de réussite des épisodes au cours de l'entraînement sur FrozenLake, moyenné sur 5 graines : il augmente à mesure que $\varepsilon$ diminue. Simulation Gymnasium.

### Contextes d'utilisation
- **Cas d'usage 1 (logistique) :** apprentissage en ligne d'une politique de routage d'entrepôt ou de robot de préparation où une erreur d'exploration est coûteuse (collision, retard) : SARSA tend à apprendre des itinéraires plus sûrs que le Q-learning.
- **Cas d'usage 2 (finance/gestion) :** politique de relance de clients en retard de paiement apprise sur le système réel, où l'on veut évaluer la stratégie effectivement appliquée (avec sa part d'essais) plutôt qu'une stratégie idéale.
- **Données et prérequis :** environnement interactif à états et actions discrets ; l'apprentissage doit se faire **avec la politique utilisée** (pas sur des données collectées par une autre politique) ; décroissance de $\varepsilon$.
- **Peu approprié :** apprendre à partir d'un historique existant collecté par une autre stratégie (ex. journal de décisions passées de gestionnaires) : SARSA est sur-politique et exige des données générées par sa propre politique → préférer le Q-learning ou des méthodes hors-ligne.
- **Métriques :** récompense cumulée et taux de réussite **pendant** et **après** l'apprentissage, risque (nombre de chutes/échecs), variation entre graines.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

- **Entrées :** Environnement **FrozenLake-v1** de Gymnasium, version déterministe (`is_slippery=False`) : grille 4 × 4 (états 0 à 15, départ en 0, but en 15, trous en 5, 7, 11 et 12), 4 actions (0 = gauche, 1 = bas, 2 = droite, 3 = haut), récompense 1 uniquement en atteignant le but. Ce sont des trajectoires **simulées par interaction**, pas un dataset réel.
- **Hyperparamètres :** identiques à l'exemple 27 (4 000 épisodes, $\alpha=0,3$, $\gamma=0,95$, $\varepsilon$ décroissant de 1 à 0,05), ce qui rend la comparaison équitable.
- **Sortie obtenue :** politique `[1 0 3 0 1 0 1 0 2 1 1 0 0 2 2 0]`.
- **Conclusion possible :** depuis l'état 0, la politique suit **le même chemin que le Q-learning** : 0 → 4 → 8 → 9 → 13 → 14 → 15 (bas, bas, droite, bas, droite, droite), chemin optimal de 6 pas. Les seules différences portent sur les états 1 et 2 (actions 0 et 3 au lieu de 2 et 1), qui ne sont **pas sur le chemin optimal** et ont été peu visités en fin d'apprentissage ; leur valeur importe peu pour la politique suivie depuis le départ.
- **Ce qu'on ne peut pas conclure :** dans un environnement déterministe et sans risque après convergence, la différence de prudence entre SARSA et Q-learning ne peut pas apparaître ; aucune récompense cumulée ni taux de réussite n'est mesuré ; une seule graine.

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Deep Q-Network** (réseau Q profond, *DQN*) — famille : **apprentissage par renforcement profond, sans modèle, hors-politique**. Objectif : étendre le Q-learning à des espaces d'états **grands ou continus** en remplaçant la table $Q$ par un **réseau de neurones** $Q(s,a;\theta)$.

**Intuition et étapes :** l'agent interagit et stocke ses transitions $(s,a,r,s',\text{fin})$ dans une **mémoire de rejeu** → il tire au hasard des mini-lots de transitions (casse les corrélations temporelles) → pour chaque transition, calcul d'une **cible de Bellman** avec un **réseau cible** gelé $\theta^-$ → mise à jour des poids $\theta$ par descente de gradient sur l'erreur quadratique → copie périodique $\theta\to\theta^-$.

$$y_i=r_i+\gamma\,(1-d_i)\max_{a'}Q(s'_i,a';\theta^-),\qquad L(\theta)=\frac1B\sum_{i=1}^{B}\big(y_i-Q(s_i,a_i;\theta)\big)^2$$

$y_i$ : cible de Bellman ; $r_i$ : récompense ; $\gamma$ : facteur d'actualisation ; $d_i\in\{0,1\}$ : indicateur d'état terminal (pas de valeur future si 1) ; $\theta$ : poids du réseau en ligne (**paramètres appris**) ; $\theta^-$ : poids du réseau cible ; $B$ : taille du mini-lot. **Hyperparamètres :** architecture du réseau, $\gamma$, taux d'apprentissage, taille de la mémoire, $B$, fréquence de mise à jour du réseau cible, schéma de $\varepsilon$.

**Avantages :** (1) gère des états continus ou de grande dimension (images, séries de prix) en généralisant entre états similaires ; (2) mémoire de rejeu et réseau cible stabilisent l'apprentissage.
**Limites :** (1) apprentissage instable, gourmand en données et en calcul, très sensible aux hyperparamètres ; (2) limité aux actions **discrètes** et peu interprétable (surestimation des valeurs → variantes Double DQN).

### Schéma explicatif

```python
# Schéma 29 — Architecture DQN : mémoire de rejeu, réseau en ligne, réseau cible (schéma conceptuel)
from matplotlib.patches import FancyBboxPatch
fig,ax=plt.subplots(figsize=(12,5));ax.axis('off');ax.set_xlim(0,10.5);ax.set_ylim(0,7)
boites={'env':(0.3,4.5,'Environnement\n(Gymnasium)','#d9ead3'),'mem':(3.4,4.5,'Mémoire de rejeu\n(s, a, r, s′, fin)','#fff2cc'),
        'ligne':(7,4.5,'Réseau en ligne\nQ(s,a;θ)','#cfe2f3'),'cible':(7,0.8,'Réseau cible\nQ(s′,·;θ⁻) gelé','#f4cccc'),
        'perte':(3.4,0.8,'Cible y = r + γ(1−fin)·max Q(s′,·;θ⁻)\nPerte L = (y − Q(s,a;θ))²','#ead1dc')}
for x,y,t,c in boites.values():
    ax.add_patch(FancyBboxPatch((x,y),2.8,1.5,boxstyle='round,pad=.1',fc=c,ec='k'));ax.text(x+1.4,y+.75,t,ha='center',va='center',fontsize=9)
fl=[((3.1,5.25),(3.4,5.25),'transitions'),((6.2,5.25),(7,5.25),'mini-lot aléatoire'),((8.4,4.5),(8.4,2.3),'copie θ→θ⁻\ntous les C pas'),
    ((7,1.55),(6.2,1.55),'max Q(s′,·)'),((6.2,2.0),(7,4.6),'gradient sur θ'),((8.4,6.25),(1.7,6.25),'action ε-greedy')]
for (a,b,t) in fl:
    ax.annotate('',b,a,arrowprops=dict(arrowstyle='->',lw=1.6));ax.text((a[0]+b[0])/2,(a[1]+b[1])/2+.15,t,ha='center',fontsize=8)
ax.set_title('DQN : le Q-learning dont la table Q est remplacée par un réseau de neurones stabilisé');plt.show()
```

**Comment lire le schéma :** en haut, le réseau en ligne choisit les actions ($\varepsilon$-greedy) ; chaque transition observée est stockée dans la mémoire de rejeu, d'où l'on tire des mini-lots au hasard. En bas, la cible est calculée avec le **réseau cible**, une copie figée du réseau en ligne mise à jour seulement tous les $C$ pas : sans lui, la cible bougerait à chaque mise à jour et l'apprentissage divergerait. La perte entre cible et prédiction sert à corriger les poids $\theta$. Schéma conceptuel original.
